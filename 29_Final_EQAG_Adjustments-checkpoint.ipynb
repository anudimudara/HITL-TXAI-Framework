```python
# ============================================================
# SECTION 13 — EQAG CRITERION 2: CORE APPLICABILITY
# ============================================================

valid_outcomes = {
    "TP",
    "TN",
    "FP",
    "FN"
}

eqag_df["Outcome_Applicable"] = (
    eqag_df["Outcome"]
    .isin(valid_outcomes)
)

eqag_df["CTS_Applicable"] = (
    eqag_df["CTS_star"].notna()
)

eqag_df["HTS_Applicable"] = (
    eqag_df["HTS"].notna()
)

eqag_df["EAS_Applicable"] = (
    eqag_df["EAS"].notna()
)

eqag_df["Applicable"] = (
    eqag_df["Outcome_Applicable"] &
    eqag_df["CTS_Applicable"] &
    eqag_df["HTS_Applicable"] &
    eqag_df["EAS_Applicable"]
)

print("Applicability:")
print(eqag_df["Applicable"].value_counts())

print(
    "\nApplicable records:",
    eqag_df["Applicable"].sum(),
    "/",
    len(eqag_df)
)
```

```python
# ============================================================
# SECTION 14 — LOCALISATION APPLICABILITY
# ============================================================

eqag_df["L_localisation"] = pd.to_numeric(
    df["L_localisation"],
    errors="coerce"
)

eqag_df["Reference_Positive"] = (
    eqag_df["Outcome"]
    .isin(["TP", "FN"])
)
```

```python
def localisation_status(row):

    # TN and FP are reference-negative
    if not row["Reference_Positive"]:
        return "Not applicable"

    # Reference-positive and localisation value exists
    if pd.notna(row["L_localisation"]):
        return "Applicable"

    # Reference-positive but no usable localisation evidence
    return "Conditional - reference region unavailable"


eqag_df["Localisation_Applicability"] = (
    eqag_df.apply(
        localisation_status,
        axis=1
    )
)

print(
    eqag_df[
        "Localisation_Applicability"
    ].value_counts()
)
```

```python
# ============================================================
# SECTION 15 — LOCALISATION INFORMATION CODE
# ============================================================

def localisation_information_code(status):

    if status == "Applicable":
        return "I01_LOCALISATION_APPLICABLE"

    elif status == "Not applicable":
        return "I02_LOCALISATION_NOT_APPLICABLE"

    else:
        return "I03_LOCALISATION_CONDITIONAL"


eqag_df["Localisation_Info_Code"] = (
    eqag_df[
        "Localisation_Applicability"
    ].apply(
        localisation_information_code
    )
)

display(
    eqag_df[
        [
            "Case_ID",
            "Method",
            "Outcome",
            "Localisation_Applicability",
            "Localisation_Info_Code"
        ]
    ].head(12)
)
```

```python
# ============================================================
# SECTION 16 — EQAG CRITERION 3: NORMALISATION
# ============================================================

def valid_unit_interval(series):

    numeric = pd.to_numeric(
        series,
        errors="coerce"
    )

    return numeric.between(
        0,
        1,
        inclusive="both"
    )


eqag_df["CTS_Normalised"] = (
    valid_unit_interval(
        eqag_df["CTS_star"]
    )
)

eqag_df["HTS_Normalised"] = (
    valid_unit_interval(
        eqag_df["HTS"]
    )
)

eqag_df["EAS_Normalised"] = (
    valid_unit_interval(
        eqag_df["EAS"]
    )
)

eqag_df["UATAI_Normalised"] = (
    valid_unit_interval(
        eqag_df["UA_TAI"]
    )
)

eqag_df["Normalised"] = (
    eqag_df["CTS_Normalised"] &
    eqag_df["HTS_Normalised"] &
    eqag_df["EAS_Normalised"] &
    eqag_df["UATAI_Normalised"]
)

print("Normalisation results:")
print(eqag_df["Normalised"].value_counts())
```

```python
# ============================================================
# SECTION 17 — RANGE SUMMARY
# ============================================================

for col in [
    "CTS_star",
    "HTS",
    "EAS",
    "UA_TAI"
]:

    print(
        f"{col}: "
        f"min={eqag_df[col].min():.6f}, "
        f"max={eqag_df[col].max():.6f}"
    )
```

```python
# ============================================================
# SECTION 18 — CORRECT EAS DISCOUNT VERIFICATION
# ============================================================

KAPPA = 0.5
FORMULA_TOLERANCE = 1e-6

eqag_df["EAS_Discount_Saved"] = pd.to_numeric(
    df["EAS_discount_factor"],
    errors="coerce"
)

eqag_df["EAS_Discount_Recomputed"] = np.exp(
    -KAPPA * (1 - eqag_df["EAS"])
)

eqag_df["EAS_Discount_Difference"] = np.abs(
    eqag_df["EAS_Discount_Saved"]
    -
    eqag_df["EAS_Discount_Recomputed"]
)

eqag_df["EAS_Discount_Consistent"] = (
    eqag_df["EAS_Discount_Difference"]
    <= FORMULA_TOLERANCE
)

print("EAS discount consistency:")
print(
    eqag_df["EAS_Discount_Consistent"]
    .value_counts()
)

print(
    "\nMaximum difference:",
    eqag_df["EAS_Discount_Difference"].max()
)
```

```python
# ============================================================
# SECTION 19 — RAW ALIGNMENT VERIFICATION
# ============================================================

eqag_df["Raw_Alignment_Recomputed"] = (
    1
    -
    np.abs(
        eqag_df["HTS"]
        -
        eqag_df["CTS_star"]
    )
)

eqag_df["Raw_Alignment_Difference"] = np.abs(
    eqag_df["Raw_Alignment"]
    -
    eqag_df["Raw_Alignment_Recomputed"]
)

eqag_df["Raw_Alignment_Consistent"] = (
    eqag_df["Raw_Alignment_Difference"]
    <= FORMULA_TOLERANCE
)

print("Raw alignment consistency:")
print(
    eqag_df["Raw_Alignment_Consistent"]
    .value_counts()
)

print(
    "\nMaximum difference:",
    eqag_df["Raw_Alignment_Difference"].max()
)
```

```python
# ============================================================
# SECTION 20 — VERIFY FINAL UA-TAI
# ============================================================

eqag_df["UA_TAI_Recomputed"] = (
    eqag_df["Raw_Alignment_Recomputed"]
    *
    eqag_df["EAS_Discount_Recomputed"]
)

eqag_df["UA_TAI_Formula_Difference"] = np.abs(
    eqag_df["UA_TAI"]
    -
    eqag_df["UA_TAI_Recomputed"]
)

eqag_df["UA_TAI_Formula_Consistent"] = (
    eqag_df["UA_TAI_Formula_Difference"]
    <= FORMULA_TOLERANCE
)

print("UA-TAI formula consistency:")
print(
    eqag_df["UA_TAI_Formula_Consistent"]
    .value_counts()
)

print(
    "\nMaximum UA-TAI difference:",
    eqag_df["UA_TAI_Formula_Difference"].max()
)
```

```python
# ============================================================
# SECTION 21 — KAPPA 0.5 SAVED RESULT VERIFICATION
# ============================================================

eqag_df["UA_TAI_Kappa_05_Saved"] = pd.to_numeric(
    df["UA_TAI_kappa_0_5_unit"],
    errors="coerce"
)

eqag_df["Kappa_05_Difference"] = np.abs(
    eqag_df["UA_TAI"]
    -
    eqag_df["UA_TAI_Kappa_05_Saved"]
)

eqag_df["Kappa_05_Consistent"] = (
    eqag_df["Kappa_05_Difference"]
    <= FORMULA_TOLERANCE
)

print("Kappa 0.5 consistency:")
print(
    eqag_df["Kappa_05_Consistent"]
    .value_counts()
)

print(
    "\nMaximum difference:",
    eqag_df["Kappa_05_Difference"].max()
)
```

```python
# ============================================================
# SECTION 24 — STANDARDISE TRUST QUADRANT
# ============================================================

def standardise_quadrant(value):

    text = str(value).strip().lower()

    if "over" in text:
        return "Over-trust"

    elif "under" in text:
        return "Under-trust"

    elif "high" in text:
        return "Aligned-high"

    elif "cautious" in text or "low" in text:
        return "Aligned-cautious"

    else:
        return str(value)


eqag_df["Trust_Quadrant_Standardised"] = (
    eqag_df["Trust_Quadrant"]
    .apply(standardise_quadrant)
)

print(
    eqag_df["Trust_Quadrant_Standardised"]
    .value_counts()
)
```

```python
# ============================================================
# SECTION 25 — FINAL OUTCOME-AWARE GOVERNANCE ACTION
# ============================================================

def final_governance_action(row):

    outcome = row["Outcome"]

    quadrant = row[
        "Trust_Quadrant_Standardised"
    ]

    # --------------------------------------------------------
    # FALSE NEGATIVE
    # --------------------------------------------------------

    if outcome == "FN":

        if quadrant == "Over-trust":
            return (
                "Mandatory high-priority false-negative review; "
                "investigate potential over-trust and whether the "
                "explanation may have encouraged excessive confidence "
                "in an incorrect prediction."
            )

        elif quadrant == "Under-trust":
            return (
                "Mandatory high-priority false-negative review; "
                "additionally review explanation clarity and "
                "specialist trust calibration."
            )

        elif quadrant == "Aligned-high":
            return (
                "Mandatory high-priority false-negative review despite "
                "relatively high trust alignment; alignment must not "
                "override prediction error."
            )

        else:
            return (
                "Mandatory high-priority false-negative review; "
                "maintain heightened caution because the model missed "
                "a reference-positive case."
            )

    # --------------------------------------------------------
    # FALSE POSITIVE
    # --------------------------------------------------------

    if outcome == "FP":

        if quadrant == "Over-trust":
            return (
                "Mandatory false-positive review; investigate potential "
                "over-trust and whether the explanation may have "
                "encouraged confidence in an incorrect positive prediction."
            )

        elif quadrant == "Under-trust":
            return (
                "Mandatory false-positive review; additionally review "
                "explanation clarity and specialist trust."
            )

        elif quadrant == "Aligned-high":
            return (
                "Mandatory false-positive review despite relatively high "
                "trust alignment; alignment does not establish prediction "
                "correctness."
            )

        else:
            return (
                "Mandatory false-positive review; maintain error-case "
                "caution and investigate the source of the incorrect "
                "positive prediction."
            )

    # --------------------------------------------------------
    # CORRECT PREDICTIONS
    # --------------------------------------------------------

    if quadrant == "Over-trust":
        return (
            "Trust review required; specialist trust is relatively high "
            "compared with computational trustworthiness."
        )

    elif quadrant == "Under-trust":
        return (
            "Review explanation clarity, communication and "
            "specialist confidence."
        )

    elif quadrant == "Aligned-high":
        return (
            "Routine review; computational and human trust are "
            "relatively well aligned."
        )

    else:
        return (
            "Standard caution; both computational and human trust "
            "are relatively low."
        )


eqag_df["Final_Governance_Action"] = (
    eqag_df.apply(
        final_governance_action,
        axis=1
    )
)
```

```python
# ============================================================
# SECTION 26 — FINAL OUTCOME-AWARE INTERPRETATION
# ============================================================

def final_outcome_interpretation(row):

    outcome = row["Outcome"]

    if outcome == "FN":
        return (
            "Incorrect prediction: false negative. "
            "The model predicted normal while the reference label "
            "indicated lung opacity suspicious for pneumonia. "
            "Mandatory outcome-aware review applies regardless of UA-TAI."
        )

    elif outcome == "FP":
        return (
            "Incorrect prediction: false positive. "
            "The model predicted lung opacity while the reference label "
            "was normal. Mandatory outcome-aware review applies "
            "regardless of UA-TAI."
        )

    elif outcome == "TP":
        return (
            "Correct reference-positive prediction. "
            "UA-TAI describes trust alignment only and does not establish "
            "clinical safety or deployment readiness."
        )

    elif outcome == "TN":
        return (
            "Correct reference-negative prediction. "
            "UA-TAI describes trust alignment only and does not establish "
            "clinical safety or deployment readiness."
        )

    return "Outcome interpretation unavailable."


eqag_df["Final_Outcome_Interpretation"] = (
    eqag_df.apply(
        final_outcome_interpretation,
        axis=1
    )
)
```

```python
# ============================================================
# SECTION 32 — SOURCE WORKBOOK HASH
# ============================================================

import hashlib

def calculate_sha256(file_path):

    sha256 = hashlib.sha256()

    with open(file_path, "rb") as f:

        for block in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            sha256.update(block)

    return sha256.hexdigest()


SOURCE_SHA256 = calculate_sha256(
    FINAL_RESULTS_FILE
)

print("Source workbook SHA-256:")
print(SOURCE_SHA256)

eqag_df["Source_Workbook_SHA256"] = (
    SOURCE_SHA256
)
```

```python
# ============================================================
# SECTION 36 — RECORD-LEVEL EQAG REASON CODES
# ============================================================

def generate_eqag_reason_codes(row):

    codes = []

    if not row["Complete"]:
        codes.append(
            "N01_MISSING_CORE_EVIDENCE"
        )

    if not row["Applicable"]:
        codes.append(
            "N02_CORE_EVIDENCE_NOT_APPLICABLE"
        )

    if not row["Normalised"]:
        codes.append(
            "N03_INVALID_NORMALISED_SCALE"
        )

    if not row["Provenance_Complete"]:
        codes.append(
            "N04_PROVENANCE_INCOMPLETE"
        )

    if not row["EAS_Discount_Consistent"]:
        codes.append(
            "N05_EAS_DISCOUNT_INCONSISTENT"
        )

    if not row["Raw_Alignment_Consistent"]:
        codes.append(
            "N06_RAW_ALIGNMENT_INCONSISTENT"
        )

    if not row["UA_TAI_Formula_Consistent"]:
        codes.append(
            "N07_UATAI_FORMULA_INCONSISTENT"
        )

    if not row["Kappa_05_Consistent"]:
        codes.append(
            "N08_KAPPA_PROFILE_INCONSISTENT"
        )

    if not row["Governance_Output_Complete"]:
        codes.append(
            "N09_GOVERNANCE_OUTPUT_INCOMPLETE"
        )

    if not row["Outcome_Override_Consistent"]:
        codes.append(
            "N10_OUTCOME_OVERRIDE_FAILURE"
        )

    if len(codes) == 0:
        codes.append(
            "Q00_RECORD_QUALIFIED"
        )

    return "; ".join(codes)


eqag_df["EQAG_Record_Reason_Codes"] = (
    eqag_df.apply(
        generate_eqag_reason_codes,
        axis=1
    )
)

print(
    eqag_df[
        "EQAG_Record_Reason_Codes"
    ].value_counts()
)
```

```python
# ============================================================
# SECTION 48 — FORMAT FINAL WORKBOOK
# ============================================================

from openpyxl import load_workbook
from openpyxl.styles import Font, Alignment
from openpyxl.utils import get_column_letter

wb = load_workbook(
    FINAL_EQAG_FILE
)

for ws in wb.worksheets:

    # Bold header
    for cell in ws[1]:
        cell.font = Font(
            bold=True
        )

    # Freeze header row
    ws.freeze_panes = "A2"

    # Add filter
    if ws.max_row > 1:
        ws.auto_filter.ref = (
            ws.dimensions
        )

    # Adjust column widths
    for column_cells in ws.columns:

        max_length = 0

        column_index = (
            column_cells[0].column
        )

        for cell in column_cells:

            if cell.value is not None:

                max_length = max(
                    max_length,
                    len(str(cell.value))
                )

        width = min(
            max_length + 2,
            60
        )

        ws.column_dimensions[
            get_column_letter(
                column_index
            )
        ].width = width

        for cell in column_cells:
            cell.alignment = Alignment(
                vertical="top",
                wrap_text=True
            )

wb.save(
    FINAL_EQAG_FILE
)

print(
    "Workbook formatting completed."
)
```
