```python
with open(temperature_path, "r", encoding="utf-8") as file:
    temperature_data = json.load(file)

print("Temperature JSON:")
print(temperature_data)

possible_temperature_keys = [
    "temperature",
    "optimal_temperature",
    "best_temperature",
    "T"
]

TEMPERATURE = None

for key in possible_temperature_keys:
    if key in temperature_data:
        TEMPERATURE = float(temperature_data[key])
        break

if TEMPERATURE is None:
    raise ValueError("Temperature value not found in JSON file.")

MODEL_DECISION_THRESHOLD = 0.35

print("Loaded temperature:", TEMPERATURE)
print("Model decision threshold:", MODEL_DECISION_THRESHOLD)
```

```python
def extract_state_dict(checkpoint):
    if isinstance(checkpoint, dict):
        for key in ["model_state_dict", "state_dict", "model"]:
            if key in checkpoint:
                return checkpoint[key]
    return checkpoint


def clean_state_dict_keys(state_dict):
    cleaned = {}

    for key, value in state_dict.items():
        new_key = key

        if new_key.startswith("module."):
            new_key = new_key.replace("module.", "", 1)

        cleaned[new_key] = value

    return cleaned


checkpoint = torch.load(resnet_checkpoint_path, map_location=device)
state_dict = extract_state_dict(checkpoint)
state_dict = clean_state_dict_keys(state_dict)

output_features = state_dict["fc.weight"].shape[0]

print("Detected ResNet50 output features:", output_features)

model = models.resnet50(weights=None)
model.fc = nn.Linear(model.fc.in_features, output_features)

model.load_state_dict(state_dict, strict=True)
model = model.to(device)
model.eval()

print("Frozen ResNet50 loaded successfully.")
```

```python
IMAGE_SIZE = 224

model_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("Preprocessing transform defined.")
```

```python
print("Searching for DICOM files...")

dicom_files = list(BASE_DIR.rglob("*.dcm"))

print("Number of DICOM files found:", len(dicom_files))

dicom_index = {}

for path in dicom_files:
    dicom_index[path.stem] = path

print("DICOM index created.")
```

```python
def read_dicom_as_rgb(dicom_path):
    dicom = pydicom.dcmread(str(dicom_path))
    image = dicom.pixel_array.astype(np.float32)

    if getattr(dicom, "PhotometricInterpretation", "") == "MONOCHROME1":
        image = np.max(image) - image

    image = image - np.min(image)

    if np.max(image) > 0:
        image = image / np.max(image)

    image = (image * 255).astype(np.uint8)

    rgb_image = np.stack([image, image, image], axis=-1)

    return rgb_image


def load_rgb_image_from_case(case_row):
    patient_id = str(case_row["patientId"])

    if patient_id not in dicom_index:
        raise FileNotFoundError(f"DICOM image not found for patientId: {patient_id}")

    dicom_path = dicom_index[patient_id]

    rgb_image = read_dicom_as_rgb(dicom_path)

    pil_image = Image.fromarray(rgb_image).convert("RGB")
    pil_image = pil_image.resize((IMAGE_SIZE, IMAGE_SIZE))

    rgb_resized = np.array(pil_image)

    return dicom_path, rgb_resized
```

```python
def predict_calibrated_probabilities_from_rgb_batch(rgb_images):
    """
    Input:
    - list or array of RGB images, each shape 224 x 224 x 3, uint8

    Output:
    - calibrated Lung Opacity probabilities
    """

    model.eval()

    tensors = []

    for image in rgb_images:
        image = np.asarray(image).astype(np.uint8)
        pil_image = Image.fromarray(image).convert("RGB")
        tensor = model_transform(pil_image)
        tensors.append(tensor)

    batch_tensor = torch.stack(tensors).to(device)

    with torch.no_grad():
        logits = model(batch_tensor)
        scaled_logits = logits / TEMPERATURE

        if scaled_logits.shape[1] == 1:
            probs = torch.sigmoid(scaled_logits)[:, 0]

        elif scaled_logits.shape[1] == 2:
            probs = torch.softmax(scaled_logits, dim=1)[:, 1]

        else:
            raise ValueError(f"Unexpected model output shape: {scaled_logits.shape}")

    return probs.detach().cpu().numpy()
```

```python
def create_blurred_baseline(rgb_image, kernel_size=51):
    """
    Creates a blurred image baseline for deletion/insertion.
    The blur preserves broad intensity structure while removing local detail.
    """

    if kernel_size % 2 == 0:
        kernel_size += 1

    blurred = cv2.GaussianBlur(
        rgb_image,
        (kernel_size, kernel_size),
        0
    )

    return blurred.astype(np.uint8)
```

```python
def create_top_fraction_mask(heatmap, fraction):
    """
    Creates a binary mask for the top fraction of attribution values.

    fraction = 0.0 means no pixels selected.
    fraction = 1.0 means all pixels selected.
    """

    heatmap = np.asarray(heatmap).astype(np.float32)

    if fraction <= 0:
        return np.zeros_like(heatmap, dtype=bool)

    if fraction >= 1:
        return np.ones_like(heatmap, dtype=bool)

    flat = heatmap.flatten()
    n_pixels = flat.shape[0]

    k = int(round(fraction * n_pixels))

    if k <= 0:
        return np.zeros_like(heatmap, dtype=bool)

    sorted_indices = np.argsort(flat)[::-1]
    selected_indices = sorted_indices[:k]

    mask_flat = np.zeros(n_pixels, dtype=bool)
    mask_flat[selected_indices] = True

    return mask_flat.reshape(heatmap.shape)
```

```python
def apply_deletion(rgb_image, baseline_image, attribution_mask):
    """
    Deletion:
    Start with original image.
    Replace highly attributed pixels with baseline.
    """

    perturbed = rgb_image.copy()
    perturbed[attribution_mask] = baseline_image[attribution_mask]

    return perturbed.astype(np.uint8)


def apply_insertion(rgb_image, baseline_image, attribution_mask):
    """
    Insertion:
    Start with baseline image.
    Restore highly attributed pixels from original image.
    """

    perturbed = baseline_image.copy()
    perturbed[attribution_mask] = rgb_image[attribution_mask]

    return perturbed.astype(np.uint8)
```

```python
PERTURBATION_FRACTIONS = np.linspace(0.0, 1.0, 11)

print("Perturbation fractions:")
print(PERTURBATION_FRACTIONS)
```

```python
def evaluate_faithfulness_for_row(manifest_row, xai_cohort, perturbation_fractions):
    """
    Calculates deletion and insertion curves for one case-method explanation.
    """

    case_id = manifest_row["xai_case_id"]
    method = manifest_row["method"]

    case_row = xai_cohort[
        xai_cohort["xai_case_id"] == case_id
    ].iloc[0]

    _, rgb_image = load_rgb_image_from_case(case_row)

    heatmap = np.load(manifest_row["normalised_map_path"]).astype(np.float32)

    if heatmap.shape != (IMAGE_SIZE, IMAGE_SIZE):
        heatmap = cv2.resize(
            heatmap,
            (IMAGE_SIZE, IMAGE_SIZE),
            interpolation=cv2.INTER_LINEAR
        )

    baseline_image = create_blurred_baseline(rgb_image)

    deletion_images = []
    insertion_images = []

    for fraction in perturbation_fractions:
        top_mask = create_top_fraction_mask(
            heatmap=heatmap,
            fraction=fraction
        )

        deletion_image = apply_deletion(
            rgb_image=rgb_image,
            baseline_image=baseline_image,
            attribution_mask=top_mask
        )

        insertion_image = apply_insertion(
            rgb_image=rgb_image,
            baseline_image=baseline_image,
            attribution_mask=top_mask
        )

        deletion_images.append(deletion_image)
        insertion_images.append(insertion_image)

    deletion_probs = predict_calibrated_probabilities_from_rgb_batch(deletion_images)
    insertion_probs = predict_calibrated_probabilities_from_rgb_batch(insertion_images)

    deletion_auc = auc(perturbation_fractions, deletion_probs)
    insertion_auc = auc(perturbation_fractions, insertion_probs)

    original_probability = float(deletion_probs[0])
    baseline_probability = float(insertion_probs[0])

    aopc = float(
        np.mean(original_probability - deletion_probs[1:])
    )

    result = {
        "xai_case_id": case_id,
        "patientId": manifest_row["patientId"],
        "method": method,
        "outcome": manifest_row["outcome"],
        "reference_class": manifest_row["reference_class"],
        "predicted_class": manifest_row["predicted_class"],
        "saved_calibrated_probability": manifest_row["calibrated_probability"],
        "reproduced_original_probability": original_probability,
        "baseline_probability": baseline_probability,
        "deletion_auc": float(deletion_auc),
        "insertion_auc": float(insertion_auc),
        "aopc": float(aopc),
        "num_perturbation_steps": int(len(perturbation_fractions)),
        "normalised_map_path": manifest_row["normalised_map_path"]
    }

    curve_df = pd.DataFrame({
        "xai_case_id": case_id,
        "method": method,
        "fraction_removed_or_inserted": perturbation_fractions,
        "deletion_probability": deletion_probs,
        "insertion_probability": insertion_probs
    })

    return result, curve_df
```

```python
faithfulness_pilot_manifest = (
    full_xai_manifest
    .groupby(["method", "outcome"], group_keys=False)
    .sample(n=1, random_state=RANDOM_SEED)
    .reset_index(drop=True)
)

print("Faithfulness pilot rows:", len(faithfulness_pilot_manifest))

display(
    faithfulness_pilot_manifest[
        [
            "xai_case_id",
            "method",
            "outcome",
            "calibrated_probability",
            "normalised_map_path"
        ]
    ]
)
```

```python
pilot_faithfulness_rows = []
pilot_curve_rows = []

for _, manifest_row in tqdm(
    faithfulness_pilot_manifest.iterrows(),
    total=len(faithfulness_pilot_manifest),
    desc="faithfulness pilot"
):
    result, curve_df = evaluate_faithfulness_for_row(
        manifest_row=manifest_row,
        xai_cohort=xai_cohort,
        perturbation_fractions=PERTURBATION_FRACTIONS
    )

    pilot_faithfulness_rows.append(result)
    pilot_curve_rows.append(curve_df)

pilot_faithfulness_results = pd.DataFrame(pilot_faithfulness_rows)
pilot_faithfulness_curves = pd.concat(pilot_curve_rows, ignore_index=True)

pilot_results_path = faithfulness_results_folder / "faithfulness_pilot_results.csv"
pilot_curves_path = faithfulness_curves_folder / "faithfulness_pilot_curves.csv"

pilot_faithfulness_results.to_csv(pilot_results_path, index=False)
pilot_faithfulness_curves.to_csv(pilot_curves_path, index=False)

print("Pilot faithfulness results saved:")
print(pilot_results_path)

print("Pilot faithfulness curves saved:")
print(pilot_curves_path)

display(pilot_faithfulness_results)
```

```python
faithfulness_results_path = faithfulness_results_folder / "xai_faithfulness_results.csv"
faithfulness_curves_path = faithfulness_curves_folder / "xai_faithfulness_curves.csv"

faithfulness_rows = []
curve_rows = []

for _, manifest_row in tqdm(
    full_xai_manifest.iterrows(),
    total=len(full_xai_manifest),
    desc="faithfulness full"
):
    result, curve_df = evaluate_faithfulness_for_row(
        manifest_row=manifest_row,
        xai_cohort=xai_cohort,
        perturbation_fractions=PERTURBATION_FRACTIONS
    )

    faithfulness_rows.append(result)
    curve_rows.append(curve_df)

    # Save progress every 20 rows
    if len(faithfulness_rows) % 20 == 0:
        pd.DataFrame(faithfulness_rows).to_csv(
            faithfulness_results_path,
            index=False
        )

        pd.concat(curve_rows, ignore_index=True).to_csv(
            faithfulness_curves_path,
            index=False
        )

        print(f"Saved progress: {len(faithfulness_rows)}/480")

    gc.collect()

faithfulness_results = pd.DataFrame(faithfulness_rows)
faithfulness_curves = pd.concat(curve_rows, ignore_index=True)

faithfulness_results.to_csv(faithfulness_results_path, index=False)
faithfulness_curves.to_csv(faithfulness_curves_path, index=False)

print("Full faithfulness results saved:")
print(faithfulness_results_path)

print("Full faithfulness curves saved:")
print(faithfulness_curves_path)
```
