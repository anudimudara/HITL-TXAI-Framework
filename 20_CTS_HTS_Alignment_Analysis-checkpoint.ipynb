```python
import pandas as pd

hts_excel = pd.ExcelFile(hts_file)

print("Sheets inside HTS file:")

for sheet in hts_excel.sheet_names:
    print(sheet)
```

```python
case_method_hts = pd.read_excel(
    hts_file,
    sheet_name="Case_Method_HTS"
)

case_method_hts.head()
```

```python
cts_file = "CTS_Survey_Case_Method_Table.xlsx"

cts_survey_table = pd.read_excel(
    cts_file,
    sheet_name="CTS_32_Rows"
)

print("CTS rows:", len(cts_survey_table))
cts_survey_table.head()
```

```python
print("HTS case-method rows:", len(case_method_hts))
print("CTS survey rows:", len(cts_survey_table))

print("\nHTS columns:")
print(case_method_hts.columns.tolist())

print("\nCTS columns:")
print(cts_survey_table.columns.tolist())
```

```python
hts_standard = case_method_hts.copy()

# Rename mean_HTS to HTS_percentage if needed
if "mean_HTS" in hts_standard.columns:
    hts_standard = hts_standard.rename(columns={"mean_HTS": "HTS_percentage"})

# Keep only needed HTS columns
hts_standard = hts_standard[
    [
        "case_id",
        "method",
        "HTS_percentage",
        "median_HTS",
        "sd_HTS",
        "min_HTS",
        "max_HTS",
        "n"
    ]
].copy()

# Clean text columns
hts_standard["case_id"] = hts_standard["case_id"].astype(str).str.strip()
hts_standard["method"] = hts_standard["method"].astype(str).str.strip()

hts_standard.head()
```

```python
cts_standard = cts_survey_table.copy()

# Rename survey_outcome to outcome if needed
if "survey_outcome" in cts_standard.columns:
    cts_standard = cts_standard.rename(columns={"survey_outcome": "outcome"})

# Clean text columns
cts_standard["case_id"] = cts_standard["case_id"].astype(str).str.strip()
cts_standard["method"] = cts_standard["method"].astype(str).str.strip()

cts_standard.head()
```

```python
print("HTS case-method combinations:")
print(hts_standard[["case_id", "method"]].drop_duplicates().sort_values(["case_id", "method"]))

print("\nCTS case-method combinations:")
print(cts_standard[["case_id", "method"]].drop_duplicates().sort_values(["case_id", "method"]))
```

```python
trust_alignment = cts_standard.merge(
    hts_standard,
    on=["case_id", "method"],
    how="left"
)

print("Merged rows:", len(trust_alignment))

trust_alignment.head()
```

```python
print("Expected rows: 32")
print("Actual rows:", len(trust_alignment))

print("\nMissing values:")
print(trust_alignment.isnull().sum())

print("\nRows per case:")
print(trust_alignment.groupby("case_id").size())

print("\nRows per method:")
print(trust_alignment.groupby("method").size())
```

```python
trust_alignment["TrustGap"] = (
    trust_alignment["HTS_percentage"] - trust_alignment["CTS_percentage"]
)

trust_alignment["Absolute_TrustGap"] = trust_alignment["TrustGap"].abs()

trust_alignment["TAI"] = (
    100 - trust_alignment["Absolute_TrustGap"]
)

trust_alignment[
    [
        "case_id",
        "xai_case_id",
        "outcome",
        "method",
        "actual_xai_method",
        "CTS_percentage",
        "HTS_percentage",
        "TrustGap",
        "TAI"
    ]
]
```

```python
overall_alignment_summary = pd.DataFrame({
    "Metric": [
        "Mean CTS",
        "Mean HTS",
        "Mean TrustGap",
        "Median TrustGap",
        "Mean Absolute TrustGap",
        "Mean TAI",
        "Median TAI",
        "Minimum TAI",
        "Maximum TAI",
        "Number of case-method pairs"
    ],
    "Value": [
        trust_alignment["CTS_percentage"].mean(),
        trust_alignment["HTS_percentage"].mean(),
        trust_alignment["TrustGap"].mean(),
        trust_alignment["TrustGap"].median(),
        trust_alignment["Absolute_TrustGap"].mean(),
        trust_alignment["TAI"].mean(),
        trust_alignment["TAI"].median(),
        trust_alignment["TAI"].min(),
        trust_alignment["TAI"].max(),
        len(trust_alignment)
    ]
})

overall_alignment_summary
```

```python
alignment_by_method = trust_alignment.groupby(
    ["method", "actual_xai_method"]
).agg(
    mean_CTS=("CTS_percentage", "mean"),
    mean_HTS=("HTS_percentage", "mean"),
    mean_TrustGap=("TrustGap", "mean"),
    median_TrustGap=("TrustGap", "median"),
    mean_abs_TrustGap=("Absolute_TrustGap", "mean"),
    mean_TAI=("TAI", "mean"),
    median_TAI=("TAI", "median"),
    min_TAI=("TAI", "min"),
    max_TAI=("TAI", "max"),
    n=("TAI", "count")
).reset_index()

alignment_by_method = alignment_by_method.sort_values(
    "mean_TAI",
    ascending=False
).reset_index(drop=True)

alignment_by_method
```

```python
alignment_by_outcome = trust_alignment.groupby("outcome").agg(
    mean_CTS=("CTS_percentage", "mean"),
    mean_HTS=("HTS_percentage", "mean"),
    mean_TrustGap=("TrustGap", "mean"),
    median_TrustGap=("TrustGap", "median"),
    mean_abs_TrustGap=("Absolute_TrustGap", "mean"),
    mean_TAI=("TAI", "mean"),
    median_TAI=("TAI", "median"),
    min_TAI=("TAI", "min"),
    max_TAI=("TAI", "max"),
    n=("TAI", "count")
).reset_index()

alignment_by_outcome = alignment_by_outcome.sort_values(
    "mean_TAI",
    ascending=False
).reset_index(drop=True)

alignment_by_outcome
```

```python
def score_level(score):
    if score >= 70:
        return "High"
    elif score >= 40:
        return "Moderate"
    else:
        return "Low"

trust_alignment["CTS_level"] = trust_alignment["CTS_percentage"].apply(score_level)
trust_alignment["HTS_level"] = trust_alignment["HTS_percentage"].apply(score_level)

trust_alignment[
    [
        "case_id",
        "method",
        "CTS_percentage",
        "CTS_level",
        "HTS_percentage",
        "HTS_level"
    ]
].head()
```

```python
def trust_category(row):
    cts = row["CTS_level"]
    hts = row["HTS_level"]
    
    if cts == "High" and hts == "High":
        return "Aligned high trust"
    elif cts == "Low" and hts == "Low":
        return "Aligned cautious trust"
    elif cts == "Low" and hts == "High":
        return "Potential over-trust"
    elif cts == "High" and hts == "Low":
        return "Potential under-trust"
    else:
        return "Partial alignment"

trust_alignment["Trust_Category"] = trust_alignment.apply(trust_category, axis=1)

trust_alignment[
    [
        "case_id",
        "outcome",
        "method",
        "actual_xai_method",
        "CTS_level",
        "HTS_level",
        "Trust_Category"
    ]
]
```

```python
trust_category_summary = trust_alignment["Trust_Category"].value_counts().reset_index()

trust_category_summary.columns = ["Trust_Category", "count"]

trust_category_summary["percentage"] = (
    trust_category_summary["count"] / len(trust_alignment) * 100
)

trust_category_summary
```

```python
trust_category_by_method = trust_alignment.groupby(
    ["method", "actual_xai_method", "Trust_Category"]
).size().reset_index(name="count")

trust_category_by_method["percentage_within_method"] = (
    trust_category_by_method.groupby("method")["count"]
    .transform(lambda x: x / x.sum() * 100)
)

trust_category_by_method
```

```python
trust_category_by_outcome = trust_alignment.groupby(
    ["outcome", "Trust_Category"]
).size().reset_index(name="count")

trust_category_by_outcome["percentage_within_outcome"] = (
    trust_category_by_outcome.groupby("outcome")["count"]
    .transform(lambda x: x / x.sum() * 100)
)

trust_category_by_outcome
```

```python
strongest_alignment_cases = trust_alignment.sort_values(
    "TAI",
    ascending=False
).head(10)

strongest_alignment_cases[
    [
        "case_id",
        "xai_case_id",
        "outcome",
        "method",
        "actual_xai_method",
        "CTS_percentage",
        "HTS_percentage",
        "TrustGap",
        "TAI",
        "Trust_Category"
    ]
]
```

```python
weakest_alignment_cases = trust_alignment.sort_values(
    "TAI",
    ascending=True
).head(10)

weakest_alignment_cases[
    [
        "case_id",
        "xai_case_id",
        "outcome",
        "method",
        "actual_xai_method",
        "CTS_percentage",
        "HTS_percentage",
        "TrustGap",
        "TAI",
        "Trust_Category"
    ]
]
```

```python
over_trust_cases = trust_alignment[
    trust_alignment["Trust_Category"] == "Potential over-trust"
].sort_values("TrustGap", ascending=False)

over_trust_cases
```

```python
under_trust_cases = trust_alignment[
    trust_alignment["Trust_Category"] == "Potential under-trust"
].sort_values("TrustGap", ascending=True)

under_trust_cases
```

```python
alignment_output_file = "CTS_HTS_Trust_Alignment_Results.xlsx"

with pd.ExcelWriter(alignment_output_file, engine="openpyxl") as writer:
    trust_alignment.to_excel(writer, sheet_name="Trust_Alignment_32", index=False)
    overall_alignment_summary.to_excel(writer, sheet_name="Overall_Summary", index=False)
    alignment_by_method.to_excel(writer, sheet_name="Alignment_By_Method", index=False)
    alignment_by_outcome.to_excel(writer, sheet_name="Alignment_By_Outcome", index=False)
    trust_category_summary.to_excel(writer, sheet_name="Trust_Category_Summary", index=False)
    trust_category_by_method.to_excel(writer, sheet_name="Category_By_Method", index=False)
    trust_category_by_outcome.to_excel(writer, sheet_name="Category_By_Outcome", index=False)
    strongest_alignment_cases.to_excel(writer, sheet_name="Strongest_Alignment", index=False)
    weakest_alignment_cases.to_excel(writer, sheet_name="Weakest_Alignment", index=False)
    over_trust_cases.to_excel(writer, sheet_name="Over_Trust_Cases", index=False)
    under_trust_cases.to_excel(writer, sheet_name="Under_Trust_Cases", index=False)
    cts_standard.to_excel(writer, sheet_name="CTS_32_Row_Table", index=False)
    hts_standard.to_excel(writer, sheet_name="HTS_Case_Method", index=False)

print("Saved:", alignment_output_file)
```
