```python
def extract_state_dict(checkpoint):
    if isinstance(checkpoint, dict):
        for key in ["model_state_dict", "state_dict", "model"]:
            if key in checkpoint:
                return checkpoint[key]
    return checkpoint


def clean_state_dict_keys(state_dict):
    cleaned = {}

    for key, value in state_dict.items():
        new_key = key

        if new_key.startswith("module."):
            new_key = new_key.replace("module.", "", 1)

        cleaned[new_key] = value

    return cleaned


checkpoint = torch.load(resnet_checkpoint_path, map_location=device)
state_dict = extract_state_dict(checkpoint)
state_dict = clean_state_dict_keys(state_dict)

if "fc.weight" not in state_dict:
    raise ValueError("Could not find fc.weight in checkpoint. Check model architecture.")

output_features = state_dict["fc.weight"].shape[0]

print("Detected ResNet50 output features:", output_features)

model = models.resnet50(weights=None)
model.fc = nn.Linear(model.fc.in_features, output_features)

model.load_state_dict(state_dict, strict=True)
model = model.to(device)
model.eval()

print("Frozen ResNet50 loaded successfully.")
```

```python
IMAGE_SIZE = 224

model_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("Preprocessing transform defined.")
```

```python
def read_dicom_as_rgb(dicom_path):
    dicom = pydicom.dcmread(str(dicom_path))
    image = dicom.pixel_array.astype(np.float32)

    if getattr(dicom, "PhotometricInterpretation", "") == "MONOCHROME1":
        image = np.max(image) - image

    image = image - np.min(image)

    if np.max(image) > 0:
        image = image / np.max(image)

    image = (image * 255).astype(np.uint8)

    rgb_image = np.stack([image, image, image], axis=-1)

    return rgb_image
```

```python
def get_dicom_path_from_row(row):
    patient_id = str(row["patientId"])

    if patient_id in dicom_index:
        return dicom_index[patient_id]

    possible_path_columns = [
        "dicom_path",
        "image_path",
        "path",
        "file_path"
    ]

    for col in possible_path_columns:
        if col in row.index and pd.notna(row[col]):
            possible_path = Path(row[col])
            if possible_path.exists():
                return possible_path

    raise FileNotFoundError(f"Could not find DICOM path for patientId: {patient_id}")


def load_image_for_xai(row):
    dicom_path = get_dicom_path_from_row(row)

    rgb_image = read_dicom_as_rgb(dicom_path)

    pil_image = Image.fromarray(rgb_image).convert("RGB")
    pil_image = pil_image.resize((IMAGE_SIZE, IMAGE_SIZE))

    rgb_resized = np.array(pil_image)

    input_tensor = model_transform(pil_image).unsqueeze(0).to(device)

    return dicom_path, rgb_resized, input_tensor
```

```python
def predict_calibrated_probability(input_tensor):
    model.eval()

    with torch.no_grad():
        logits = model(input_tensor)
        scaled_logits = logits / TEMPERATURE

        if scaled_logits.shape[1] == 1:
            probability = torch.sigmoid(scaled_logits)[0, 0].item()

        elif scaled_logits.shape[1] == 2:
            probability = torch.softmax(scaled_logits, dim=1)[0, 1].item()

        else:
            raise ValueError(f"Unexpected model output shape: {scaled_logits.shape}")

    return float(probability)
```

```python
def normalise_map(raw_map):
    raw_map = np.asarray(raw_map).astype(np.float32)

    raw_map = np.nan_to_num(
        raw_map,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    raw_map = raw_map - np.min(raw_map)

    if np.max(raw_map) > 0:
        raw_map = raw_map / np.max(raw_map)

    return raw_map.astype(np.float32)


def overlay_heatmap(rgb_image, heatmap, alpha=0.45):
    heatmap = normalise_map(heatmap)

    heatmap_uint8 = np.uint8(255 * heatmap)
    heatmap_color = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)
    heatmap_color = cv2.cvtColor(heatmap_color, cv2.COLOR_BGR2RGB)

    overlay = cv2.addWeighted(
        rgb_image.astype(np.uint8),
        1 - alpha,
        heatmap_color,
        alpha,
        0
    )

    return overlay


def save_map_and_overlay(case_id, method_name, raw_map, rgb_image):
    raw_map = np.asarray(raw_map).astype(np.float32)

    if raw_map.shape != (IMAGE_SIZE, IMAGE_SIZE):
        raw_map = cv2.resize(
            raw_map,
            (IMAGE_SIZE, IMAGE_SIZE),
            interpolation=cv2.INTER_LINEAR
        )

    normalised_map = normalise_map(raw_map)
    overlay = overlay_heatmap(rgb_image, normalised_map)

    raw_path = full_maps_folder / f"{case_id}_{method_name}_raw.npy"
    normalised_path = full_maps_folder / f"{case_id}_{method_name}_normalised.npy"
    overlay_path = full_overlays_folder / f"{case_id}_{method_name}_overlay.png"

    np.save(raw_path, raw_map)
    np.save(normalised_path, normalised_map)
    Image.fromarray(overlay).save(overlay_path)

    return raw_path, normalised_path, overlay_path, normalised_map
```

```python
from lime import lime_image
from skimage.segmentation import slic

LIME_NUM_SAMPLES = 500
LIME_NUM_FEATURES = 10
LIME_N_SEGMENTS = 50
LIME_COMPACTNESS = 10
LIME_SIGMA = 1

lime_explainer = lime_image.LimeImageExplainer(
    random_state=RANDOM_SEED
)

print("LIME explainer initialised.")
```

```python
def lime_predict_function(images):
    model.eval()

    batch_tensors = []

    for image in images:
        image = np.asarray(image)

        if image.max() <= 1.0:
            image = (image * 255).astype(np.uint8)
        else:
            image = image.astype(np.uint8)

        pil_image = Image.fromarray(image).convert("RGB")
        tensor = model_transform(pil_image)
        batch_tensors.append(tensor)

    batch_tensor = torch.stack(batch_tensors).to(device)

    with torch.no_grad():
        logits = model(batch_tensor)
        scaled_logits = logits / TEMPERATURE

        if scaled_logits.shape[1] == 1:
            pneumonia_probability = torch.sigmoid(scaled_logits)
            normal_probability = 1 - pneumonia_probability

            probabilities = torch.cat(
                [normal_probability, pneumonia_probability],
                dim=1
            )

        elif scaled_logits.shape[1] == 2:
            probabilities = torch.softmax(scaled_logits, dim=1)

        else:
            raise ValueError(f"Unexpected output shape: {scaled_logits.shape}")

    return probabilities.detach().cpu().numpy()
```

```python
def lime_segmentation_function(image):
    return slic(
        image,
        n_segments=LIME_N_SEGMENTS,
        compactness=LIME_COMPACTNESS,
        sigma=LIME_SIGMA,
        start_label=0
    )
```

```python
lime_manifest_rows = []

for _, row in tqdm(
    xai_cohort.iterrows(),
    total=len(xai_cohort),
    desc="lime"
):
    case_id = row["xai_case_id"]

    _, rgb_image, input_tensor = load_image_for_xai(row)

    start_time = time.time()

    explanation = lime_explainer.explain_instance(
        image=rgb_image,
        classifier_fn=lime_predict_function,
        labels=[1],
        hide_color=0,
        num_samples=LIME_NUM_SAMPLES,
        segmentation_fn=lime_segmentation_function
    )

    segments = explanation.segments

    if 1 not in explanation.local_exp:
        raise ValueError(f"LIME did not return label 1 explanation for {case_id}")

    local_exp = dict(explanation.local_exp[1])

    lime_map = np.zeros(segments.shape, dtype=np.float32)

    for segment_id, weight in local_exp.items():
        if weight > 0:
            lime_map[segments == segment_id] = weight

    runtime_seconds = time.time() - start_time

    raw_path, normalised_path, overlay_path, normalised_map = save_map_and_overlay(
        case_id=case_id,
        method_name="lime",
        raw_map=lime_map,
        rgb_image=rgb_image
    )

    lime_manifest_rows.append({
        "xai_case_id": case_id,
        "patientId": row["patientId"],
        "method": "lime",
        "outcome": row["outcome"],
        "reference_class": row["reference_class"],
        "predicted_class": row["predicted_class"],
        "calibrated_probability": row["calibrated_probability"],
        "map_shape": str(normalised_map.shape),
        "map_min": float(normalised_map.min()),
        "map_max": float(normalised_map.max()),
        "runtime_seconds": runtime_seconds,
        "raw_map_path": str(raw_path),
        "normalised_map_path": str(normalised_path),
        "overlay_path": str(overlay_path),
        "lime_num_samples": LIME_NUM_SAMPLES,
        "lime_num_features": LIME_NUM_FEATURES,
        "lime_n_segments": LIME_N_SEGMENTS
    })

lime_manifest = pd.DataFrame(lime_manifest_rows)

lime_manifest_path = full_manifests_folder / "full_lime_manifest.csv"
lime_manifest.to_csv(lime_manifest_path, index=False)

print("LIME manifest saved:")
print(lime_manifest_path)

print("\nRows:", len(lime_manifest))
print(lime_manifest["method"].value_counts())

assert len(lime_manifest) == 120
assert set(lime_manifest["method"]) == {"lime"}
assert lime_manifest["map_shape"].eq("(224, 224)").all()

print("\nLIME checks passed.")
```

```python
class CalibratedProbabilityWrapper(nn.Module):
    def __init__(self, base_model, temperature):
        super().__init__()
        self.base_model = base_model
        self.temperature = float(temperature)

    def forward(self, x):
        logits = self.base_model(x)
        scaled_logits = logits / self.temperature

        if scaled_logits.shape[1] == 1:
            pneumonia_probability = torch.sigmoid(scaled_logits)
            normal_probability = 1 - pneumonia_probability

            probabilities = torch.cat(
                [normal_probability, pneumonia_probability],
                dim=1
            )

        elif scaled_logits.shape[1] == 2:
            probabilities = torch.softmax(scaled_logits, dim=1)

        else:
            raise ValueError(f"Unexpected model output shape: {scaled_logits.shape}")

        return probabilities


shap_model = CalibratedProbabilityWrapper(
    base_model=model,
    temperature=TEMPERATURE
).to(device)

shap_model.eval()

print("SHAP calibrated probability wrapper created.")
```

```python
def extract_state_dict(checkpoint):
    if isinstance(checkpoint, dict):
        for key in ["model_state_dict", "state_dict", "model"]:
            if key in checkpoint:
                return checkpoint[key]
    return checkpoint


def clean_state_dict_keys(state_dict):
    cleaned = {}

    for key, value in state_dict.items():
        new_key = key

        if new_key.startswith("module."):
            new_key = new_key.replace("module.", "", 1)

        cleaned[new_key] = value

    return cleaned


checkpoint = torch.load(resnet_checkpoint_path, map_location=device)
state_dict = extract_state_dict(checkpoint)
state_dict = clean_state_dict_keys(state_dict)

if "fc.weight" not in state_dict:
    raise ValueError("Could not find fc.weight in checkpoint.")

output_features = state_dict["fc.weight"].shape[0]

print("Detected ResNet50 output features:", output_features)

model = models.resnet50(weights=None)
model.fc = nn.Linear(model.fc.in_features, output_features)

model.load_state_dict(state_dict, strict=True)
model = model.to(device)
model.eval()

print("Frozen ResNet50 loaded successfully.")
```

```python
IMAGE_SIZE = 224

model_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("Preprocessing transform defined.")
```

```python
def read_dicom_as_rgb(dicom_path):
    dicom = pydicom.dcmread(str(dicom_path))
    image = dicom.pixel_array.astype(np.float32)

    if getattr(dicom, "PhotometricInterpretation", "") == "MONOCHROME1":
        image = np.max(image) - image

    image = image - np.min(image)

    if np.max(image) > 0:
        image = image / np.max(image)

    image = (image * 255).astype(np.uint8)

    rgb_image = np.stack([image, image, image], axis=-1)

    return rgb_image


def get_dicom_path_from_row(row):
    patient_id = str(row["patientId"])

    if patient_id in dicom_index:
        return dicom_index[patient_id]

    possible_path_columns = [
        "dicom_path",
        "image_path",
        "path",
        "file_path"
    ]

    for col in possible_path_columns:
        if col in row.index and pd.notna(row[col]):
            possible_path = Path(row[col])
            if possible_path.exists():
                return possible_path

    raise FileNotFoundError(f"Could not find DICOM path for patientId: {patient_id}")


def load_image_for_xai(row):
    dicom_path = get_dicom_path_from_row(row)

    rgb_image = read_dicom_as_rgb(dicom_path)

    pil_image = Image.fromarray(rgb_image).convert("RGB")
    pil_image = pil_image.resize((IMAGE_SIZE, IMAGE_SIZE))

    rgb_resized = np.array(pil_image)

    input_tensor = model_transform(pil_image).unsqueeze(0).to(device)

    return dicom_path, rgb_resized, input_tensor
```

```python
def normalise_map(raw_map):
    raw_map = np.asarray(raw_map).astype(np.float32)

    raw_map = np.nan_to_num(
        raw_map,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    raw_map = raw_map - np.min(raw_map)

    if np.max(raw_map) > 0:
        raw_map = raw_map / np.max(raw_map)

    return raw_map.astype(np.float32)


def overlay_heatmap(rgb_image, heatmap, alpha=0.45):
    heatmap = normalise_map(heatmap)

    heatmap_uint8 = np.uint8(255 * heatmap)
    heatmap_color = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)
    heatmap_color = cv2.cvtColor(heatmap_color, cv2.COLOR_BGR2RGB)

    overlay = cv2.addWeighted(
        rgb_image.astype(np.uint8),
        1 - alpha,
        heatmap_color,
        alpha,
        0
    )

    return overlay


def save_map_and_overlay(case_id, method_name, raw_map, rgb_image):
    raw_map = np.asarray(raw_map).astype(np.float32)

    if raw_map.shape != (IMAGE_SIZE, IMAGE_SIZE):
        raw_map = cv2.resize(
            raw_map,
            (IMAGE_SIZE, IMAGE_SIZE),
            interpolation=cv2.INTER_LINEAR
        )

    normalised_map = normalise_map(raw_map)
    overlay = overlay_heatmap(rgb_image, normalised_map)

    raw_path = full_maps_folder / f"{case_id}_{method_name}_raw.npy"
    normalised_path = full_maps_folder / f"{case_id}_{method_name}_normalised.npy"
    overlay_path = full_overlays_folder / f"{case_id}_{method_name}_overlay.png"

    np.save(raw_path, raw_map)
    np.save(normalised_path, normalised_map)
    Image.fromarray(overlay).save(overlay_path)

    return raw_path, normalised_path, overlay_path, normalised_map
```

```python
SHAP_NSAMPLES = 200

if shap_manifest_path.exists():
    shap_manifest = pd.read_csv(shap_manifest_path)

    shap_manifest = shap_manifest.drop_duplicates(
        subset=["xai_case_id"],
        keep="last"
    ).reset_index(drop=True)

    shap_manifest_rows = shap_manifest.to_dict("records")
    completed_shap_cases = set(shap_manifest["xai_case_id"].tolist())

    print("Existing SHAP manifest loaded.")
    print("Completed cases:", len(completed_shap_cases))

else:
    shap_manifest_rows = []
    completed_shap_cases = set()

    print("No existing SHAP manifest found.")

for _, row in tqdm(
    xai_cohort.iterrows(),
    total=len(xai_cohort),
    desc="shap resume"
):
    case_id = row["xai_case_id"]

    if case_id in completed_shap_cases:
        continue

    shap_values = None
    shap_map = None
    normalised_map = None
    input_tensor = None

    try:
        _, rgb_image, input_tensor = load_image_for_xai(row)

        start_time = time.time()

        shap_values = shap_explainer.shap_values(
            input_tensor,
            nsamples=SHAP_NSAMPLES,
            rseed=RANDOM_SEED
        )

        shap_map = extract_positive_class_shap_map(
            shap_values,
            target_index=1
        )

        runtime_seconds = time.time() - start_time

        raw_path, normalised_path, overlay_path, normalised_map = save_map_and_overlay(
            case_id=case_id,
            method_name="shap",
            raw_map=shap_map,
            rgb_image=rgb_image
        )

        shap_manifest_rows.append({
            "xai_case_id": case_id,
            "patientId": row["patientId"],
            "method": "shap",
            "outcome": row["outcome"],
            "reference_class": row["reference_class"],
            "predicted_class": row["predicted_class"],
            "calibrated_probability": row["calibrated_probability"],
            "map_shape": str(normalised_map.shape),
            "map_min": float(normalised_map.min()),
            "map_max": float(normalised_map.max()),
            "runtime_seconds": runtime_seconds,
            "raw_map_path": str(raw_path),
            "normalised_map_path": str(normalised_path),
            "overlay_path": str(overlay_path),
            "shap_explainer": "GradientExplainer",
            "shap_background_size": int(shap_background.shape[0]),
            "shap_nsamples": SHAP_NSAMPLES,
            "recovered_after_kernel_restart": False
        })

        shap_manifest = pd.DataFrame(shap_manifest_rows)

        shap_manifest = shap_manifest.drop_duplicates(
            subset=["xai_case_id"],
            keep="last"
        ).reset_index(drop=True)

        shap_manifest.to_csv(shap_manifest_path, index=False)

        completed_shap_cases.add(case_id)

        print(
            f"{case_id} | SHAP completed | "
            f"runtime={runtime_seconds:.2f}s | "
            f"completed={len(completed_shap_cases)}/120"
        )

    except MemoryError:
        shap_manifest = pd.DataFrame(shap_manifest_rows)

        shap_manifest = shap_manifest.drop_duplicates(
            subset=["xai_case_id"],
            keep="last"
        ).reset_index(drop=True)

        shap_manifest.to_csv(shap_manifest_path, index=False)

        print("\nMemoryError occurred again.")
        print("Progress saved safely.")
        print("Completed cases:", len(completed_shap_cases))
        print("Restart kernel and run from Step 1 again, then resume.")

        raise

    finally:
        del shap_values
        del shap_map
        del normalised_map
        del input_tensor

        gc.collect()

print("\nSHAP full-cohort generation completed.")
```
