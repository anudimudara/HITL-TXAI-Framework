```python
def find_file(base_dir, filename):
    """
    Finds the newest matching file under BASE_DIR.
    """
    matches = list(base_dir.rglob(filename))
    if len(matches) == 0:
        raise FileNotFoundError(f"Could not find {filename} under {base_dir}")
    
    matches = sorted(matches, key=lambda p: p.stat().st_mtime, reverse=True)
    return matches[0]


def safe_numeric(series):
    return pd.to_numeric(series, errors="coerce")


def safe_quantile(series, q, default=None):
    series = safe_numeric(series).dropna()
    if len(series) == 0:
        return default
    return float(series.quantile(q))


def minmax(series):
    series = safe_numeric(series)
    min_val = series.min()
    max_val = series.max()
    
    if pd.isna(min_val) or pd.isna(max_val) or max_val == min_val:
        return pd.Series(np.ones(len(series)) * 0.5, index=series.index)
    
    return (series - min_val) / (max_val - min_val)


def outcome_correct(outcome):
    outcome = str(outcome).upper()
    return outcome in ["TP", "TN"]


def outcome_error(outcome):
    outcome = str(outcome).upper()
    return outcome in ["FP", "FN"]


def clean_outcome(outcome):
    return str(outcome).upper().strip()


def calculate_metrics_for_subset(data, deferred_mask, strategy_name, strategy_type, target_deferral_rate=None):
    """
    Calculates selective prediction metrics.
    
    deferred_mask = True for cases referred for human review.
    retained cases = AI output kept.
    """
    temp = data.copy()
    temp["deferred"] = deferred_mask.astype(bool)
    temp["retained"] = ~temp["deferred"]
    
    n_total = len(temp)
    n_deferred = int(temp["deferred"].sum())
    n_retained = int(temp["retained"].sum())
    
    total_errors = int(temp["is_error"].sum())
    total_correct = int(temp["is_correct"].sum())
    
    retained_df = temp[temp["retained"]]
    deferred_df = temp[temp["deferred"]]
    
    retained_correct = int(retained_df["is_correct"].sum())
    retained_errors = int(retained_df["is_error"].sum())
    
    deferred_correct = int(deferred_df["is_correct"].sum())
    deferred_errors = int(deferred_df["is_error"].sum())
    
    coverage = n_retained / n_total if n_total > 0 else np.nan
    actual_deferral_rate = n_deferred / n_total if n_total > 0 else np.nan
    
    selective_accuracy = retained_correct / n_retained if n_retained > 0 else np.nan
    selective_risk = retained_errors / n_retained if n_retained > 0 else np.nan
    
    baseline_accuracy = total_correct / n_total if n_total > 0 else np.nan
    baseline_risk = total_errors / n_total if n_total > 0 else np.nan
    
    error_capture_rate = deferred_errors / total_errors if total_errors > 0 else np.nan
    deferred_error_rate = deferred_errors / n_deferred if n_deferred > 0 else np.nan
    deferred_correct_rate = deferred_correct / n_deferred if n_deferred > 0 else np.nan
    
    error_enrichment = (
        deferred_error_rate / baseline_risk
        if baseline_risk is not None and baseline_risk > 0 and not pd.isna(deferred_error_rate)
        else np.nan
    )
    
    result = {
        "strategy_name": strategy_name,
        "strategy_type": strategy_type,
        "target_deferral_rate": target_deferral_rate,
        "actual_deferral_rate": actual_deferral_rate,
        "coverage": coverage,
        "n_total": n_total,
        "n_retained": n_retained,
        "n_deferred": n_deferred,
        "baseline_accuracy": baseline_accuracy,
        "baseline_risk": baseline_risk,
        "selective_accuracy": selective_accuracy,
        "selective_risk": selective_risk,
        "retained_correct": retained_correct,
        "retained_errors": retained_errors,
        "deferred_correct": deferred_correct,
        "deferred_errors": deferred_errors,
        "total_errors": total_errors,
        "error_capture_rate": error_capture_rate,
        "deferred_error_rate": deferred_error_rate,
        "deferred_correct_rate": deferred_correct_rate,
        "error_enrichment": error_enrichment,
        "deferred_TP": int((deferred_df["outcome_upper"] == "TP").sum()),
        "deferred_TN": int((deferred_df["outcome_upper"] == "TN").sum()),
        "deferred_FP": int((deferred_df["outcome_upper"] == "FP").sum()),
        "deferred_FN": int((deferred_df["outcome_upper"] == "FN").sum()),
        "retained_TP": int((retained_df["outcome_upper"] == "TP").sum()),
        "retained_TN": int((retained_df["outcome_upper"] == "TN").sum()),
        "retained_FP": int((retained_df["outcome_upper"] == "FP").sum()),
        "retained_FN": int((retained_df["outcome_upper"] == "FN").sum())
    }
    
    deferred_cases = deferred_df.copy()
    deferred_cases["strategy_name"] = strategy_name
    deferred_cases["strategy_type"] = strategy_type
    deferred_cases["target_deferral_rate"] = target_deferral_rate
    deferred_cases["actual_deferral_rate"] = actual_deferral_rate
    
    return result, deferred_cases


def evaluate_score_based_deferral(data, score_col, strategy_name, deferral_rate):
    """
    Defers lowest-scoring cases.
    Higher score = stronger case.
    Lower score = weaker case.
    """
    temp = data.copy()
    temp = temp.sort_values(score_col, ascending=True).reset_index(drop=True)
    
    n_total = len(temp)
    n_defer = int(math.ceil(n_total * deferral_rate))
    
    deferred_ids = set(temp.head(n_defer)["xai_case_id"])
    deferred_mask = data["xai_case_id"].isin(deferred_ids)
    
    return calculate_metrics_for_subset(
        data=data,
        deferred_mask=deferred_mask,
        strategy_name=strategy_name,
        strategy_type="Score-based",
        target_deferral_rate=deferral_rate
    )


def evaluate_rule_based_deferral(data, flag_col, strategy_name):
    """
    Defers cases based on a Boolean rule flag.
    """
    deferred_mask = data[flag_col].astype(bool)
    
    return calculate_metrics_for_subset(
        data=data,
        deferred_mask=deferred_mask,
        strategy_name=strategy_name,
        strategy_type="Rule-based",
        target_deferral_rate=None
    )
```

```python
case_level_path = find_file(BASE_DIR, "case_level_hard_case_summary.csv")
case_method_path = find_file(BASE_DIR, "case_method_failure_modes.csv")

print("Files found:")
print("Case-level hard-case summary:", case_level_path)
print("Case-method failure modes:", case_method_path)

case_level = pd.read_csv(case_level_path)
case_method = pd.read_csv(case_method_path)

print("\nRows loaded:")
print("case_level:", len(case_level))
print("case_method:", len(case_method))

print("\nCase-level columns:")
print(list(case_level.columns))

print("\nCase-method columns:")
print(list(case_method.columns))
```

```python
case_analysis = case_level.copy()

# Basic validation
required_cols = [
    "xai_case_id",
    "outcome",
    "confidence_score",
    "prediction_robustness_score",
    "mean_faithfulness_score",
    "mean_explanation_stability_score"
]

missing_cols = [col for col in required_cols if col not in case_analysis.columns]

if len(missing_cols) > 0:
    raise ValueError(f"Missing required columns in case-level file: {missing_cols}")

# Clean outcome and correctness
case_analysis["outcome_upper"] = case_analysis["outcome"].apply(clean_outcome)
case_analysis["is_correct"] = case_analysis["outcome_upper"].apply(outcome_correct).astype(int)
case_analysis["is_error"] = case_analysis["outcome_upper"].apply(outcome_error).astype(int)

# Ensure numeric columns
numeric_cols = [
    "confidence_score",
    "prediction_robustness_score",
    "mean_faithfulness_score",
    "mean_explanation_stability_score",
    "mean_localisation_score",
    "total_failure_flags",
    "methods_with_review_flag"
]

for col in numeric_cols:
    if col in case_analysis.columns:
        case_analysis[col] = safe_numeric(case_analysis[col])

# Localisation may be NaN because only positive cases with bounding boxes have localisation evidence
if "mean_localisation_score" not in case_analysis.columns:
    case_analysis["mean_localisation_score"] = np.nan

print("Prepared case-level analysis table:")
print("Rows:", len(case_analysis))
print("Unique cases:", case_analysis["xai_case_id"].nunique())

display(case_analysis.head())
```

```python
# Core score used for all 120 cases
# Localisation is not included in the main score because it is not available for all cases.

core_components = [
    "confidence_score",
    "prediction_robustness_score",
    "mean_faithfulness_score",
    "mean_explanation_stability_score"
]

case_analysis["preliminary_multidimensional_trust_score"] = case_analysis[core_components].mean(axis=1)

# Extended score only when localisation is available
extended_components = [
    "confidence_score",
    "prediction_robustness_score",
    "mean_faithfulness_score",
    "mean_explanation_stability_score",
    "mean_localisation_score"
]

case_analysis["preliminary_extended_trust_score"] = case_analysis[extended_components].mean(axis=1, skipna=True)

# Ranking: lower score = weaker case
case_analysis["trust_score_rank_weakest_first"] = case_analysis[
    "preliminary_multidimensional_trust_score"
].rank(method="min", ascending=True)

print("Preliminary trust score created.")
display(
    case_analysis[
        [
            "xai_case_id",
            "outcome",
            "confidence_score",
            "prediction_robustness_score",
            "mean_faithfulness_score",
            "mean_explanation_stability_score",
            "mean_localisation_score",
            "preliminary_multidimensional_trust_score",
            "preliminary_extended_trust_score"
        ]
    ].head(10)
)
```

```python
# Thresholds
confidence_low_threshold = 0.25
robustness_low_q25 = safe_quantile(case_analysis["prediction_robustness_score"], 0.25, default=0.75)
faithfulness_low_q25 = safe_quantile(case_analysis["mean_faithfulness_score"], 0.25, default=0.25)
stability_low_q25 = safe_quantile(case_analysis["mean_explanation_stability_score"], 0.25, default=0.25)
trust_score_low_q25 = safe_quantile(case_analysis["preliminary_multidimensional_trust_score"], 0.25, default=0.25)

thresholds = {
    "confidence_low_threshold": confidence_low_threshold,
    "robustness_low_q25": robustness_low_q25,
    "faithfulness_low_q25": faithfulness_low_q25,
    "stability_low_q25": stability_low_q25,
    "preliminary_trust_score_low_q25": trust_score_low_q25
}

print("Technical weakness thresholds:")
print(json.dumps(thresholds, indent=4))

# Weakness flags
case_analysis["weak_low_confidence"] = (
    case_analysis["confidence_score"] < confidence_low_threshold
).astype(int)

case_analysis["weak_prediction_robustness"] = (
    case_analysis["prediction_robustness_score"] <= robustness_low_q25
).astype(int)

case_analysis["weak_faithfulness"] = (
    case_analysis["mean_faithfulness_score"] <= faithfulness_low_q25
).astype(int)

case_analysis["weak_explanation_stability"] = (
    case_analysis["mean_explanation_stability_score"] <= stability_low_q25
).astype(int)

case_analysis["weak_preliminary_trustworthiness"] = (
    case_analysis["preliminary_multidimensional_trust_score"] <= trust_score_low_q25
).astype(int)

weakness_cols = [
    "weak_low_confidence",
    "weak_prediction_robustness",
    "weak_faithfulness",
    "weak_explanation_stability",
    "weak_preliminary_trustworthiness"
]

case_analysis["technical_weakness_count"] = case_analysis[weakness_cols].sum(axis=1)

# Rule-based multidimensional deferral
# This flags cases with at least two weakness indicators
# or very low preliminary trustworthiness.
case_analysis["multidimensional_rule_deferral_flag"] = (
    (case_analysis["technical_weakness_count"] >= 2) |
    (case_analysis["weak_preliminary_trustworthiness"] == 1)
).astype(int)

def weakness_reason(row):
    reasons = []
    
    if row["weak_low_confidence"] == 1:
        reasons.append("low confidence")
    if row["weak_prediction_robustness"] == 1:
        reasons.append("weak robustness")
    if row["weak_faithfulness"] == 1:
        reasons.append("low faithfulness")
    if row["weak_explanation_stability"] == 1:
        reasons.append("low stability")
    if row["weak_preliminary_trustworthiness"] == 1:
        reasons.append("low preliminary trustworthiness")
    
    if len(reasons) == 0:
        return "No major technical weakness"
    
    return "; ".join(reasons)

case_analysis["technical_weakness_reason"] = case_analysis.apply(weakness_reason, axis=1)

# Save case-level analysis
case_analysis_path = OUTPUT_DIR / "selective_prediction_case_level_analysis.csv"
case_analysis.to_csv(case_analysis_path, index=False)

print("Case-level selective prediction analysis saved:")
print(case_analysis_path)

display(
    case_analysis[
        [
            "xai_case_id",
            "outcome",
            "is_correct",
            "confidence_score",
            "prediction_robustness_score",
            "mean_faithfulness_score",
            "mean_explanation_stability_score",
            "preliminary_multidimensional_trust_score",
            "technical_weakness_count",
            "multidimensional_rule_deferral_flag",
            "technical_weakness_reason"
        ]
    ].head(15)
)
```

```python
deferral_rates = [0.00, 0.05, 0.10, 0.15, 0.20, 0.25, 0.30, 0.40, 0.50]

confidence_results = []
confidence_deferred_cases = []

for rate in deferral_rates:
    result, deferred_cases = evaluate_score_based_deferral(
        data=case_analysis,
        score_col="confidence_score",
        strategy_name="Confidence-only deferral",
        deferral_rate=rate
    )
    
    confidence_results.append(result)
    
    if len(deferred_cases) > 0:
        confidence_deferred_cases.append(deferred_cases)

confidence_results_df = pd.DataFrame(confidence_results)

if len(confidence_deferred_cases) > 0:
    confidence_deferred_cases_df = pd.concat(confidence_deferred_cases, ignore_index=True)
else:
    confidence_deferred_cases_df = pd.DataFrame()

confidence_results_path = OUTPUT_DIR / "confidence_only_deferral_results.csv"
confidence_results_df.to_csv(confidence_results_path, index=False)

print("Confidence-only deferral results saved:")
print(confidence_results_path)

display(confidence_results_df)
```

```python
multidimensional_results = []
multidimensional_deferred_cases = []

for rate in deferral_rates:
    result, deferred_cases = evaluate_score_based_deferral(
        data=case_analysis,
        score_col="preliminary_multidimensional_trust_score",
        strategy_name="Multidimensional trustworthiness deferral",
        deferral_rate=rate
    )
    
    multidimensional_results.append(result)
    
    if len(deferred_cases) > 0:
        multidimensional_deferred_cases.append(deferred_cases)

multidimensional_results_df = pd.DataFrame(multidimensional_results)

if len(multidimensional_deferred_cases) > 0:
    multidimensional_deferred_cases_df = pd.concat(multidimensional_deferred_cases, ignore_index=True)
else:
    multidimensional_deferred_cases_df = pd.DataFrame()

multidimensional_results_path = OUTPUT_DIR / "multidimensional_deferral_results.csv"
multidimensional_results_df.to_csv(multidimensional_results_path, index=False)

print("Multidimensional deferral results saved:")
print(multidimensional_results_path)

display(multidimensional_results_df)
```

```python
rule_result, rule_deferred_cases_df = evaluate_rule_based_deferral(
    data=case_analysis,
    flag_col="multidimensional_rule_deferral_flag",
    strategy_name="Rule-based multidimensional deferral"
)

rule_based_results_df = pd.DataFrame([rule_result])

rule_based_results_path = OUTPUT_DIR / "rule_based_deferral_results.csv"
rule_based_results_df.to_csv(rule_based_results_path, index=False)

print("Rule-based deferral results saved:")
print(rule_based_results_path)

display(rule_based_results_df)
```

```python
deferred_case_frames = []

if len(confidence_deferred_cases_df) > 0:
    deferred_case_frames.append(confidence_deferred_cases_df)

if len(multidimensional_deferred_cases_df) > 0:
    deferred_case_frames.append(multidimensional_deferred_cases_df)

if len(rule_deferred_cases_df) > 0:
    deferred_case_frames.append(rule_deferred_cases_df)

if len(deferred_case_frames) > 0:
    all_deferred_cases = pd.concat(deferred_case_frames, ignore_index=True)
else:
    all_deferred_cases = pd.DataFrame()

deferred_cases_path = OUTPUT_DIR / "deferred_cases_by_strategy.csv"
all_deferred_cases.to_csv(deferred_cases_path, index=False)

print("Deferred cases by strategy saved:")
print(deferred_cases_path)

display(all_deferred_cases.head(20))
```

```python
comparison = confidence_results_df.merge(
    multidimensional_results_df,
    on="target_deferral_rate",
    suffixes=("_confidence_only", "_multidimensional")
)

comparison["selective_accuracy_gain"] = (
    comparison["selective_accuracy_multidimensional"] -
    comparison["selective_accuracy_confidence_only"]
)

comparison["selective_risk_reduction"] = (
    comparison["selective_risk_confidence_only"] -
    comparison["selective_risk_multidimensional"]
)

comparison["error_capture_gain"] = (
    comparison["error_capture_rate_multidimensional"] -
    comparison["error_capture_rate_confidence_only"]
)

comparison["deferred_error_rate_gain"] = (
    comparison["deferred_error_rate_multidimensional"] -
    comparison["deferred_error_rate_confidence_only"]
)

comparison_path = OUTPUT_DIR / "deferral_comparison_summary.csv"
comparison.to_csv(comparison_path, index=False)

print("Deferral comparison summary saved:")
print(comparison_path)

display(
    comparison[
        [
            "target_deferral_rate",
            "coverage_confidence_only",
            "coverage_multidimensional",
            "selective_accuracy_confidence_only",
            "selective_accuracy_multidimensional",
            "selective_accuracy_gain",
            "selective_risk_confidence_only",
            "selective_risk_multidimensional",
            "selective_risk_reduction",
            "error_capture_rate_confidence_only",
            "error_capture_rate_multidimensional",
            "error_capture_gain"
        ]
    ]
)
```

```python
all_results_df = pd.concat(
    [
        confidence_results_df,
        multidimensional_results_df,
        rule_based_results_df
    ],
    ignore_index=True
)

error_capture_summary = all_results_df[
    [
        "strategy_name",
        "strategy_type",
        "target_deferral_rate",
        "actual_deferral_rate",
        "coverage",
        "n_deferred",
        "selective_accuracy",
        "selective_risk",
        "error_capture_rate",
        "deferred_error_rate",
        "error_enrichment",
        "deferred_FP",
        "deferred_FN",
        "deferred_TP",
        "deferred_TN"
    ]
].copy()

error_capture_summary_path = OUTPUT_DIR / "deferral_error_capture_summary.csv"
error_capture_summary.to_csv(error_capture_summary_path, index=False)

print("Deferral error-capture summary saved:")
print(error_capture_summary_path)

display(error_capture_summary)
```
