```python
# ============================================================
# STEP 7 — VERIFY SOURCE COLUMNS
# ============================================================

required_columns = [

    "case_id",
    "xai_case_id",
    "outcome",
    "prediction_probability",

    "CTS_star_unit",
    "CTS_star_percentage",

    "HTS_unit",
    "HTS_percentage",

    "EAS_unit",
    "EAS_Percentage",

    "UA_TAI_unit",
    "UA_TAI_percentage",

    "Raw_Alignment_unit",
    "Raw_Alignment_percentage",

    "UA_TAI_Quartile_Band",
    "Trust_Quadrant",

    "actual_xai_method",

    "EAS_discount_factor",

    "UA_TAI_kappa_0_5_unit"
]

missing = [
    col
    for col in required_columns
    if col not in df.columns
]

if missing:

    print("Missing fields:")

    for col in missing:
        print("-", col)

else:

    print("PASS: Required source fields available.")
```

```python
# ============================================================
# STEP 8 — CREATE CLEAN DOSSIER TABLE
# ============================================================

dossier_df = pd.DataFrame({

    "Case_ID":
        df["xai_case_id"],

    "Case_Number":
        df["case_id"],

    "Method_Original":
        df["actual_xai_method"],

    "Outcome":
        df["outcome"]
        .astype(str)
        .str.strip()
        .str.upper(),

    "Prediction_Probability":
        pd.to_numeric(
            df["prediction_probability"],
            errors="coerce"
        ),

    "CTS_star":
        pd.to_numeric(
            df["CTS_star_unit"],
            errors="coerce"
        ),

    "HTS":
        pd.to_numeric(
            df["HTS_unit"],
            errors="coerce"
        ),

    "EAS":
        pd.to_numeric(
            df["EAS_unit"],
            errors="coerce"
        ),

    "UA_TAI":
        pd.to_numeric(
            df["UA_TAI_unit"],
            errors="coerce"
        ),

    "CTS_star_pct":
        pd.to_numeric(
            df["CTS_star_percentage"],
            errors="coerce"
        ),

    "HTS_pct":
        pd.to_numeric(
            df["HTS_percentage"],
            errors="coerce"
        ),

    "EAS_pct":
        pd.to_numeric(
            df["EAS_Percentage"],
            errors="coerce"
        ),

    "UA_TAI_pct":
        pd.to_numeric(
            df["UA_TAI_percentage"],
            errors="coerce"
        ),

    "UA_TAI_Band":
        df["UA_TAI_Quartile_Band"],

    "Trust_Quadrant":
        df["Trust_Quadrant"]
})

print(
    "Dossier table shape:",
    dossier_df.shape
)

display(dossier_df.head())
```

```python
# ============================================================
# STEP 9 — STANDARDISE XAI METHOD NAMES
# ============================================================

method_map = {

    "Method A": "Grad-CAM",
    "A": "Grad-CAM",

    "Method B": "Grad-CAM++",
    "B": "Grad-CAM++",

    "Method C": "LIME",
    "C": "LIME",

    "Method D": "SHAP",
    "D": "SHAP"
}


def standardise_method(value):

    text = str(value).strip()

    return method_map.get(
        text,
        text
    )


dossier_df["XAI_Method"] = (
    dossier_df["Method_Original"]
    .apply(standardise_method)
)

print(
    dossier_df[
        "XAI_Method"
    ].value_counts()
)
```

```python
# ============================================================
# STEP 10 — TRUST GAP
# ============================================================

dossier_df["TrustGap"] = (
    dossier_df["HTS"]
    -
    dossier_df["CTS_star"]
)

dossier_df["TrustGap_pct_points"] = (
    dossier_df["TrustGap"]
    *
    100
)

display(
    dossier_df[
        [
            "Case_ID",
            "XAI_Method",
            "CTS_star",
            "HTS",
            "TrustGap"
        ]
    ].head()
)
```

```python
# ============================================================
# STEP 11 — RAW TRUST ALIGNMENT (RTA)
# ============================================================

dossier_df["RTA"] = (
    1
    -
    np.abs(
        dossier_df["HTS"]
        -
        dossier_df["CTS_star"]
    )
)

dossier_df["RTA_pct"] = (
    dossier_df["RTA"]
    *
    100
)

display(
    dossier_df[
        [
            "Case_ID",
            "XAI_Method",
            "TrustGap",
            "RTA"
        ]
    ].head()
)
```

```python
# ============================================================
# STEP 12 — VERIFY RTA
# ============================================================

saved_rta = pd.to_numeric(
    df["Raw_Alignment_unit"],
    errors="coerce"
)

dossier_df[
    "Saved_Raw_Alignment"
] = saved_rta

dossier_df[
    "RTA_Verification_Difference"
] = np.abs(
    dossier_df["RTA"]
    -
    dossier_df["Saved_Raw_Alignment"]
)

dossier_df[
    "RTA_Consistent"
] = (
    dossier_df[
        "RTA_Verification_Difference"
    ]
    <= 1e-6
)

print(
    dossier_df[
        "RTA_Consistent"
    ].value_counts()
)

print(
    "\nMaximum difference:",
    dossier_df[
        "RTA_Verification_Difference"
    ].max()
)
```

```python
# ============================================================
# STEP 13 — VERIFY UA-TAI USING RTA
# ============================================================

KAPPA = 0.5

dossier_df[
    "EAS_Discount"
] = np.exp(
    -KAPPA
    *
    (
        1
        -
        dossier_df["EAS"]
    )
)

dossier_df[
    "UA_TAI_Recomputed"
] = (
    dossier_df["RTA"]
    *
    dossier_df["EAS_Discount"]
)

dossier_df[
    "UA_TAI_Difference"
] = np.abs(
    dossier_df["UA_TAI"]
    -
    dossier_df[
        "UA_TAI_Recomputed"
    ]
)

dossier_df[
    "UA_TAI_Consistent"
] = (
    dossier_df[
        "UA_TAI_Difference"
    ]
    <= 1e-6
)

print(
    dossier_df[
        "UA_TAI_Consistent"
    ].value_counts()
)
```

```python
# ============================================================
# STEP 14 — VERIFY KAPPA PROFILE
# ============================================================

saved_kappa_05 = pd.to_numeric(
    df["UA_TAI_kappa_0_5_unit"],
    errors="coerce"
)

dossier_df[
    "Saved_UA_TAI_Kappa_05"
] = saved_kappa_05

dossier_df[
    "Kappa_05_Difference"
] = np.abs(
    dossier_df["UA_TAI"]
    -
    dossier_df[
        "Saved_UA_TAI_Kappa_05"
    ]
)

dossier_df[
    "Kappa_05_Consistent"
] = (
    dossier_df[
        "Kappa_05_Difference"
    ]
    <= 1e-6
)

print(
    dossier_df[
        "Kappa_05_Consistent"
    ].value_counts()
)
```

```python
# ============================================================
# STEP 15 — OUTCOME FLAGS
# ============================================================

dossier_df["Correct_Prediction"] = (
    dossier_df["Outcome"]
    .isin(["TP", "TN"])
)

dossier_df["Error_Flag"] = (
    dossier_df["Outcome"]
    .isin(["FP", "FN"])
)

dossier_df["False_Positive"] = (
    dossier_df["Outcome"]
    == "FP"
)

dossier_df["False_Negative"] = (
    dossier_df["Outcome"]
    == "FN"
)

print(
    "Correct records:",
    dossier_df[
        "Correct_Prediction"
    ].sum()
)

print(
    "Error records:",
    dossier_df[
        "Error_Flag"
    ].sum()
)
```

```python
# ============================================================
# STEP 16 — GOVERNANCE PRECEDENCE
# ============================================================

dossier_df[
    "Mandatory_Outcome_Review"
] = dossier_df[
    "Error_Flag"
]

dossier_df[
    "Governance_Priority"
] = np.where(

    dossier_df[
        "Mandatory_Outcome_Review"
    ],

    "P1 - Mandatory outcome review",

    "P2 - Trust-alignment governance"
)

dossier_df[
    "Governance_Precedence_Rule"
] = np.where(

    dossier_df[
        "Mandatory_Outcome_Review"
    ],

    (
        "Outcome override supersedes "
        "band/quadrant recommendation"
    ),

    (
        "Band/quadrant recommendation applies"
    )
)

print(
    dossier_df[
        "Governance_Priority"
    ].value_counts()
)
```

```python
# ============================================================
# STEP 17 — TRUST QUADRANT STANDARDISATION
# ============================================================

def standardise_quadrant(value):

    text = str(value).lower().strip()

    if "over" in text:
        return "Over-trust"

    if "under" in text:
        return "Under-trust"

    if "high" in text:
        return "Aligned-high"

    if (
        "cautious" in text
        or "low" in text
    ):
        return "Aligned-cautious"

    return str(value)


dossier_df[
    "Trust_Quadrant_Final"
] = (
    dossier_df[
        "Trust_Quadrant"
    ].apply(
        standardise_quadrant
    )
)
```

```python
# ============================================================
# STEP 18 — FINAL GOVERNANCE ACTION
# ============================================================

def governance_action(row):

    outcome = row["Outcome"]
    quadrant = row[
        "Trust_Quadrant_Final"
    ]

    # --------------------------------------------------------
    # FALSE NEGATIVE — HIGHEST ERROR PRIORITY
    # --------------------------------------------------------

    if outcome == "FN":

        if quadrant == "Over-trust":

            return (
                "Mandatory high-priority false-negative review; "
                "investigate potential over-trust and whether "
                "the explanation encouraged excessive confidence."
            )

        elif quadrant == "Under-trust":

            return (
                "Mandatory high-priority false-negative review; "
                "additionally investigate explanation clarity "
                "and trust calibration."
            )

        elif quadrant == "Aligned-high":

            return (
                "Mandatory high-priority false-negative review "
                "despite relatively high trust alignment."
            )

        else:

            return (
                "Mandatory high-priority false-negative review; "
                "maintain heightened caution."
            )

    # --------------------------------------------------------
    # FALSE POSITIVE
    # --------------------------------------------------------

    if outcome == "FP":

        if quadrant == "Over-trust":

            return (
                "Mandatory false-positive review; "
                "investigate potential excessive human trust."
            )

        elif quadrant == "Under-trust":

            return (
                "Mandatory false-positive review; "
                "additionally review explanation clarity."
            )

        elif quadrant == "Aligned-high":

            return (
                "Mandatory false-positive review despite "
                "relatively high trust alignment."
            )

        else:

            return (
                "Mandatory false-positive review; "
                "maintain error-case caution."
            )

    # --------------------------------------------------------
    # CORRECT CASES
    # --------------------------------------------------------

    if quadrant == "Over-trust":

        return (
            "Trust review required; human trust exceeds "
            "relative computational evidence."
        )

    elif quadrant == "Under-trust":

        return (
            "Review explanation clarity and specialist confidence."
        )

    elif quadrant == "Aligned-high":

        return (
            "Routine review; computational and human "
            "trust are relatively aligned."
        )

    else:

        return (
            "Standard caution; computational and human "
            "trust are both relatively low."
        )


dossier_df[
    "Final_Governance_Action"
] = dossier_df.apply(
    governance_action,
    axis=1
)
```

```python
# ============================================================
# STEP 19 — GOVERNANCE OVERRIDE VALIDATION
# ============================================================

error_records = dossier_df.loc[
    dossier_df["Error_Flag"]
]

mandatory_count = (
    error_records[
        "Final_Governance_Action"
    ]
    .str.contains(
        "Mandatory",
        case=False,
        na=False
    )
    .sum()
)

print(
    "FP/FN records:",
    len(error_records)
)

print(
    "FP/FN with mandatory action:",
    mandatory_count
)
```

```python
# ============================================================
# STEP 27 — SOURCE FILE HASH
# ============================================================

def sha256_file(file_path):

    sha = hashlib.sha256()

    with open(
        file_path,
        "rb"
    ) as f:

        for block in iter(
            lambda:
                f.read(
                    1024 * 1024
                ),
            b""
        ):
            sha.update(block)

    return sha.hexdigest()


SOURCE_SHA256 = sha256_file(
    FINAL_RESULTS_FILE
)

dossier_df[
    "Source_Workbook_SHA256"
] = SOURCE_SHA256

print(
    "Source workbook SHA-256:"
)

print(SOURCE_SHA256)
```
