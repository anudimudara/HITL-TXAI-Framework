```python
def detect_case_column(df):
    priority_names = [
        "xai_case_id",
        "xai_case",
        "case_id",
        "case",
        "survey_case",
        "case_name"
    ]
    
    for name in priority_names:
        if name in df.columns:
            return name
    
    for c in df.columns:
        lc = str(c).lower()
        if "xai" in lc and "case" in lc:
            return c
    
    for c in df.columns:
        lc = str(c).lower()
        if "case" in lc:
            return c
    
    return None


def detect_image_path_column(df):
    priority_names = [
        "image_path",
        "original_image_path",
        "file_path",
        "saved_path",
        "png_path",
        "original_path",
        "path"
    ]
    
    for name in priority_names:
        if name in df.columns:
            return name
    
    for c in df.columns:
        lc = str(c).lower()
        if "path" in lc or "file" in lc or "image" in lc or "png" in lc or "jpg" in lc:
            return c
    
    return None


manifest_case_col = detect_case_column(original_manifest_df)
manifest_image_col = detect_image_path_column(original_manifest_df)

print("Detected case column:", manifest_case_col)
print("Detected image path column:", manifest_image_col)

if manifest_case_col is None:
    raise ValueError("Could not detect case column in original image manifest.")

if manifest_image_col is None:
    raise ValueError("Could not detect image path column in original image manifest.")

display(original_manifest_df[[manifest_case_col, manifest_image_col]].head(20))
```

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

def load_image_224_rgb_any(image_path):
    """
    Loads PNG/JPG/JPEG image as 224 x 224 RGB float array in range 0-1.
    """
    image_path = Path(image_path)
    
    if not image_path.exists():
        raise FileNotFoundError(f"Image not found: {image_path}")
    
    suffix = image_path.suffix.lower()
    
    if suffix in [".png", ".jpg", ".jpeg"]:
        img = Image.open(image_path).convert("RGB")
        img = img.resize((224, 224))
        arr = np.asarray(img).astype(np.float32) / 255.0
        return arr
    
    else:
        raise ValueError(f"Unsupported image type: {suffix}")
```

```python
def extract_state_dict(checkpoint):
    if isinstance(checkpoint, dict):
        for key in ["model_state_dict", "state_dict", "model"]:
            if key in checkpoint:
                return checkpoint[key]
    return checkpoint


def clean_state_dict_keys(state_dict):
    cleaned = {}

    for key, value in state_dict.items():
        new_key = key

        if new_key.startswith("module."):
            new_key = new_key.replace("module.", "", 1)

        cleaned[new_key] = value

    return cleaned


checkpoint = torch.load(resnet_checkpoint_path, map_location=device)

state_dict = extract_state_dict(checkpoint)
state_dict = clean_state_dict_keys(state_dict)

if "fc.weight" not in state_dict:
    raise ValueError("Could not find fc.weight in checkpoint. Check model architecture.")

output_features = state_dict["fc.weight"].shape[0]

print("Detected ResNet50 output features:", output_features)

model = models.resnet50(weights=None)
model.fc = nn.Linear(model.fc.in_features, output_features)

model.load_state_dict(state_dict, strict=True)
model = model.to(device)
model.eval()

print("Frozen ResNet50 loaded successfully.")
```

```python
import torch

imagenet_mean = torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1).to(device)
imagenet_std = torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1).to(device)


def image_array_to_tensor(image_224_rgb_float):
    """
    Converts 224 x 224 x 3 image array in range 0-1 into model tensor.
    """
    img = np.asarray(image_224_rgb_float, dtype=np.float32)

    if img.max() > 1.5:
        img = img / 255.0

    tensor = torch.tensor(img).permute(2, 0, 1).unsqueeze(0).to(device)
    tensor = (tensor - imagenet_mean) / imagenet_std

    return tensor


def predict_calibrated_probability(input_tensor):
    model.eval()

    with torch.no_grad():
        logits = model(input_tensor)
        scaled_logits = logits / TEMPERATURE

        if scaled_logits.shape[1] == 1:
            probability = torch.sigmoid(scaled_logits)[0, 0].item()

        elif scaled_logits.shape[1] == 2:
            probability = torch.softmax(scaled_logits, dim=1)[0, 1].item()

        else:
            raise ValueError(f"Unexpected model output shape: {scaled_logits.shape}")

    return float(probability)


def predict_probability(image_224_rgb_float):
    """
    TCPA wrapper.
    Input: 224 x 224 x 3 image array, range 0-1.
    Output: calibrated lung-opacity probability between 0 and 1.
    """
    input_tensor = image_array_to_tensor(image_224_rgb_float)
    probability = predict_calibrated_probability(input_tensor)
    return float(probability)
```

```python
FINAL_THRESHOLD = 0.35
PATCH_SIZE = 32
IMAGE_SIZE = 224

def predicted_class_from_prob(prob):
    if prob >= FINAL_THRESHOLD:
        return "Lung opacity"
    else:
        return "Normal"


def target_score_from_prob(prob, predicted_class):
    if predicted_class == "Lung opacity":
        return prob
    else:
        return 1.0 - prob


def predict_batch_probability(images):
    probs = []
    
    for img in images:
        probs.append(predict_probability(img))
    
    return np.array(probs, dtype=np.float32)
```

```python
def get_patch_coordinates(image_size=224, patch_size=32):
    coords = []
    
    for y in range(0, image_size, patch_size):
        for x in range(0, image_size, patch_size):
            y2 = min(y + patch_size, image_size)
            x2 = min(x + patch_size, image_size)
            
            coords.append((x, y, x2, y2))
    
    return coords

patch_coords = get_patch_coordinates(IMAGE_SIZE, PATCH_SIZE)

print("Number of patches:", len(patch_coords))
print(patch_coords[:5])
```

```python
from PIL import ImageFilter, ImageEnhance

def make_blurred_baseline(image, blur_radius=8):
    img_uint8 = (image * 255).astype(np.uint8)
    pil_img = Image.fromarray(img_uint8)
    blurred = pil_img.filter(ImageFilter.GaussianBlur(radius=blur_radius))
    arr = np.asarray(blurred).astype(np.float32) / 255.0
    return arr


def create_deletion_images(image, baseline, patch_coords):
    deletion_images = []
    
    for (x1, y1, x2, y2) in patch_coords:
        modified = image.copy()
        modified[y1:y2, x1:x2, :] = baseline[y1:y2, x1:x2, :]
        deletion_images.append(modified)
    
    return deletion_images


def create_insertion_images(image, baseline, patch_coords):
    insertion_images = []
    
    for (x1, y1, x2, y2) in patch_coords:
        modified = baseline.copy()
        modified[y1:y2, x1:x2, :] = image[y1:y2, x1:x2, :]
        insertion_images.append(modified)
    
    return insertion_images
```

```python
def normalise_01(values):
    values = np.asarray(values, dtype=np.float32)
    
    min_val = np.nanmin(values)
    max_val = np.nanmax(values)
    
    if max_val - min_val < 1e-8:
        return np.zeros_like(values, dtype=np.float32)
    
    return (values - min_val) / (max_val - min_val)


def calculate_patch_faithfulness(image, patch_coords, predicted_class):
    baseline = make_blurred_baseline(image)
    
    original_prob = predict_probability(image)
    original_target_score = target_score_from_prob(original_prob, predicted_class)
    
    baseline_prob = predict_probability(baseline)
    baseline_target_score = target_score_from_prob(baseline_prob, predicted_class)
    
    deletion_images = create_deletion_images(image, baseline, patch_coords)
    insertion_images = create_insertion_images(image, baseline, patch_coords)
    
    deletion_probs = predict_batch_probability(deletion_images)
    insertion_probs = predict_batch_probability(insertion_images)
    
    deletion_target_scores = np.array([
        target_score_from_prob(p, predicted_class) for p in deletion_probs
    ])
    
    insertion_target_scores = np.array([
        target_score_from_prob(p, predicted_class) for p in insertion_probs
    ])
    
    deletion_scores = np.maximum(0, original_target_score - deletion_target_scores)
    insertion_scores = np.maximum(0, insertion_target_scores - baseline_target_score)
    
    faithfulness_raw = 0.6 * deletion_scores + 0.4 * insertion_scores
    faithfulness_norm = normalise_01(faithfulness_raw)
    
    return {
        "original_prob": original_prob,
        "original_target_score": original_target_score,
        "baseline_prob": baseline_prob,
        "baseline_target_score": baseline_target_score,
        "deletion_scores": deletion_scores,
        "insertion_scores": insertion_scores,
        "faithfulness_raw": faithfulness_raw,
        "faithfulness_norm": faithfulness_norm
    }
```

```python
from PIL import ImageEnhance
import numpy as np

def perturb_gaussian_noise(image, std=8/255.0):
    noisy = image + np.random.normal(0, std, image.shape).astype(np.float32)
    return np.clip(noisy, 0, 1)


def perturb_brightness(image, factor=1.10):
    img_uint8 = (image * 255).astype(np.uint8)
    pil_img = Image.fromarray(img_uint8)
    enhanced = ImageEnhance.Brightness(pil_img).enhance(factor)
    return np.asarray(enhanced).astype(np.float32) / 255.0


def perturb_contrast(image, factor=1.10):
    img_uint8 = (image * 255).astype(np.uint8)
    pil_img = Image.fromarray(img_uint8)
    enhanced = ImageEnhance.Contrast(pil_img).enhance(factor)
    return np.asarray(enhanced).astype(np.float32) / 255.0


def perturb_translation(image, shift_x=5, shift_y=5):
    shifted = np.zeros_like(image)
    
    h, w, c = image.shape
    
    src_x1 = max(0, -shift_x)
    src_x2 = min(w, w - shift_x)
    dst_x1 = max(0, shift_x)
    dst_x2 = min(w, w + shift_x)
    
    src_y1 = max(0, -shift_y)
    src_y2 = min(h, h - shift_y)
    dst_y1 = max(0, shift_y)
    dst_y2 = min(h, h + shift_y)
    
    shifted[dst_y1:dst_y2, dst_x1:dst_x2, :] = image[src_y1:src_y2, src_x1:src_x2, :]
    
    return shifted


def perturb_rotation(image, angle=5):
    img_uint8 = (image * 255).astype(np.uint8)
    pil_img = Image.fromarray(img_uint8)
    rotated = pil_img.rotate(angle, resample=Image.BILINEAR, fillcolor=(0, 0, 0))
    return np.asarray(rotated).astype(np.float32) / 255.0


def generate_perturbations(image):
    return {
        "gaussian_noise_std8": perturb_gaussian_noise(image),
        "brightness_plus_10": perturb_brightness(image),
        "contrast_plus_10": perturb_contrast(image),
        "translation_5px": perturb_translation(image, shift_x=5, shift_y=5),
        "rotation_5deg": perturb_rotation(image, angle=5)
    }
```

```python
def calculate_patch_stability(image, patch_coords, predicted_class):
    """
    Recomputes patch faithfulness under perturbations and calculates patch stability.
    """
    perturbations = generate_perturbations(image)
    
    faithfulness_matrix = []
    perturbation_rows = []
    
    # Original image faithfulness
    original_result = calculate_patch_faithfulness(
        image=image,
        patch_coords=patch_coords,
        predicted_class=predicted_class
    )
    
    faithfulness_matrix.append(original_result["faithfulness_norm"])
    
    perturbation_rows.append({
        "perturbation": "original",
        "probability": original_result["original_prob"],
        "target_score": original_result["original_target_score"]
    })
    
    # Perturbed images
    for perturb_name, perturbed_img in perturbations.items():
        print("Running perturbation:", perturb_name)
        
        result = calculate_patch_faithfulness(
            image=perturbed_img,
            patch_coords=patch_coords,
            predicted_class=predicted_class
        )
        
        faithfulness_matrix.append(result["faithfulness_norm"])
        
        perturbation_rows.append({
            "perturbation": perturb_name,
            "probability": result["original_prob"],
            "target_score": result["original_target_score"]
        })
    
    faithfulness_matrix = np.array(faithfulness_matrix)
    
    patch_mean = np.mean(faithfulness_matrix, axis=0)
    patch_std = np.std(faithfulness_matrix, axis=0)
    
    epsilon = 1e-6
    
    stability = 1 - (patch_std / (patch_mean + epsilon))
    stability = np.clip(stability, 0, 1)
    
    perturbation_df = pd.DataFrame(perturbation_rows)
    
    return {
        "faithfulness_matrix": faithfulness_matrix,
        "patch_mean": patch_mean,
        "patch_std": patch_std,
        "stability": stability,
        "perturbation_df": perturbation_df,
        "original_faithfulness_result": original_result
    }
```

```python
def run_tcpa_for_image(case_id, image, patch_size=32):
    patch_coords = get_patch_coordinates(224, patch_size)
    
    original_prob = predict_probability(image)
    predicted_class = predicted_class_from_prob(original_prob)
    confidence_score = target_score_from_prob(original_prob, predicted_class)
    
    stability_result = calculate_patch_stability(
        image=image,
        patch_coords=patch_coords,
        predicted_class=predicted_class
    )
    
    original_faithfulness = stability_result["original_faithfulness_result"]["faithfulness_norm"]
    patch_stability = stability_result["stability"]
    
    tcpa_patch_scores = original_faithfulness * patch_stability
    tcpa_patch_scores = normalise_01(tcpa_patch_scores)
    
    tcpa_heatmap = patch_scores_to_heatmap(
        patch_scores=tcpa_patch_scores,
        patch_coords=patch_coords,
        image_size=224
    )
    
    top_k = max(1, int(np.ceil(len(tcpa_patch_scores) * 0.20)))
    top_indices = np.argsort(tcpa_patch_scores)[-top_k:]
    
    global_faithfulness = float(np.mean(original_faithfulness[top_indices]) * 100)
    
    weights = original_faithfulness + 1e-6
    global_stability = float(np.average(patch_stability, weights=weights) * 100)
    
    perturb_df = stability_result["perturbation_df"]
    original_target = perturb_df[perturb_df["perturbation"] == "original"]["target_score"].iloc[0]
    perturb_only = perturb_df[perturb_df["perturbation"] != "original"].copy()
    
    mean_target_change = np.mean(np.abs(perturb_only["target_score"] - original_target))
    prediction_robustness = float(np.clip((1 - mean_target_change) * 100, 0, 100))
    calibrated_confidence = float(confidence_score * 100)
    
    tcpa_reliability_general = (
        0.05 * calibrated_confidence
        + 0.15 * prediction_robustness
        + 0.50 * global_faithfulness
        + 0.30 * global_stability
    )
    
    warning_flags = []
    
    if tcpa_reliability_general < 50:
        warning_flags.append("Weak explanation reliability")
    
    if calibrated_confidence >= 80 and tcpa_reliability_general < 60:
        warning_flags.append("High-confidence weak-explanation risk")
    
    if global_faithfulness < 50:
        warning_flags.append("Low faithfulness warning")
    
    if global_stability < 50:
        warning_flags.append("Explanation instability warning")
    
    if len(warning_flags) == 0:
        warning_flags.append("No major TCPA warning")
    
    result_summary = {
        "xai_case_id": case_id,
        "original_probability": original_prob,
        "predicted_class": predicted_class,
        "calibrated_confidence_pct": calibrated_confidence,
        "prediction_robustness_pct": prediction_robustness,
        "tcpa_global_faithfulness_pct": global_faithfulness,
        "tcpa_global_stability_pct": global_stability,
        "tcpa_reliability_general_pct": tcpa_reliability_general,
        "warning_flags": "; ".join(warning_flags)
    }
    
    patch_df = pd.DataFrame(patch_coords, columns=["x1", "y1", "x2", "y2"])
    patch_df["patch_index"] = range(len(patch_df))
    patch_df["faithfulness_score"] = original_faithfulness
    patch_df["stability_score"] = patch_stability
    patch_df["tcpa_patch_score"] = tcpa_patch_scores
    
    return {
        "summary": result_summary,
        "patch_df": patch_df,
        "perturbation_df": perturb_df,
        "heatmap": tcpa_heatmap
    }
```

```python
from scipy.stats import spearmanr
import numpy as np
import pandas as pd
from PIL import Image

try:
    from skimage.metrics import structural_similarity as ssim
    SKIMAGE_AVAILABLE = True
    print("scikit-image available. SSIM will be calculated.")
except ImportError:
    SKIMAGE_AVAILABLE = False
    print("scikit-image not available. SSIM will be skipped.")

METHOD_KEYS = {
    "Grad-CAM": "gradcam",
    "Grad-CAM++": "gradcampp",
    "LIME": "lime",
    "SHAP": "shap"
}

methods = list(METHOD_KEYS.keys())

XAI_MAP_ROOT = PROJECT_ROOT / "xai_analysis" / "full_cohort" / "maps"
XAI_PILOT_MAP_ROOT = PROJECT_ROOT / "xai_analysis" / "xai_pilot" / "maps"


def find_existing_xai_map(case_id, method):
    method_key = METHOD_KEYS[method]
    filename = f"{case_id}_{method_key}_normalised.npy"
    
    full_path = XAI_MAP_ROOT / filename
    pilot_path = XAI_PILOT_MAP_ROOT / filename
    
    if full_path.exists():
        return full_path
    
    if pilot_path.exists():
        return pilot_path
    
    return None


def load_heatmap_npy(path, target_size=(224, 224)):
    arr = np.load(path)
    arr = np.squeeze(arr).astype(np.float32)
    
    if arr.ndim == 3:
        arr = arr.mean(axis=2)
    
    arr = normalise_01(arr)
    
    if arr.shape != target_size:
        img = Image.fromarray((arr * 255).astype(np.uint8))
        img = img.resize(target_size, Image.BILINEAR)
        arr = np.asarray(img).astype(np.float32) / 255.0
        arr = normalise_01(arr)
    
    return arr


def cosine_similarity_map(a, b):
    a = a.flatten().astype(np.float32)
    b = b.flatten().astype(np.float32)
    
    denom = np.linalg.norm(a) * np.linalg.norm(b)
    
    if denom < 1e-8:
        return np.nan
    
    return float(np.clip(np.dot(a, b) / denom, -1, 1))


def spearman_similarity_map(a, b):
    a = a.flatten()
    b = b.flatten()
    
    if np.std(a) < 1e-8 or np.std(b) < 1e-8:
        return np.nan
    
    rho, p = spearmanr(a, b)
    return float(rho)


def topk_overlap(a, b, top_percent=10):
    a = a.flatten()
    b = b.flatten()
    
    n = len(a)
    k = int(np.ceil(n * top_percent / 100))
    
    top_a = np.argsort(a)[-k:]
    top_b = np.argsort(b)[-k:]
    
    return float(len(set(top_a).intersection(set(top_b))) / k)


def ssim_similarity_map(a, b):
    if not SKIMAGE_AVAILABLE:
        return np.nan
    
    return float(ssim(a, b, data_range=1.0))


def spearman_to_01(rho):
    if pd.isna(rho):
        return np.nan
    return (rho + 1) / 2


def safe_01(value):
    if pd.isna(value):
        return np.nan
    return float(np.clip(value, 0, 1))
```

```python
tcpa_similarity_rows = []

for _, row in tcpa_summary_df.iterrows():
    case_id = row["xai_case_id"]
    outcome = row["outcome"]
    
    tcpa_path = TCPA_MAP_DIR / f"{case_id}_TCPA_normalised.npy"
    
    if not tcpa_path.exists():
        print("Missing TCPA map:", case_id)
        continue
    
    tcpa_map = load_heatmap_npy(tcpa_path)
    
    for method in methods:
        method_path = find_existing_xai_map(case_id, method)
        
        if method_path is None:
            print("Missing existing XAI map:", case_id, method)
            continue
        
        method_map = load_heatmap_npy(method_path)
        
        ssim_val = ssim_similarity_map(tcpa_map, method_map)
        cosine_val = cosine_similarity_map(tcpa_map, method_map)
        spearman_val = spearman_similarity_map(tcpa_map, method_map)
        top10_val = topk_overlap(tcpa_map, method_map, top_percent=10)
        
        composite = np.nanmean([
            safe_01(ssim_val),
            (cosine_val + 1) / 2 if not pd.isna(cosine_val) else np.nan,
            spearman_to_01(spearman_val),
            safe_01(top10_val)
        ])
        
        tcpa_similarity_rows.append({
            "xai_case_id": case_id,
            "outcome": outcome,
            "compared_method": method,
            "SSIM": ssim_val,
            "Cosine": cosine_val,
            "Spearman": spearman_val,
            "Top10_Overlap": top10_val,
            "Composite_Similarity": composite,
            "Composite_Similarity_Percentage": composite * 100
        })

tcpa_similarity_df = pd.DataFrame(tcpa_similarity_rows)

display(tcpa_similarity_df.head())
print("Rows:", tcpa_similarity_df.shape)
```
