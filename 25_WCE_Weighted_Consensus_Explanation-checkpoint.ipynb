```python
import pandas as pd
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

from scipy.stats import spearmanr

try:
    from skimage.metrics import structural_similarity as ssim
    SKIMAGE_AVAILABLE = True
    print("scikit-image available. SSIM will be calculated.")
except ImportError:
    SKIMAGE_AVAILABLE = False
    print("scikit-image not installed. Run: !pip install scikit-image")

# -----------------------------
# Project paths
# -----------------------------
PROJECT_ROOT = Path(r"C:\Users\anudi\Downloads\MSc_Research")

# Input files from previous steps
TAI_FILE = (
    PROJECT_ROOT
    / "Advanced_Framework_Analysis"
    / "TAI_Optimised_CTS"
    / "TAI_Optimised_CTS_Results.xlsx"
)

EAS_FILE = (
    PROJECT_ROOT
    / "Advanced_Framework_Analysis"
    / "Explanation_Agreement_Score"
    / "Explanation_Agreement_Score_Results.xlsx"
)

# Existing XAI maps
XAI_MAP_ROOT = PROJECT_ROOT / "xai_analysis" / "full_cohort" / "maps"
XAI_PILOT_MAP_ROOT = PROJECT_ROOT / "xai_analysis" / "xai_pilot" / "maps"

# Output folders
OUT_DIR = PROJECT_ROOT / "Advanced_Framework_Analysis" / "WCE_Weighted_Consensus_Explanation"
WCE_MAP_DIR = OUT_DIR / "wce_maps_npy"
WCE_PNG_DIR = OUT_DIR / "wce_maps_png"

OUT_DIR.mkdir(parents=True, exist_ok=True)
WCE_MAP_DIR.mkdir(parents=True, exist_ok=True)
WCE_PNG_DIR.mkdir(parents=True, exist_ok=True)

print("TAI file exists:", TAI_FILE.exists())
print("EAS file exists:", EAS_FILE.exists())
print("XAI map folder exists:", XAI_MAP_ROOT.exists())
print("Output folder:", OUT_DIR)
```

```python
analysis_df = pd.read_excel(TAI_FILE, sheet_name="Analysis_Data_With_TAI_CTS")

print("Shape:", analysis_df.shape)
display(analysis_df.head())

print("\nColumns:")
for c in analysis_df.columns:
    print(c)
```

```python
case_mapping = {
    "Case 1": "XAI_030",
    "Case 2": "XAI_003",
    "Case 3": "XAI_052",
    "Case 4": "XAI_032",
    "Case 5": "XAI_087",
    "Case 6": "XAI_066",
    "Case 7": "XAI_120",
    "Case 8": "XAI_099"
}

# Detect case column
possible_case_cols = ["xai_case_id_standard", "xai_case_id", "case_id", "case"]

case_col = None
for col in possible_case_cols:
    if col in analysis_df.columns:
        case_col = col
        break

if case_col is None:
    for col in analysis_df.columns:
        if "case" in str(col).lower():
            case_col = col
            break

if case_col is None:
    raise ValueError("Could not find case column.")

analysis_df["xai_case_id_standard"] = analysis_df[case_col].replace(case_mapping)

print("Detected case column:", case_col)
display(analysis_df[["xai_case_id_standard"]].drop_duplicates())
```

```python
# Detect method column
method_col = None
for col in analysis_df.columns:
    if "method" in str(col).lower():
        method_col = col
        break

if method_col is None:
    raise ValueError("Could not find method column.")

print("Detected method column:", method_col)
print(analysis_df[method_col].unique())

method_mapping = {
    "Method A": "Grad-CAM",
    "Method B": "Grad-CAM++",
    "Method C": "LIME",
    "Method D": "SHAP",
    "A": "Grad-CAM",
    "B": "Grad-CAM++",
    "C": "LIME",
    "D": "SHAP",
    "gradcam": "Grad-CAM",
    "grad-cam": "Grad-CAM",
    "GradCAM": "Grad-CAM",
    "gradcampp": "Grad-CAM++",
    "gradcam++": "Grad-CAM++",
    "Grad-CAM++": "Grad-CAM++",
    "lime": "LIME",
    "shap": "SHAP"
}

analysis_df["method_standard"] = analysis_df[method_col].replace(method_mapping)

display(analysis_df[["xai_case_id_standard", method_col, "method_standard"]].head(10))
print(analysis_df["method_standard"].unique())
```

```python
required_cols = [
    "xai_case_id_standard",
    "method_standard",
    "faithfulness",
    "stability"
]

missing = [c for c in required_cols if c not in analysis_df.columns]

if missing:
    raise ValueError(f"Missing required columns: {missing}")

# Convert to numeric
for col in ["faithfulness", "stability"]:
    analysis_df[col] = pd.to_numeric(analysis_df[col], errors="coerce")

# Make sure scores are 0-100
def to_percentage(series):
    s = pd.to_numeric(series, errors="coerce")
    if s.max(skipna=True) <= 1.5:
        return s * 100
    return s

analysis_df["faithfulness_pct"] = to_percentage(analysis_df["faithfulness"])
analysis_df["stability_pct"] = to_percentage(analysis_df["stability"])

display(analysis_df[[
    "xai_case_id_standard",
    "method_standard",
    "faithfulness_pct",
    "stability_pct"
]].head())
```

```python
METHOD_KEYS = {
    "Grad-CAM": "gradcam",
    "Grad-CAM++": "gradcampp",
    "LIME": "lime",
    "SHAP": "shap"
}

methods = list(METHOD_KEYS.keys())

def find_heatmap_file_exact(case_id, method):
    method_key = METHOD_KEYS[method]
    filename = f"{case_id}_{method_key}_normalised.npy"
    
    full_path = XAI_MAP_ROOT / filename
    pilot_path = XAI_PILOT_MAP_ROOT / filename
    
    if full_path.exists():
        return full_path
    
    if pilot_path.exists():
        return pilot_path
    
    return None
```

```python
def normalise_01(arr):
    arr = np.asarray(arr, dtype=np.float32)
    
    if arr.ndim == 3:
        arr = arr.mean(axis=2)
    
    min_val = np.nanmin(arr)
    max_val = np.nanmax(arr)
    
    if max_val - min_val < 1e-8:
        return np.zeros_like(arr, dtype=np.float32)
    
    return (arr - min_val) / (max_val - min_val)


def load_heatmap(path, target_size=(224, 224)):
    path = Path(path)
    
    if path.suffix.lower() == ".npy":
        arr = np.load(path)
        arr = np.squeeze(arr)
        
        if arr.ndim == 3:
            arr = arr.mean(axis=2)
        
        arr = normalise_01(arr)
        
        img = Image.fromarray((arr * 255).astype(np.uint8))
        img = img.resize(target_size, Image.BILINEAR)
        arr = np.asarray(img).astype(np.float32) / 255.0
        
        return normalise_01(arr)
    
    else:
        img = Image.open(path).convert("L")
        img = img.resize(target_size, Image.BILINEAR)
        arr = np.asarray(img).astype(np.float32) / 255.0
        
        return normalise_01(arr)


def save_heatmap_png(arr, path):
    arr = normalise_01(arr)
    img = Image.fromarray((arr * 255).astype(np.uint8))
    img.save(path)
```

```python
survey_cases = analysis_df["xai_case_id_standard"].drop_duplicates().tolist()

check_rows = []

for case_id in survey_cases:
    for method in methods:
        path = find_heatmap_file_exact(case_id, method)
        check_rows.append({
            "xai_case_id": case_id,
            "method": method,
            "file_found": path is not None,
            "file_path": str(path) if path else None
        })

map_check_df = pd.DataFrame(check_rows)
display(map_check_df)

missing = map_check_df[map_check_df["file_found"] == False]

if len(missing) > 0:
    print("Missing maps:")
    display(missing)
else:
    print("All maps found.")
```

```python
def safe_weight_normalise(scores):
    scores = np.asarray(scores, dtype=np.float32)
    scores = np.nan_to_num(scores, nan=0.0, posinf=0.0, neginf=0.0)
    scores = np.clip(scores, 0, None)
    
    total = scores.sum()
    
    if total <= 1e-8:
        return np.ones_like(scores) / len(scores)
    
    return scores / total


wce_weight_rows = []

for case_id in survey_cases:
    case_df = analysis_df[analysis_df["xai_case_id_standard"] == case_id].copy()
    
    for method in methods:
        method_row = case_df[case_df["method_standard"] == method]
        
        if len(method_row) == 0:
            continue
        
        row = method_row.iloc[0]
        
        faith = row["faithfulness_pct"]
        stab = row["stability_pct"]
        
        fs_score = faith * stab
        
        cts_score = row["CTS_TAI_optimised"] if "CTS_TAI_optimised" in row.index else np.nan
        tai_score = row["TAI_TAI_optimised"] if "TAI_TAI_optimised" in row.index else np.nan
        
        wce_weight_rows.append({
            "xai_case_id": case_id,
            "method": method,
            "faithfulness_pct": faith,
            "stability_pct": stab,
            "FS_score": fs_score,
            "CTS_TAI_optimised": cts_score,
            "TAI_TAI_optimised": tai_score
        })

wce_weights_long = pd.DataFrame(wce_weight_rows)

# Add normalised weights per case
final_weight_rows = []

for case_id, group in wce_weights_long.groupby("xai_case_id"):
    group = group.copy()
    
    group["WCE_FS_weight"] = safe_weight_normalise(group["FS_score"].values)
    
    if group["CTS_TAI_optimised"].notna().sum() == len(group):
        group["WCE_CTS_weight"] = safe_weight_normalise(group["CTS_TAI_optimised"].values)
    else:
        group["WCE_CTS_weight"] = np.nan
    
    if group["TAI_TAI_optimised"].notna().sum() == len(group):
        group["WCE_TAI_weight"] = safe_weight_normalise(group["TAI_TAI_optimised"].values)
    else:
        group["WCE_TAI_weight"] = np.nan
    
    final_weight_rows.append(group)

wce_weights_df = pd.concat(final_weight_rows, ignore_index=True)

display(wce_weights_df)
```

```python
def generate_wce_map(case_id, weight_col):
    maps = []
    weights = []
    
    case_weights = wce_weights_df[wce_weights_df["xai_case_id"] == case_id]
    
    for _, row in case_weights.iterrows():
        method = row["method"]
        weight = row[weight_col]
        
        if pd.isna(weight):
            continue
        
        path = find_heatmap_file_exact(case_id, method)
        
        if path is None:
            continue
        
        heatmap = load_heatmap(path)
        maps.append(heatmap)
        weights.append(weight)
    
    if len(maps) == 0:
        return None
    
    weights = safe_weight_normalise(weights)
    
    wce = np.zeros_like(maps[0], dtype=np.float32)
    
    for m, w in zip(maps, weights):
        wce += m * w
    
    return normalise_01(wce)


wce_map_rows = []

wce_versions = {
    "WCE_FS": "WCE_FS_weight",
    "WCE_CTS": "WCE_CTS_weight",
    "WCE_TAI": "WCE_TAI_weight"
}

for case_id in survey_cases:
    for version_name, weight_col in wce_versions.items():
        if weight_col not in wce_weights_df.columns:
            continue
        
        wce_map = generate_wce_map(case_id, weight_col)
        
        if wce_map is None:
            continue
        
        npy_path = WCE_MAP_DIR / f"{case_id}_{version_name}_normalised.npy"
        png_path = WCE_PNG_DIR / f"{case_id}_{version_name}_normalised.png"
        
        np.save(npy_path, wce_map)
        save_heatmap_png(wce_map, png_path)
        
        wce_map_rows.append({
            "xai_case_id": case_id,
            "WCE_version": version_name,
            "npy_path": str(npy_path),
            "png_path": str(png_path)
        })

wce_maps_df = pd.DataFrame(wce_map_rows)

display(wce_maps_df)
print("WCE maps generated:", len(wce_maps_df))
```

```python
def cosine_similarity_map(a, b):
    a = a.flatten().astype(np.float32)
    b = b.flatten().astype(np.float32)
    
    denom = np.linalg.norm(a) * np.linalg.norm(b)
    
    if denom < 1e-8:
        return np.nan
    
    return float(np.clip(np.dot(a, b) / denom, -1, 1))


def spearman_similarity_map(a, b):
    a = a.flatten()
    b = b.flatten()
    
    if np.std(a) < 1e-8 or np.std(b) < 1e-8:
        return np.nan
    
    rho, p = spearmanr(a, b)
    return float(rho)


def topk_overlap(a, b, top_percent=10):
    a = a.flatten()
    b = b.flatten()
    
    n = len(a)
    k = int(np.ceil(n * top_percent / 100))
    
    top_a = np.argsort(a)[-k:]
    top_b = np.argsort(b)[-k:]
    
    return float(len(set(top_a).intersection(set(top_b))) / k)


def ssim_similarity_map(a, b):
    if not SKIMAGE_AVAILABLE:
        return np.nan
    return float(ssim(a, b, data_range=1.0))


def spearman_to_01(rho):
    if pd.isna(rho):
        return np.nan
    return (rho + 1) / 2


def safe_01(value):
    if pd.isna(value):
        return np.nan
    return float(np.clip(value, 0, 1))
```

```python
wce_comparison_rows = []

for case_id in survey_cases:
    for wce_version in wce_versions.keys():
        wce_path = WCE_MAP_DIR / f"{case_id}_{wce_version}_normalised.npy"
        
        if not wce_path.exists():
            continue
        
        wce_map = load_heatmap(wce_path)
        
        for method in methods:
            method_path = find_heatmap_file_exact(case_id, method)
            
            if method_path is None:
                continue
            
            method_map = load_heatmap(method_path)
            
            ssim_val = ssim_similarity_map(wce_map, method_map)
            cosine_val = cosine_similarity_map(wce_map, method_map)
            spearman_val = spearman_similarity_map(wce_map, method_map)
            top10_val = topk_overlap(wce_map, method_map, top_percent=10)
            
            composite = np.nanmean([
                safe_01(ssim_val),
                (cosine_val + 1) / 2 if not pd.isna(cosine_val) else np.nan,
                spearman_to_01(spearman_val),
                safe_01(top10_val)
            ])
            
            wce_comparison_rows.append({
                "xai_case_id": case_id,
                "WCE_version": wce_version,
                "compared_method": method,
                "SSIM": ssim_val,
                "Cosine": cosine_val,
                "Spearman": spearman_val,
                "Top10_Overlap": top10_val,
                "Composite_Similarity": composite,
                "Composite_Similarity_Percentage": composite * 100
            })

wce_similarity_df = pd.DataFrame(wce_comparison_rows)

display(wce_similarity_df.head())

wce_similarity_summary = wce_similarity_df.groupby(["WCE_version", "compared_method"]).agg(
    mean_similarity=("Composite_Similarity_Percentage", "mean"),
    median_similarity=("Composite_Similarity_Percentage", "median"),
    sd_similarity=("Composite_Similarity_Percentage", "std"),
    n=("Composite_Similarity_Percentage", "count")
).reset_index()

display(wce_similarity_summary)
```

```python
# Load previous EAS case-level result
case_eas_df = pd.read_excel(EAS_FILE, sheet_name="Case_Level_EAS")

# Standardise ID just in case
case_eas_df["xai_case_id"] = case_eas_df["xai_case_id"].replace(case_mapping)

# WCE mean similarity to the four original methods
wce_case_agreement = wce_similarity_df.groupby(["xai_case_id", "WCE_version"]).agg(
    WCE_mean_similarity=("Composite_Similarity_Percentage", "mean"),
    WCE_min_similarity=("Composite_Similarity_Percentage", "min"),
    WCE_max_similarity=("Composite_Similarity_Percentage", "max")
).reset_index()

wce_eas_comparison = wce_case_agreement.merge(
    case_eas_df[["xai_case_id", "outcome", "EAS_Percentage"]],
    on="xai_case_id",
    how="left"
)

wce_eas_comparison["WCE_minus_original_EAS"] = (
    wce_eas_comparison["WCE_mean_similarity"] - wce_eas_comparison["EAS_Percentage"]
)

display(wce_eas_comparison)

wce_eas_summary = wce_eas_comparison.groupby("WCE_version").agg(
    mean_original_EAS=("EAS_Percentage", "mean"),
    mean_WCE_similarity=("WCE_mean_similarity", "mean"),
    mean_improvement_over_EAS=("WCE_minus_original_EAS", "mean"),
    median_improvement_over_EAS=("WCE_minus_original_EAS", "median"),
    n=("xai_case_id", "count")
).reset_index()

display(wce_eas_summary)
```

```python
wce_by_outcome = wce_eas_comparison.groupby(["WCE_version", "outcome"]).agg(
    mean_original_EAS=("EAS_Percentage", "mean"),
    mean_WCE_similarity=("WCE_mean_similarity", "mean"),
    mean_improvement=("WCE_minus_original_EAS", "mean"),
    n=("xai_case_id", "count")
).reset_index()

display(wce_by_outcome)
```

```python
dominant_rows = []

for case_id, group in wce_weights_df.groupby("xai_case_id"):
    for weight_col in ["WCE_FS_weight", "WCE_CTS_weight", "WCE_TAI_weight"]:
        if weight_col not in group.columns:
            continue
        
        temp = group[["method", weight_col]].dropna()
        
        if len(temp) == 0:
            continue
        
        dominant = temp.loc[temp[weight_col].idxmax()]
        
        dominant_rows.append({
            "xai_case_id": case_id,
            "WCE_version": weight_col.replace("_weight", ""),
            "dominant_method": dominant["method"],
            "dominant_weight": dominant[weight_col]
        })

dominant_method_df = pd.DataFrame(dominant_rows)

display(dominant_method_df)

dominant_summary = dominant_method_df.groupby(["WCE_version", "dominant_method"]).size().reset_index(name="count")
display(dominant_summary)
```
