```python
import pandas as pd
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import LeaveOneOut
from sklearn.metrics import mean_absolute_error, r2_score

import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

# -----------------------------
# Project paths
# -----------------------------
PROJECT_ROOT = Path(r"C:\Users\anudi\Downloads\MSc_Research")

# Your existing CTS-HTS alignment workbook
ALIGNMENT_FILE = PROJECT_ROOT / "CTS_HTS_Trust_Alignment_Results.xlsx"

# Output folder
OUT_DIR = PROJECT_ROOT / "Advanced_Framework_Analysis" / "Regression_Derived_CTS"
OUT_DIR.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
print("Alignment file exists:", ALIGNMENT_FILE.exists())
print("Output folder:", OUT_DIR)
```

```python
# -----------------------------
# Inspect workbook sheets
# -----------------------------
xls = pd.ExcelFile(ALIGNMENT_FILE)
print("Available sheets:")
for s in xls.sheet_names:
    print("-", s)

# -----------------------------
# Try to find the best sheet automatically
# -----------------------------
candidate_sheets = xls.sheet_names

required_keywords = [
    "confidence",
    "robust",
    "faith",
    "stability",
    "hts"
]

best_sheet = None
best_score = -1

for sheet in candidate_sheets:
    temp = pd.read_excel(ALIGNMENT_FILE, sheet_name=sheet)
    cols_joined = " ".join([str(c).lower() for c in temp.columns])
    score = sum(k in cols_joined for k in required_keywords)
    
    if score > best_score:
        best_score = score
        best_sheet = sheet

print("\nSelected sheet:", best_sheet)
df = pd.read_excel(ALIGNMENT_FILE, sheet_name=best_sheet)

print("\nShape:", df.shape)
display(df.head())
print("\nColumns:")
for c in df.columns:
    print(c)
```

```python
# -----------------------------
# Helper function to find columns
# -----------------------------
def find_col(df, keywords, required=True):
    """
    Finds the first column containing all keywords.
    """
    cols = list(df.columns)
    lower_cols = {c: str(c).lower() for c in cols}
    
    for c, lc in lower_cols.items():
        if all(k.lower() in lc for k in keywords):
            return c
    
    if required:
        raise ValueError(f"Could not find column with keywords: {keywords}")
    return None


# -----------------------------
# Detect important columns
# -----------------------------
case_col = find_col(df, ["case"], required=False)
xai_case_col = find_col(df, ["xai"], required=False)
method_col = find_col(df, ["method"], required=True)
outcome_col = find_col(df, ["outcome"], required=False)

confidence_col = find_col(df, ["confidence"], required=True)
robustness_col = find_col(df, ["robust"], required=True)
faithfulness_col = find_col(df, ["faith"], required=True)
stability_col = find_col(df, ["stability"], required=True)

# HTS column can have different names
try:
    hts_col = find_col(df, ["hts"], required=True)
except:
    hts_col = find_col(df, ["human", "trust"], required=True)

# Existing baseline CTS column if available
cts_baseline_col = find_col(df, ["cts"], required=False)

print("Detected columns:")
print("case_col:", case_col)
print("xai_case_col:", xai_case_col)
print("method_col:", method_col)
print("outcome_col:", outcome_col)
print("confidence_col:", confidence_col)
print("robustness_col:", robustness_col)
print("faithfulness_col:", faithfulness_col)
print("stability_col:", stability_col)
print("hts_col:", hts_col)
print("cts_baseline_col:", cts_baseline_col)
```

```python
# -----------------------------
# Convert scores to percentage scale
# -----------------------------
def to_percentage(series):
    """
    Converts a score column to 0-100 scale.
    If max <= 1.5, assumes the column is 0-1 and multiplies by 100.
    """
    s = pd.to_numeric(series, errors="coerce")
    if s.max(skipna=True) <= 1.5:
        return s * 100
    return s


analysis_df = pd.DataFrame()

if case_col:
    analysis_df["case_id"] = df[case_col]

if xai_case_col:
    analysis_df["xai_case_id"] = df[xai_case_col]

analysis_df["method"] = df[method_col]

if outcome_col:
    analysis_df["outcome"] = df[outcome_col]

analysis_df["confidence"] = to_percentage(df[confidence_col])
analysis_df["robustness"] = to_percentage(df[robustness_col])
analysis_df["faithfulness"] = to_percentage(df[faithfulness_col])
analysis_df["stability"] = to_percentage(df[stability_col])
analysis_df["HTS"] = to_percentage(df[hts_col])

# Baseline CTS
if cts_baseline_col:
    analysis_df["CTS_baseline"] = to_percentage(df[cts_baseline_col])
else:
    analysis_df["CTS_baseline"] = analysis_df[
        ["confidence", "robustness", "faithfulness", "stability"]
    ].mean(axis=1)

# Remove incomplete rows
analysis_df = analysis_df.dropna(
    subset=["confidence", "robustness", "faithfulness", "stability", "HTS"]
).reset_index(drop=True)

print("Clean analysis dataset shape:", analysis_df.shape)
display(analysis_df.head())

print("\nScore ranges:")
display(analysis_df[["confidence", "robustness", "faithfulness", "stability", "CTS_baseline", "HTS"]].describe())
```

```python
def add_alignment_metrics(data, cts_col, label):
    temp = data.copy()
    temp["CTS_version"] = label
    temp["CTS"] = temp[cts_col]
    temp["TrustGap"] = temp["HTS"] - temp["CTS"]
    temp["Abs_TrustGap"] = temp["TrustGap"].abs()
    temp["TAI"] = 100 - temp["Abs_TrustGap"]
    return temp


baseline_rows = add_alignment_metrics(
    analysis_df,
    cts_col="CTS_baseline",
    label="Baseline equal-weight CTS"
)

baseline_summary = baseline_rows.agg({
    "CTS": ["mean", "median", "std", "min", "max"],
    "HTS": ["mean", "median", "std", "min", "max"],
    "TrustGap": ["mean", "median", "std", "min", "max"],
    "Abs_TrustGap": ["mean", "median", "std", "min", "max"],
    "TAI": ["mean", "median", "std", "min", "max"]
})

display(baseline_summary)
```

```python
predictor_cols = ["confidence", "robustness", "faithfulness", "stability"]
target_col = "HTS"

X = analysis_df[predictor_cols].copy()
y = analysis_df[target_col].copy()

# -----------------------------
# Unstandardised OLS
# -----------------------------
X_const = sm.add_constant(X)
ols_model = sm.OLS(y, X_const).fit()

print("Unstandardised OLS regression:")
print(ols_model.summary())

# -----------------------------
# Standardised OLS
# -----------------------------
scaler_X = StandardScaler()
scaler_y = StandardScaler()

X_z = pd.DataFrame(
    scaler_X.fit_transform(X),
    columns=predictor_cols
)

y_z = pd.Series(
    scaler_y.fit_transform(y.values.reshape(-1, 1)).ravel(),
    name="HTS_z"
)

X_z_const = sm.add_constant(X_z)
std_model = sm.OLS(y_z, X_z_const).fit()

print("\nStandardised OLS regression:")
print(std_model.summary())
```

```python
vif_df = pd.DataFrame()
vif_df["component"] = predictor_cols
vif_df["VIF"] = [
    variance_inflation_factor(X_z.values, i)
    for i in range(X_z.shape[1])
]

display(vif_df)
```

```python
# Extract standardised beta coefficients
std_betas = std_model.params.drop("const")

weights_df = pd.DataFrame({
    "component": std_betas.index,
    "standardised_beta": std_betas.values
})

# Use only positive beta values for CTS weights
# Negative beta means that component did not align positively with HTS in this sample.
weights_df["positive_beta"] = weights_df["standardised_beta"].clip(lower=0)

if weights_df["positive_beta"].sum() > 0:
    weights_df["regression_weight"] = weights_df["positive_beta"] / weights_df["positive_beta"].sum()
    weight_method = "positive standardised betas"
else:
    # fallback if all betas are zero/negative
    weights_df["absolute_beta"] = weights_df["standardised_beta"].abs()
    weights_df["regression_weight"] = weights_df["absolute_beta"] / weights_df["absolute_beta"].sum()
    weight_method = "absolute standardised betas fallback"

print("Weight derivation method:", weight_method)
display(weights_df)

# Dictionary of weights
reg_weights = dict(zip(weights_df["component"], weights_df["regression_weight"]))

print("\nRegression-derived CTS weights:")
for k, v in reg_weights.items():
    print(f"{k}: {v:.4f}")
```

```python
analysis_df["CTS_regression_derived"] = (
    analysis_df["confidence"] * reg_weights["confidence"] +
    analysis_df["robustness"] * reg_weights["robustness"] +
    analysis_df["faithfulness"] * reg_weights["faithfulness"] +
    analysis_df["stability"] * reg_weights["stability"]
)

analysis_df["TrustGap_regression"] = analysis_df["HTS"] - analysis_df["CTS_regression_derived"]
analysis_df["Abs_TrustGap_regression"] = analysis_df["TrustGap_regression"].abs()
analysis_df["TAI_regression"] = 100 - analysis_df["Abs_TrustGap_regression"]

display(analysis_df.head())
```

```python
comparison_rows = pd.concat([
    add_alignment_metrics(analysis_df, "CTS_baseline", "Baseline equal-weight CTS"),
    add_alignment_metrics(analysis_df, "CTS_regression_derived", "Regression-derived CTS")
], ignore_index=True)

comparison_summary = comparison_rows.groupby("CTS_version").agg(
    mean_CTS=("CTS", "mean"),
    median_CTS=("CTS", "median"),
    mean_HTS=("HTS", "mean"),
    mean_TrustGap=("TrustGap", "mean"),
    median_TrustGap=("TrustGap", "median"),
    mean_abs_TrustGap=("Abs_TrustGap", "mean"),
    median_abs_TrustGap=("Abs_TrustGap", "median"),
    mean_TAI=("TAI", "mean"),
    median_TAI=("TAI", "median"),
    min_TAI=("TAI", "min"),
    max_TAI=("TAI", "max"),
    n=("TAI", "count")
).reset_index()

display(comparison_summary)
```

```python
# By method
method_comparison = comparison_rows.groupby(["CTS_version", "method"]).agg(
    mean_CTS=("CTS", "mean"),
    mean_HTS=("HTS", "mean"),
    mean_TrustGap=("TrustGap", "mean"),
    mean_abs_TrustGap=("Abs_TrustGap", "mean"),
    mean_TAI=("TAI", "mean"),
    n=("TAI", "count")
).reset_index()

display(method_comparison)

# By outcome, if outcome column exists
if "outcome" in comparison_rows.columns:
    outcome_comparison = comparison_rows.groupby(["CTS_version", "outcome"]).agg(
        mean_CTS=("CTS", "mean"),
        mean_HTS=("HTS", "mean"),
        mean_TrustGap=("TrustGap", "mean"),
        mean_abs_TrustGap=("Abs_TrustGap", "mean"),
        mean_TAI=("TAI", "mean"),
        n=("TAI", "count")
    ).reset_index()

    display(outcome_comparison)
else:
    outcome_comparison = pd.DataFrame()
```

```python
loo = LeaveOneOut()

actual = []
predicted = []

for train_idx, test_idx in loo.split(X):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]
    
    model = sm.OLS(y_train, sm.add_constant(X_train)).fit()
    
    pred = model.predict(sm.add_constant(X_test, has_constant="add"))
    
    actual.append(float(y_test.iloc[0]))
    predicted.append(float(pred.iloc[0]))

loo_results = pd.DataFrame({
    "HTS_actual": actual,
    "HTS_predicted": predicted
})

loo_mae = mean_absolute_error(loo_results["HTS_actual"], loo_results["HTS_predicted"])

try:
    loo_r2 = r2_score(loo_results["HTS_actual"], loo_results["HTS_predicted"])
except:
    loo_r2 = np.nan

print("Leave-one-out MAE:", loo_mae)
print("Leave-one-out R²:", loo_r2)

display(loo_results.head())
```

```python
N_BOOT = 1000
rng = np.random.default_rng(42)

bootstrap_weights = []

for i in range(N_BOOT):
    sample_idx = rng.choice(analysis_df.index, size=len(analysis_df), replace=True)
    boot_df = analysis_df.loc[sample_idx].reset_index(drop=True)
    
    X_boot = boot_df[predictor_cols]
    y_boot = boot_df[target_col]
    
    # Standardise within bootstrap sample
    X_boot_z = pd.DataFrame(
        StandardScaler().fit_transform(X_boot),
        columns=predictor_cols
    )
    
    y_boot_z = pd.Series(
        StandardScaler().fit_transform(y_boot.values.reshape(-1, 1)).ravel()
    )
    
    try:
        boot_model = sm.OLS(y_boot_z, sm.add_constant(X_boot_z)).fit()
        boot_betas = boot_model.params.drop("const")
        
        positive = boot_betas.clip(lower=0)
        
        if positive.sum() > 0:
            boot_w = positive / positive.sum()
        else:
            absolute = boot_betas.abs()
            boot_w = absolute / absolute.sum()
        
        bootstrap_weights.append(boot_w.values)
    except:
        continue

bootstrap_weights_df = pd.DataFrame(
    bootstrap_weights,
    columns=predictor_cols
)

bootstrap_weight_summary = bootstrap_weights_df.agg(["mean", "std"]).T
bootstrap_weight_summary["ci_lower_2_5"] = bootstrap_weights_df.quantile(0.025)
bootstrap_weight_summary["ci_upper_97_5"] = bootstrap_weights_df.quantile(0.975)
bootstrap_weight_summary = bootstrap_weight_summary.reset_index().rename(columns={"index": "component"})

display(bootstrap_weight_summary)
```
