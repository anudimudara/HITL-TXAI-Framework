```python
def calculate_ece(
    y_true,
    y_prob,
    n_bins=10
):
    y_true = np.asarray(y_true, dtype=int)
    y_prob = np.asarray(y_prob, dtype=float)

    bin_edges = np.linspace(
        0.0,
        1.0,
        n_bins + 1
    )

    bin_numbers = np.digitize(
        y_prob,
        bin_edges[1:-1],
        right=True
    )

    total_ece = 0.0
    rows = []

    for bin_index in range(n_bins):

        mask = bin_numbers == bin_index
        count = int(mask.sum())

        if count == 0:
            continue

        mean_probability = float(
            y_prob[mask].mean()
        )

        observed_frequency = float(
            y_true[mask].mean()
        )

        weight = count / len(y_true)

        absolute_error = abs(
            observed_frequency
            - mean_probability
        )

        total_ece += (
            weight * absolute_error
        )

        rows.append({
            "bin": bin_index + 1,
            "lower_boundary": bin_edges[bin_index],
            "upper_boundary": bin_edges[bin_index + 1],
            "count": count,
            "mean_probability": mean_probability,
            "observed_frequency": observed_frequency,
            "absolute_error": absolute_error
        })

    return float(total_ece), pd.DataFrame(rows)


print("ECE function created.")
```

```python
class TemperatureScaler(nn.Module):

    def __init__(self):
        super().__init__()

        self.log_temperature = nn.Parameter(
            torch.zeros(1)
        )

    def forward(self, logits):

        temperature = torch.exp(
            self.log_temperature
        )

        return logits / temperature

    def get_temperature(self):

        return float(
            torch.exp(
                self.log_temperature
            ).detach().cpu().item()
        )


print("Temperature-scaling class created.")
```

```python
def fit_temperature(
    validation_logits,
    validation_labels
):
    validation_logits = (
        validation_logits
        .detach()
        .float()
        .cpu()
    )

    validation_labels = (
        validation_labels
        .detach()
        .long()
        .cpu()
    )

    temperature_model = (
        TemperatureScaler()
        .cpu()
    )

    criterion = nn.CrossEntropyLoss()

    optimizer = optim.LBFGS(
        temperature_model.parameters(),
        lr=0.1,
        max_iter=100,
        line_search_fn="strong_wolfe"
    )

    nll_before = criterion(
        validation_logits,
        validation_labels
    ).item()

    def closure():

        optimizer.zero_grad()

        scaled_logits = temperature_model(
            validation_logits
        )

        loss = criterion(
            scaled_logits,
            validation_labels
        )

        loss.backward()

        return loss

    optimizer.step(closure)

    with torch.no_grad():

        scaled_logits = temperature_model(
            validation_logits
        )

        nll_after = criterion(
            scaled_logits,
            validation_labels
        ).item()

    temperature = (
        temperature_model.get_temperature()
    )

    return (
        temperature_model,
        temperature,
        nll_before,
        nll_after
    )


print("Temperature-fitting function created.")
```

```python
def calculate_calibration_metrics(
    logits,
    labels,
    temperature=1.0
):
    logits = logits.detach().float().cpu()
    labels = labels.detach().long().cpu()

    scaled_logits = logits / float(temperature)

    probabilities = torch.softmax(
        scaled_logits,
        dim=1
    )

    pneumonia_probabilities = (
        probabilities[:, 1]
        .numpy()
    )

    predictions = (
        probabilities.argmax(dim=1)
        .numpy()
    )

    y_true = labels.numpy()

    clipped_probabilities = np.clip(
        pneumonia_probabilities,
        1e-7,
        1 - 1e-7
    )

    ece, calibration_table = (
        calculate_ece(
            y_true,
            pneumonia_probabilities,
            n_bins=10
        )
    )

    metrics = {
        "temperature": float(temperature),

        "accuracy": float(
            accuracy_score(
                y_true,
                predictions
            )
        ),

        "roc_auc": float(
            roc_auc_score(
                y_true,
                pneumonia_probabilities
            )
        ),

        "average_precision": float(
            average_precision_score(
                y_true,
                pneumonia_probabilities
            )
        ),

        "brier_score": float(
            brier_score_loss(
                y_true,
                pneumonia_probabilities
            )
        ),

        "log_loss": float(
            log_loss(
                y_true,
                np.column_stack([
                    1 - clipped_probabilities,
                    clipped_probabilities
                ]),
                labels=[0, 1]
            )
        ),

        "ece_10_bins": float(ece)
    }

    return (
        metrics,
        pneumonia_probabilities,
        predictions,
        calibration_table
    )


print("Calibration metric function created.")
```

```python
def collect_validation_logits(
    model,
    data_loader,
    device,
    description
):
    model.eval()

    logits_list = []
    labels_list = []

    with torch.no_grad():

        for images, labels in tqdm(
            data_loader,
            desc=description
        ):

            images = images.to(device)

            logits = model(images)

            logits_list.append(
                logits.cpu()
            )

            labels_list.append(
                labels.cpu()
            )

    all_logits = torch.cat(
        logits_list,
        dim=0
    )

    all_labels = torch.cat(
        labels_list,
        dim=0
    )

    assert all_logits.shape == (
        len(val_dataset),
        2
    )

    assert all_labels.shape[0] == (
        len(val_dataset)
    )

    return all_logits, all_labels


print("Logit collection function created.")
```

```python
def plot_calibration_before_after(
    y_true,
    probability_before,
    probability_after,
    model_name,
    output_path
):
    observed_before, predicted_before = (
        calibration_curve(
            y_true,
            probability_before,
            n_bins=10,
            strategy="uniform"
        )
    )

    observed_after, predicted_after = (
        calibration_curve(
            y_true,
            probability_after,
            n_bins=10,
            strategy="uniform"
        )
    )

    plt.figure(figsize=(7, 6))

    plt.plot(
        predicted_before,
        observed_before,
        marker="o",
        label="Before temperature scaling"
    )

    plt.plot(
        predicted_after,
        observed_after,
        marker="o",
        label="After temperature scaling"
    )

    plt.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        label="Perfect calibration"
    )

    plt.xlabel("Mean Predicted Probability")
    plt.ylabel("Observed Pneumonia Frequency")

    plt.title(
        f"{model_name} Validation Calibration"
    )

    plt.legend()
    plt.grid(alpha=0.3)
    plt.tight_layout()

    plt.savefig(
        output_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


plot_calibration_before_after(
    y_true=resnet_val_labels.numpy(),
    probability_before=resnet_prob_before,
    probability_after=resnet_prob_after,
    model_name="ResNet50",
    output_path=os.path.join(
        CALIBRATION_RESULTS_DIR,
        "resnet50_calibration_before_after.png"
    )
)

plot_calibration_before_after(
    y_true=densenet_val_labels.numpy(),
    probability_before=densenet_prob_before,
    probability_after=densenet_prob_after,
    model_name="DenseNet121",
    output_path=os.path.join(
        CALIBRATION_RESULTS_DIR,
        "densenet121_calibration_before_after.png"
    )
)
```
