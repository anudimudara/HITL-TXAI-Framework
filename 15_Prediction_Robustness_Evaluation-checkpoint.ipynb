```python
RANDOM_SEED = 42

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)
torch.manual_seed(RANDOM_SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", device)
print("Random seed:", RANDOM_SEED)
```

```python
with open(temperature_path, "r", encoding="utf-8") as file:
    temperature_data = json.load(file)

print("Temperature JSON:")
print(temperature_data)

possible_temperature_keys = [
    "temperature",
    "optimal_temperature",
    "best_temperature",
    "T"
]

TEMPERATURE = None

for key in possible_temperature_keys:
    if key in temperature_data:
        TEMPERATURE = float(temperature_data[key])
        break

if TEMPERATURE is None:
    raise ValueError("Temperature value not found in JSON file.")

MODEL_DECISION_THRESHOLD = 0.35

print("Loaded temperature:", TEMPERATURE)
print("Model decision threshold:", MODEL_DECISION_THRESHOLD)
```

```python
def extract_state_dict(checkpoint):
    if isinstance(checkpoint, dict):
        for key in ["model_state_dict", "state_dict", "model"]:
            if key in checkpoint:
                return checkpoint[key]
    return checkpoint


def clean_state_dict_keys(state_dict):
    cleaned = {}

    for key, value in state_dict.items():
        new_key = key

        if new_key.startswith("module."):
            new_key = new_key.replace("module.", "", 1)

        cleaned[new_key] = value

    return cleaned


checkpoint = torch.load(resnet_checkpoint_path, map_location=device)
state_dict = extract_state_dict(checkpoint)
state_dict = clean_state_dict_keys(state_dict)

if "fc.weight" not in state_dict:
    raise ValueError("Could not find fc.weight in checkpoint.")

output_features = state_dict["fc.weight"].shape[0]

print("Detected ResNet50 output features:", output_features)

model = models.resnet50(weights=None)
model.fc = nn.Linear(model.fc.in_features, output_features)

model.load_state_dict(state_dict, strict=True)
model = model.to(device)
model.eval()

print("Frozen ResNet50 loaded successfully.")
```

```python
IMAGE_SIZE = 224

model_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("Preprocessing transform defined.")
```

```python
print("Searching for DICOM files...")

dicom_files = list(BASE_DIR.rglob("*.dcm"))

print("Number of DICOM files found:", len(dicom_files))

dicom_index = {}

for path in dicom_files:
    dicom_index[path.stem] = path

print("DICOM index created.")
```

```python
def read_dicom_as_rgb(dicom_path):
    dicom = pydicom.dcmread(str(dicom_path))
    image = dicom.pixel_array.astype(np.float32)

    if getattr(dicom, "PhotometricInterpretation", "") == "MONOCHROME1":
        image = np.max(image) - image

    image = image - np.min(image)

    if np.max(image) > 0:
        image = image / np.max(image)

    image = (image * 255).astype(np.uint8)

    rgb_image = np.stack([image, image, image], axis=-1)

    return rgb_image


def load_rgb_image_from_case(case_row):
    patient_id = str(case_row["patientId"])

    if patient_id not in dicom_index:
        raise FileNotFoundError(f"DICOM image not found for patientId: {patient_id}")

    dicom_path = dicom_index[patient_id]

    rgb_image = read_dicom_as_rgb(dicom_path)

    pil_image = Image.fromarray(rgb_image).convert("RGB")
    pil_image = pil_image.resize((IMAGE_SIZE, IMAGE_SIZE))

    rgb_resized = np.array(pil_image)

    return dicom_path, rgb_resized
```

```python
def predict_calibrated_probabilities_from_rgb_batch(rgb_images):
    """
    Input:
    - list of RGB images, each 224 x 224 x 3

    Output:
    - calibrated Lung Opacity probabilities
    """

    model.eval()

    tensors = []

    for image in rgb_images:
        image = np.asarray(image).astype(np.uint8)

        pil_image = Image.fromarray(image).convert("RGB")
        tensor = model_transform(pil_image)

        tensors.append(tensor)

    batch_tensor = torch.stack(tensors).to(device)

    with torch.no_grad():
        logits = model(batch_tensor)
        scaled_logits = logits / TEMPERATURE

        if scaled_logits.shape[1] == 1:
            probs = torch.sigmoid(scaled_logits)[:, 0]

        elif scaled_logits.shape[1] == 2:
            probs = torch.softmax(scaled_logits, dim=1)[:, 1]

        else:
            raise ValueError(f"Unexpected model output shape: {scaled_logits.shape}")

    return probs.detach().cpu().numpy()
```

```python
verification_rows = []

for _, row in xai_cohort.head(10).iterrows():
    _, rgb_image = load_rgb_image_from_case(row)

    reproduced_probability = predict_calibrated_probabilities_from_rgb_batch(
        [rgb_image]
    )[0]

    saved_probability = float(row["calibrated_probability"])

    reproduced_prediction = int(reproduced_probability >= MODEL_DECISION_THRESHOLD)
    saved_prediction = int(saved_probability >= MODEL_DECISION_THRESHOLD)

    verification_rows.append({
        "xai_case_id": row["xai_case_id"],
        "outcome": row["outcome"],
        "saved_probability": saved_probability,
        "reproduced_probability": float(reproduced_probability),
        "absolute_difference": abs(saved_probability - float(reproduced_probability)),
        "saved_prediction": saved_prediction,
        "reproduced_prediction": reproduced_prediction,
        "prediction_match": saved_prediction == reproduced_prediction
    })

verification_df = pd.DataFrame(verification_rows)

display(verification_df)

print("Maximum probability difference:", verification_df["absolute_difference"].max())
print("All predictions match:", verification_df["prediction_match"].all())
```

```python
def apply_gaussian_noise(rgb_image, noise_std=8):
    noise = np.random.normal(
        loc=0,
        scale=noise_std,
        size=rgb_image.shape
    )

    noisy = rgb_image.astype(np.float32) + noise

    return np.clip(noisy, 0, 255).astype(np.uint8)


def apply_brightness_change(rgb_image, factor=1.10):
    bright = rgb_image.astype(np.float32) * factor

    return np.clip(bright, 0, 255).astype(np.uint8)


def apply_contrast_change(rgb_image, factor=1.10):
    mean = np.mean(
        rgb_image,
        axis=(0, 1),
        keepdims=True
    )

    contrast = (rgb_image.astype(np.float32) - mean) * factor + mean

    return np.clip(contrast, 0, 255).astype(np.uint8)


def apply_translation(rgb_image, shift_x=5, shift_y=5):
    h, w = rgb_image.shape[:2]

    matrix = np.float32([
        [1, 0, shift_x],
        [0, 1, shift_y]
    ])

    translated = cv2.warpAffine(
        rgb_image,
        matrix,
        (w, h),
        borderMode=cv2.BORDER_REFLECT
    )

    return translated.astype(np.uint8)


def apply_rotation(rgb_image, angle=5):
    h, w = rgb_image.shape[:2]
    center = (w // 2, h // 2)

    matrix = cv2.getRotationMatrix2D(
        center,
        angle,
        1.0
    )

    rotated = cv2.warpAffine(
        rgb_image,
        matrix,
        (w, h),
        borderMode=cv2.BORDER_REFLECT
    )

    return rotated.astype(np.uint8)
```

```python
PERTURBATIONS = {
    "gaussian_noise_std8": lambda img: apply_gaussian_noise(img, noise_std=8),
    "brightness_plus_10": lambda img: apply_brightness_change(img, factor=1.10),
    "contrast_plus_10": lambda img: apply_contrast_change(img, factor=1.10),
    "translation_5px": lambda img: apply_translation(img, shift_x=5, shift_y=5),
    "rotation_5deg": lambda img: apply_rotation(img, angle=5)
}

print("Perturbations defined:")

for name in PERTURBATIONS.keys():
    print("-", name)
```

```python
test_case = xai_cohort.iloc[0]
_, rgb_image = load_rgb_image_from_case(test_case)

plt.figure(figsize=(15, 6))

plt.subplot(2, 3, 1)
plt.imshow(rgb_image, cmap="gray")
plt.axis("off")
plt.title("Original")

for i, (name, function) in enumerate(PERTURBATIONS.items(), start=2):
    perturbed_image = function(rgb_image)

    plt.subplot(2, 3, i)
    plt.imshow(perturbed_image, cmap="gray")
    plt.axis("off")
    plt.title(name)

plt.tight_layout()
plt.show()
```

```python
robustness_rows = []

for _, case_row in tqdm(
    xai_cohort.iterrows(),
    total=len(xai_cohort),
    desc="prediction robustness"
):
    case_id = case_row["xai_case_id"]

    _, rgb_image = load_rgb_image_from_case(case_row)

    original_probability = predict_calibrated_probabilities_from_rgb_batch(
        [rgb_image]
    )[0]

    original_prediction = int(
        original_probability >= MODEL_DECISION_THRESHOLD
    )

    for perturbation_name, perturbation_function in PERTURBATIONS.items():
        perturbed_image = perturbation_function(rgb_image)

        perturbed_probability = predict_calibrated_probabilities_from_rgb_batch(
            [perturbed_image]
        )[0]

        perturbed_prediction = int(
            perturbed_probability >= MODEL_DECISION_THRESHOLD
        )

        absolute_probability_change = abs(
            float(original_probability) - float(perturbed_probability)
        )

        threshold_crossed = original_prediction != perturbed_prediction

        robustness_rows.append({
            "xai_case_id": case_id,
            "patientId": case_row["patientId"],
            "outcome": case_row["outcome"],
            "reference_class": case_row["reference_class"],
            "predicted_class": case_row["predicted_class"],
            "saved_calibrated_probability": case_row["calibrated_probability"],
            "original_probability": float(original_probability),
            "original_prediction": int(original_prediction),
            "perturbation": perturbation_name,
            "perturbed_probability": float(perturbed_probability),
            "perturbed_prediction": int(perturbed_prediction),
            "absolute_probability_change": float(absolute_probability_change),
            "threshold_crossed": bool(threshold_crossed)
        })

robustness_results = pd.DataFrame(robustness_rows)

robustness_results_path = robustness_results_folder / "prediction_robustness_results.csv"
robustness_results.to_csv(robustness_results_path, index=False)

print("Prediction robustness results saved:")
print(robustness_results_path)

print("\nRows:", len(robustness_results))

display(robustness_results.head())
```

```python
case_robustness = robustness_results.groupby(
    [
        "xai_case_id",
        "patientId",
        "outcome",
        "reference_class",
        "predicted_class"
    ]
).agg(
    original_probability=("original_probability", "first"),
    mean_absolute_probability_change=("absolute_probability_change", "mean"),
    max_absolute_probability_change=("absolute_probability_change", "max"),
    threshold_crossing_count=("threshold_crossed", "sum"),
    class_preservation_rate=("threshold_crossed", lambda x: 1 - x.mean())
).reset_index()

case_robustness["prediction_robustness_score"] = (
    1 - case_robustness["mean_absolute_probability_change"]
)

case_robustness["prediction_robustness_score"] = case_robustness[
    "prediction_robustness_score"
].clip(0, 1)

case_robustness_path = robustness_results_folder / "case_level_prediction_robustness.csv"
case_robustness.to_csv(case_robustness_path, index=False)

print("Case-level robustness saved:")
print(case_robustness_path)

print("\nRows:", len(case_robustness))

display(case_robustness.head())
```
