```python
import os
import random
import numpy as np
import torch

from torch.utils.data import DataLoader
from torchvision import datasets, transforms

print("Imports completed.")
```

```python
SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)
```

```python
DATASET_DIR = r"C:\Users\anudi\Downloads\MSc_Research\final_dataset"

TRAIN_DIR = os.path.join(DATASET_DIR, "train")
VAL_DIR = os.path.join(DATASET_DIR, "val")
TEST_DIR = os.path.join(DATASET_DIR, "test")

print("Train folder:", TRAIN_DIR)
print("Validation folder:", VAL_DIR)
print("Test folder:", TEST_DIR)

print("\nFolder checks:")
print("Train exists:", os.path.exists(TRAIN_DIR))
print("Validation exists:", os.path.exists(VAL_DIR))
print("Test exists:", os.path.exists(TEST_DIR))
```

```python
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

validation_test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("Transformations created.")
```

```python
train_dataset = datasets.ImageFolder(
    root=TRAIN_DIR,
    transform=train_transform
)

val_dataset = datasets.ImageFolder(
    root=VAL_DIR,
    transform=validation_test_transform
)

test_dataset = datasets.ImageFolder(
    root=TEST_DIR,
    transform=validation_test_transform
)

print("Training images:", len(train_dataset))
print("Validation images:", len(val_dataset))
print("Test images:", len(test_dataset))

print("\nClass mapping:")
print(train_dataset.class_to_idx)
```

```python
assert train_dataset.class_to_idx == val_dataset.class_to_idx
assert train_dataset.class_to_idx == test_dataset.class_to_idx

assert train_dataset.class_to_idx["Normal"] == 0
assert train_dataset.class_to_idx["Pneumonia"] == 1

assert len(train_dataset) == 10403
assert len(val_dataset) == 2230
assert len(test_dataset) == 2230

print("Dataset verification passed.")
```

```python
BATCH_SIZE = 16

train_generator = torch.Generator()
train_generator.manual_seed(SEED)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    generator=train_generator
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

print("DataLoaders created.")
```

```python
print("Class mapping:", train_dataset.class_to_idx)
print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

```python
import time
import pandas as pd
import torch.nn as nn
import torch.optim as optim

from torchvision.models import resnet50, ResNet50_Weights
from tqdm import tqdm

print("ResNet50 training libraries imported.")
```

```python
resnet_model = resnet50(
    weights=ResNet50_Weights.DEFAULT
)

print("Pretrained ResNet50 loaded.")
```

```python
for parameter in resnet_model.parameters():
    parameter.requires_grad = False
```

```python
number_of_features = resnet_model.fc.in_features

resnet_model.fc = nn.Linear(
    number_of_features,
    2
)

resnet_model = resnet_model.to(device)

print(resnet_model.fc)
```

```python
trainable_parameters = []

for name, parameter in resnet_model.named_parameters():
    if parameter.requires_grad:
        trainable_parameters.append(name)
        print(name)
```

```python
assert trainable_parameters == [
    "fc.weight",
    "fc.bias"
]

print("Only the ResNet50 classifier is trainable.")
```

```python
total_parameters = sum(
    parameter.numel()
    for parameter in resnet_model.parameters()
)

trainable_parameter_count = sum(
    parameter.numel()
    for parameter in resnet_model.parameters()
    if parameter.requires_grad
)

print(f"Total parameters: {total_parameters:,}")
print(f"Trainable parameters: {trainable_parameter_count:,}")
```

```python
criterion = nn.CrossEntropyLoss()

optimizer = optim.Adam(
    resnet_model.fc.parameters(),
    lr=0.001
)

NUM_EPOCHS = 3

print("Loss function and optimizer created.")
```

```python
def set_resnet_classifier_training_mode(model):
    model.train()

    for module in model.modules():
        if isinstance(module, nn.BatchNorm2d):
            module.eval()

    model.fc.train()
```

```python
training_history = []

best_validation_accuracy = 0.0
best_epoch = 0

for epoch in range(NUM_EPOCHS):

    epoch_start_time = time.time()

    # -------------------------
    # Training phase
    # -------------------------

    set_resnet_classifier_training_mode(
        resnet_model
    )

    running_train_loss = 0.0
    train_correct = 0
    train_total = 0

    training_progress = tqdm(
        train_loader,
        desc=f"Epoch {epoch + 1}/{NUM_EPOCHS} - Training"
    )

    for images, labels in training_progress:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = resnet_model(images)

        loss = criterion(
            outputs,
            labels
        )

        loss.backward()
        optimizer.step()

        running_train_loss += loss.item()

        predictions = outputs.argmax(
            dim=1
        )

        train_correct += (
            predictions == labels
        ).sum().item()

        train_total += labels.size(0)

        current_accuracy = (
            100 * train_correct / train_total
        )

        training_progress.set_postfix({
            "loss": f"{loss.item():.4f}",
            "accuracy": f"{current_accuracy:.2f}%"
        })

    average_train_loss = (
        running_train_loss /
        len(train_loader)
    )

    train_accuracy = (
        100 * train_correct /
        train_total
    )

    # -------------------------
    # Validation phase
    # -------------------------

    resnet_model.eval()

    running_validation_loss = 0.0
    validation_correct = 0
    validation_total = 0

    validation_progress = tqdm(
        val_loader,
        desc=f"Epoch {epoch + 1}/{NUM_EPOCHS} - Validation"
    )

    with torch.no_grad():

        for images, labels in validation_progress:

            images = images.to(device)
            labels = labels.to(device)

            outputs = resnet_model(images)

            loss = criterion(
                outputs,
                labels
            )

            running_validation_loss += loss.item()

            predictions = outputs.argmax(
                dim=1
            )

            validation_correct += (
                predictions == labels
            ).sum().item()

            validation_total += labels.size(0)

    average_validation_loss = (
        running_validation_loss /
        len(val_loader)
    )

    validation_accuracy = (
        100 * validation_correct /
        validation_total
    )

    epoch_time_minutes = (
        time.time() - epoch_start_time
    ) / 60

    epoch_results = {
        "epoch": epoch + 1,
        "train_loss": average_train_loss,
        "train_accuracy": train_accuracy,
        "validation_loss": average_validation_loss,
        "validation_accuracy": validation_accuracy,
        "time_minutes": epoch_time_minutes
    }

    training_history.append(
        epoch_results
    )

    print()
    print(f"Epoch [{epoch + 1}/{NUM_EPOCHS}]")
    print(f"Train Loss: {average_train_loss:.4f}")
    print(f"Validation Loss: {average_validation_loss:.4f}")
    print(f"Train Accuracy: {train_accuracy:.2f}%")
    print(f"Validation Accuracy: {validation_accuracy:.2f}%")
    print(f"Time: {epoch_time_minutes:.2f} minutes")

    # Save only the best validation checkpoint
    if validation_accuracy > best_validation_accuracy:

        best_validation_accuracy = validation_accuracy
        best_epoch = epoch + 1

        torch.save(
            resnet_model.state_dict(),
            "best_resnet50_rsna.pth"
        )

        print("New best ResNet50 checkpoint saved.")

    print(
        f"Best Validation Accuracy: "
        f"{best_validation_accuracy:.2f}%"
    )

    print("-" * 60)
```
