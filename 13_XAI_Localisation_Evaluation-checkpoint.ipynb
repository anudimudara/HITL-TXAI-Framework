```python
IMAGE_SIZE = 224
ORIGINAL_IMAGE_SIZE = 1024

def create_union_reference_mask(patient_id, bbox_df, output_size=224, original_size=1024):
    """
    Converts all RSNA bounding boxes for one patient/image into a single binary union mask.
    
    Output:
    - mask shape: 224 x 224
    - mask value 1: RSNA annotated opacity region
    - mask value 0: outside annotated region
    """

    case_boxes = bbox_df[bbox_df["patientId"] == patient_id]

    mask = np.zeros((output_size, output_size), dtype=np.uint8)

    scale_x = output_size / original_size
    scale_y = output_size / original_size

    for _, box in case_boxes.iterrows():
        x = float(box["x"])
        y = float(box["y"])
        w = float(box["width"])
        h = float(box["height"])

        x1 = int(round(x * scale_x))
        y1 = int(round(y * scale_y))
        x2 = int(round((x + w) * scale_x))
        y2 = int(round((y + h) * scale_y))

        x1 = max(0, min(output_size - 1, x1))
        y1 = max(0, min(output_size - 1, y1))
        x2 = max(0, min(output_size, x2))
        y2 = max(0, min(output_size, y2))

        mask[y1:y2, x1:x2] = 1

    return mask
```

```python
import pydicom
from PIL import Image
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
print("Searching for DICOM files...")

dicom_files = list(BASE_DIR.rglob("*.dcm"))

print("Number of DICOM files found:", len(dicom_files))

dicom_index = {}

for path in dicom_files:
    dicom_index[path.stem] = path

print("DICOM index created.")
```

```python
IMAGE_SIZE = 224

def read_dicom_as_rgb(dicom_path):
    dicom = pydicom.dcmread(str(dicom_path))
    image = dicom.pixel_array.astype(np.float32)

    if getattr(dicom, "PhotometricInterpretation", "") == "MONOCHROME1":
        image = np.max(image) - image

    image = image - np.min(image)

    if np.max(image) > 0:
        image = image / np.max(image)

    image = (image * 255).astype(np.uint8)

    rgb_image = np.stack([image, image, image], axis=-1)

    return rgb_image


def load_image_for_visual_check(row):
    patient_id = str(row["patientId"])

    if patient_id not in dicom_index:
        raise FileNotFoundError(f"DICOM image not found for patientId: {patient_id}")

    dicom_path = dicom_index[patient_id]

    rgb_image = read_dicom_as_rgb(dicom_path)

    pil_image = Image.fromarray(rgb_image).convert("RGB")
    pil_image = pil_image.resize((IMAGE_SIZE, IMAGE_SIZE))

    rgb_resized = np.array(pil_image)

    return dicom_path, rgb_resized
```

```python
def attribution_energy_inside_mask(heatmap, mask):
    """
    Measures the proportion of total attribution energy inside the RSNA reference mask.
    Range: 0 to 1.
    """

    heatmap = np.asarray(heatmap).astype(np.float32)
    mask = np.asarray(mask).astype(bool)

    total_energy = heatmap.sum()

    if total_energy <= 0:
        return 0.0

    inside_energy = heatmap[mask].sum()

    return float(inside_energy / total_energy)


def pointing_game_accuracy(heatmap, mask):
    """
    Checks whether the maximum attribution point is inside the RSNA reference mask.
    Returns:
    1.0 = maximum attribution point is inside the mask
    0.0 = maximum attribution point is outside the mask
    """

    heatmap = np.asarray(heatmap).astype(np.float32)
    mask = np.asarray(mask).astype(bool)

    if heatmap.max() <= 0:
        return 0.0

    max_position = np.unravel_index(np.argmax(heatmap), heatmap.shape)

    return float(mask[max_position])


def binarise_heatmap(heatmap, threshold=0.50):
    """
    Converts a continuous XAI heatmap into a binary highlighted region.
    """

    heatmap = np.asarray(heatmap).astype(np.float32)

    return (heatmap >= threshold).astype(np.uint8)


def iou_score(binary_heatmap, mask):
    """
    Calculates Intersection over Union between thresholded heatmap and RSNA mask.
    """

    binary_heatmap = np.asarray(binary_heatmap).astype(bool)
    mask = np.asarray(mask).astype(bool)

    intersection = np.logical_and(binary_heatmap, mask).sum()
    union = np.logical_or(binary_heatmap, mask).sum()

    if union == 0:
        return 0.0

    return float(intersection / union)


def dice_score(binary_heatmap, mask):
    """
    Calculates Dice similarity between thresholded heatmap and RSNA mask.
    """

    binary_heatmap = np.asarray(binary_heatmap).astype(bool)
    mask = np.asarray(mask).astype(bool)

    intersection = np.logical_and(binary_heatmap, mask).sum()
    denominator = binary_heatmap.sum() + mask.sum()

    if denominator == 0:
        return 0.0

    return float((2 * intersection) / denominator)
```

```python
HEATMAP_THRESHOLD = 0.50

print("Heatmap threshold for IoU and Dice:", HEATMAP_THRESHOLD)
```

```python
localisation_rows = []

positive_case_ids = set(reference_mask_manifest["xai_case_id"].tolist())

positive_xai_manifest = full_xai_manifest[
    full_xai_manifest["xai_case_id"].isin(positive_case_ids)
].copy()

print("XAI explanation rows with reference masks:", len(positive_xai_manifest))

print("\nMethod counts:")
print(positive_xai_manifest["method"].value_counts())

for _, row in positive_xai_manifest.iterrows():
    case_id = row["xai_case_id"]
    method = row["method"]

    mask_row = reference_mask_manifest[
        reference_mask_manifest["xai_case_id"] == case_id
    ].iloc[0]

    mask = np.load(mask_row["reference_mask_path"])
    heatmap = np.load(row["normalised_map_path"])

    if heatmap.shape != mask.shape:
        heatmap = cv2.resize(
            heatmap,
            (mask.shape[1], mask.shape[0]),
            interpolation=cv2.INTER_LINEAR
        )

    binary_heatmap = binarise_heatmap(
        heatmap,
        threshold=HEATMAP_THRESHOLD
    )

    energy_inside = attribution_energy_inside_mask(heatmap, mask)
    pointing = pointing_game_accuracy(heatmap, mask)
    iou = iou_score(binary_heatmap, mask)
    dice = dice_score(binary_heatmap, mask)

    localisation_rows.append({
        "xai_case_id": case_id,
        "patientId": row["patientId"],
        "method": method,
        "outcome": row["outcome"],
        "reference_class": row["reference_class"],
        "predicted_class": row["predicted_class"],
        "calibrated_probability": row["calibrated_probability"],
        "relative_opacity_area": mask_row["relative_opacity_area"],
        "reference_mask_area": mask_row["reference_mask_area"],
        "heatmap_threshold": HEATMAP_THRESHOLD,
        "attribution_energy_inside_mask": energy_inside,
        "pointing_game_accuracy": pointing,
        "iou": iou,
        "dice": dice,
        "normalised_map_path": row["normalised_map_path"],
        "overlay_path": row["overlay_path"],
        "reference_mask_path": mask_row["reference_mask_path"]
    })

localisation_results = pd.DataFrame(localisation_rows)

localisation_results_path = localisation_results_folder / "xai_localisation_results.csv"
localisation_results.to_csv(localisation_results_path, index=False)

print("Localisation results saved:")
print(localisation_results_path)

print("\nRows:", len(localisation_results))

print("\nMethod counts:")
print(localisation_results["method"].value_counts())

display(localisation_results.head())
```

```python
def create_overlap_visual(heatmap, mask):
    heatmap = np.asarray(heatmap).astype(np.float32)
    mask = np.asarray(mask).astype(np.uint8)

    heatmap_uint8 = np.uint8(255 * heatmap)
    heatmap_color = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)
    heatmap_color = cv2.cvtColor(heatmap_color, cv2.COLOR_BGR2RGB)

    mask_boundary = cv2.Canny((mask * 255).astype(np.uint8), 50, 150)

    overlap_visual = heatmap_color.copy()
    overlap_visual[mask_boundary > 0] = [255, 255, 255]

    return overlap_visual
```

```python
thresholds = [0.30, 0.40, 0.50, 0.60, 0.70]

sensitivity_rows = []

for _, row in positive_xai_manifest.iterrows():
    case_id = row["xai_case_id"]
    method = row["method"]

    mask_row = reference_mask_manifest[
        reference_mask_manifest["xai_case_id"] == case_id
    ].iloc[0]

    mask = np.load(mask_row["reference_mask_path"])
    heatmap = np.load(row["normalised_map_path"])

    if heatmap.shape != mask.shape:
        heatmap = cv2.resize(
            heatmap,
            (mask.shape[1], mask.shape[0]),
            interpolation=cv2.INTER_LINEAR
        )

    for threshold in thresholds:
        binary_heatmap = binarise_heatmap(
            heatmap,
            threshold=threshold
        )

        sensitivity_rows.append({
            "xai_case_id": case_id,
            "patientId": row["patientId"],
            "method": method,
            "outcome": row["outcome"],
            "threshold": threshold,
            "iou": iou_score(binary_heatmap, mask),
            "dice": dice_score(binary_heatmap, mask)
        })

threshold_sensitivity = pd.DataFrame(sensitivity_rows)

threshold_sensitivity_path = localisation_results_folder / "localisation_threshold_sensitivity.csv"
threshold_sensitivity.to_csv(threshold_sensitivity_path, index=False)

print("Threshold sensitivity saved:")
print(threshold_sensitivity_path)

display(
    threshold_sensitivity
    .groupby(["method", "threshold"])[["iou", "dice"]]
    .mean()
)
```

```python
def enhance_xray_for_display(rgb_image):
    """
    Improves X-ray visibility for display only.
    Does not affect localisation metrics.
    """
    gray = cv2.cvtColor(rgb_image, cv2.COLOR_RGB2GRAY)

    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )

    enhanced = clahe.apply(gray)

    enhanced_rgb = np.stack(
        [enhanced, enhanced, enhanced],
        axis=-1
    )

    return enhanced_rgb


def create_xray_heatmap_mask_visual(rgb_image, heatmap, mask, heatmap_alpha=0.30):
    """
    Creates a readable display:
    original X-ray + transparent XAI heatmap + white RSNA mask boundary.
    """

    rgb_display = enhance_xray_for_display(rgb_image)

    heatmap = np.asarray(heatmap).astype(np.float32)
    heatmap = heatmap - heatmap.min()

    if heatmap.max() > 0:
        heatmap = heatmap / heatmap.max()

    heatmap_uint8 = np.uint8(255 * heatmap)

    heatmap_color = cv2.applyColorMap(
        heatmap_uint8,
        cv2.COLORMAP_JET
    )

    heatmap_color = cv2.cvtColor(
        heatmap_color,
        cv2.COLOR_BGR2RGB
    )

    overlay = cv2.addWeighted(
        rgb_display.astype(np.uint8),
        1 - heatmap_alpha,
        heatmap_color.astype(np.uint8),
        heatmap_alpha,
        0
    )

    mask_boundary = cv2.Canny(
        (mask * 255).astype(np.uint8),
        50,
        150
    )

    overlay[mask_boundary > 0] = [255, 255, 255]

    return overlay
```

```python
sample_localisation = localisation_results.groupby("method").sample(
    n=2,
    random_state=42
)

for _, row in sample_localisation.iterrows():
    heatmap = np.load(row["normalised_map_path"])
    mask = np.load(row["reference_mask_path"])

    case_row = xai_cohort[
        xai_cohort["xai_case_id"] == row["xai_case_id"]
    ].iloc[0]

    _, rgb_image = load_image_for_visual_check(case_row)

    readable_overlay = create_xray_heatmap_mask_visual(
        rgb_image=rgb_image,
        heatmap=heatmap,
        mask=mask,
        heatmap_alpha=0.30
    )

    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(enhance_xray_for_display(rgb_image), cmap="gray")
    plt.axis("off")
    plt.title("Original X-ray")

    plt.subplot(1, 3, 2)
    plt.imshow(mask, cmap="gray")
    plt.axis("off")
    plt.title("RSNA reference mask")

    plt.subplot(1, 3, 3)
    plt.imshow(readable_overlay)
    plt.axis("off")
    plt.title(
        f"{row['xai_case_id']} | {row['method']} | "
        f"Energy={row['attribution_energy_inside_mask']:.2f}"
    )

    plt.tight_layout()
    plt.show()
```
