```python
# ============================================================
# COMMON SETUP FOR SECTION 4.4 FIGURES
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Output directory
# ------------------------------------------------------------

OUTPUT_DIR = "chapter4_4_figures"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ------------------------------------------------------------
# General matplotlib configuration
# ------------------------------------------------------------

plt.rcParams.update({
    "font.family": "Arial",
    "font.size": 11,
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "legend.fontsize": 10,
    "figure.dpi": 120,
    "savefig.dpi": 300
})

# ------------------------------------------------------------
# Helper function for saving each figure
# ------------------------------------------------------------

def save_figure(fig, filename):
    """
    Save a dissertation figure in both high-resolution PNG
    and vector PDF formats.
    """
    
    png_path = os.path.join(OUTPUT_DIR, f"{filename}.png")
    pdf_path = os.path.join(OUTPUT_DIR, f"{filename}.pdf")

    fig.savefig(
        png_path,
        dpi=300,
        bbox_inches="tight"
    )

    fig.savefig(
        pdf_path,
        bbox_inches="tight"
    )

    print("Saved:")
    print(png_path)
    print(pdf_path)
```

```python
# ============================================================
# FIGURE: HUMAN-EVALUATION DESIGN
# ============================================================

import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

fig, ax = plt.subplots(figsize=(13, 4.2))

ax.set_xlim(0, 13)
ax.set_ylim(0, 4)
ax.axis("off")

# ------------------------------------------------------------
# Box positions and text
# ------------------------------------------------------------

boxes = [
    (0.3, 1.4, 1.7, 1.2,
     "13 eligible\nspecialists"),

    (2.5, 1.4, 1.7, 1.2,
     "8 selected\ncases"),

    (4.7, 1.4, 1.7, 1.2,
     "4 blinded\nXAI methods"),

    (6.9, 1.4, 1.8, 1.2,
     "32 case-method\nevaluations\nper specialist"),

    (9.2, 1.4, 1.6, 1.2,
     "416 HTS\nrecords"),

    (11.2, 1.4, 1.5, 1.2,
     "2,496\nitem ratings")
]

# ------------------------------------------------------------
# Draw boxes
# ------------------------------------------------------------

for x, y, w, h, label in boxes:
    
    box = FancyBboxPatch(
        (x, y),
        w,
        h,
        boxstyle="round,pad=0.03,rounding_size=0.08",
        linewidth=1.3,
        facecolor="white"
    )
    
    ax.add_patch(box)
    
    ax.text(
        x + w/2,
        y + h/2,
        label,
        ha="center",
        va="center",
        fontsize=10.5,
        fontweight="bold"
    )

# ------------------------------------------------------------
# Draw arrows
# ------------------------------------------------------------

for i in range(len(boxes)-1):
    
    x1 = boxes[i][0] + boxes[i][2]
    y1 = boxes[i][1] + boxes[i][3]/2
    
    x2 = boxes[i+1][0]
    y2 = boxes[i+1][1] + boxes[i+1][3]/2
    
    arrow = FancyArrowPatch(
        (x1 + 0.05, y1),
        (x2 - 0.05, y2),
        arrowstyle="->",
        mutation_scale=15,
        linewidth=1.4
    )
    
    ax.add_patch(arrow)

# ------------------------------------------------------------
# Add design information
# ------------------------------------------------------------

ax.text(
    6.5,
    3.45,
    "Specialist human-evaluation design",
    ha="center",
    fontsize=15,
    fontweight="bold"
)

ax.text(
    6.5,
    0.65,
    "Fully crossed repeated-measures design: every specialist evaluated every case under every XAI condition",
    ha="center",
    fontsize=10.5
)

plt.tight_layout()

save_figure(
    fig,
    "Figure_Human_Evaluation_Design"
)

plt.show()
```

```python
# ============================================================
# STEP 1
# CREATE ANALYSIS-READY VARIABLES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# ------------------------------------------------------------
# Your real columns
# ------------------------------------------------------------

PARTICIPANT_COL = "participant_id"
CASE_COL = "case_id"
METHOD_COL = "method"
HTS_COL = "HTS_percentage"

# ------------------------------------------------------------
# Decode blinded XAI method labels
# ------------------------------------------------------------

method_mapping = {
    "Method A": "Grad-CAM",
    "Method B": "Grad-CAM++",
    "Method C": "LIME",
    "Method D": "SHAP"
}

df["XAI_Method"] = df[METHOD_COL].map(method_mapping)

# ------------------------------------------------------------
# Map the eight survey cases to prediction outcomes
#
# Case 1 = XAI_030 = TP
# Case 2 = XAI_003 = TP
# Case 3 = XAI_052 = TN
# Case 4 = XAI_032 = TN
# Case 5 = XAI_087 = FP
# Case 6 = XAI_066 = FP
# Case 7 = XAI_120 = FN
# Case 8 = XAI_099 = FN
# ------------------------------------------------------------

case_outcome_mapping = {
    "Case 1": "TP",
    "Case 2": "TP",
    "Case 3": "TN",
    "Case 4": "TN",
    "Case 5": "FP",
    "Case 6": "FP",
    "Case 7": "FN",
    "Case 8": "FN"
}

df["Outcome"] = df[CASE_COL].map(case_outcome_mapping)

# ------------------------------------------------------------
# Correctness
# ------------------------------------------------------------

df["Correctness"] = df["Outcome"].map({
    "TP": "Correct",
    "TN": "Correct",
    "FP": "Error",
    "FN": "Error"
})

# ------------------------------------------------------------
# Descriptive trust bands
# ------------------------------------------------------------

def classify_hts(score):
    if score < 40:
        return "Low"
    elif score < 70:
        return "Moderate"
    else:
        return "High"

df["HTS_Band"] = df[HTS_COL].apply(classify_hts)

print(df[
    [
        PARTICIPANT_COL,
        CASE_COL,
        METHOD_COL,
        "XAI_Method",
        "Outcome",
        "Correctness",
        HTS_COL,
        "HTS_Band"
    ]
].head(12))
```

```python
# ============================================================
# FIGURE 1
# OVERALL HTS DISTRIBUTION
# ============================================================

hts = df[HTS_COL].dropna()

mean_hts = hts.mean()
median_hts = hts.median()
sd_hts = hts.std()

fig, ax = plt.subplots(
    figsize=(9, 5.7)
)

bins = np.arange(
    0,
    105,
    5
)

ax.hist(
    hts,
    bins=bins,
    edgecolor="white",
    linewidth=0.8
)

# ------------------------------------------------------------
# Mean and median
# ------------------------------------------------------------

ax.axvline(
    mean_hts,
    linestyle="--",
    linewidth=1.8,
    label=f"Mean = {mean_hts:.2f}%"
)

ax.axvline(
    median_hts,
    linestyle=":",
    linewidth=1.8,
    label=f"Median = {median_hts:.2f}%"
)

# ------------------------------------------------------------
# Trust-band boundaries
# ------------------------------------------------------------

ax.axvline(
    40,
    linewidth=1.1
)

ax.axvline(
    70,
    linewidth=1.1
)

ymax = ax.get_ylim()[1]

ax.text(
    20,
    ymax * 0.92,
    "Low",
    ha="center",
    fontweight="bold"
)

ax.text(
    55,
    ymax * 0.92,
    "Moderate",
    ha="center",
    fontweight="bold"
)

ax.text(
    85,
    ymax * 0.92,
    "High",
    ha="center",
    fontweight="bold"
)

# ------------------------------------------------------------
# Summary box
# ------------------------------------------------------------

summary = (
    f"n = {len(hts)}\n"
    f"Mean = {mean_hts:.2f}%\n"
    f"Median = {median_hts:.2f}%\n"
    f"SD = {sd_hts:.2f}"
)

ax.text(
    0.98,
    0.96,
    summary,
    transform=ax.transAxes,
    ha="right",
    va="top",
    bbox=dict(
        boxstyle="round,pad=0.4",
        facecolor="white",
        alpha=0.9
    )
)

ax.set_title(
    "Distribution of Specialist Human Trust Scores"
)

ax.set_xlabel(
    "Human Trust Score (HTS, %)"
)

ax.set_ylabel(
    "Case–method evaluations"
)

ax.set_xlim(
    0,
    100
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.legend(
    frameon=False,
    loc="upper left"
)

plt.tight_layout()

save_figure(
    fig,
    "HTS_Overall_Distribution"
)

plt.show()
```

```python
# ============================================================
# PARTICIPANT-LEVEL OUTCOME DATA
# ============================================================

participant_outcome = (
    df.groupby(
        [
            PARTICIPANT_COL,
            "Outcome"
        ]
    )[HTS_COL]
    .mean()
    .reset_index()
)

print(
    participant_outcome.shape
)

# 13 participants × 4 outcomes = 52
```

```python
outcome_order = [
    "TP",
    "TN",
    "FP",
    "FN"
]

outcome_summary = (
    participant_outcome
    .groupby("Outcome")[HTS_COL]
    .agg(
        Mean="mean",
        SD="std",
        N="count"
    )
    .reindex(outcome_order)
)

outcome_summary["SE"] = (
    outcome_summary["SD"]
    /
    np.sqrt(
        outcome_summary["N"]
    )
)

outcome_summary["CI95"] = (
    1.96 *
    outcome_summary["SE"]
)

print(
    outcome_summary.round(2)
)
```

```python
# ============================================================
# FIGURE 4
# HTS ACROSS TP / TN / FP / FN
# ============================================================

fig, ax = plt.subplots(
    figsize=(8.5, 5.6)
)

x = np.arange(
    len(outcome_order)
)

ax.errorbar(
    x,
    outcome_summary["Mean"],
    yerr=outcome_summary["CI95"],
    fmt="o",
    markersize=9,
    linewidth=1.7,
    capsize=5
)

for i, outcome in enumerate(
    outcome_order
):

    mean_value = (
        outcome_summary.loc[
            outcome,
            "Mean"
        ]
    )

    ci = (
        outcome_summary.loc[
            outcome,
            "CI95"
        ]
    )

    ax.text(
        i,
        mean_value + ci + 1.5,
        f"{mean_value:.2f}%",
        ha="center",
        fontweight="bold"
    )

ax.set_xticks(x)

ax.set_xticklabels(
    [
        "TP\nCorrect positive",
        "TN\nCorrect negative",
        "FP\nError",
        "FN\nError"
    ]
)

ax.set_ylabel(
    "Participant mean HTS (%)"
)

ax.set_xlabel(
    "Prediction outcome"
)

ax.set_ylim(
    25,
    70
)

ax.set_title(
    "Specialist Human Trust Across Prediction Outcomes"
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()

save_figure(
    fig,
    "HTS_By_Prediction_Outcome"
)

plt.show()
```

```python
correctness_pivot = (
    participant_correctness
    .pivot(
        index=PARTICIPANT_COL,
        columns="Correctness",
        values=HTS_COL
    )
)

print(
    correctness_pivot
)
```

```python
# ============================================================
# FIGURE 5
# PAIRED CORRECT VS ERROR SPECIALIST TRUST
# ============================================================

fig, ax = plt.subplots(
    figsize=(7.5, 6)
)

x_positions = [
    0,
    1
]

for participant, row in correctness_pivot.iterrows():

    ax.plot(
        x_positions,
        [
            row["Error"],
            row["Correct"]
        ],
        marker="o",
        linewidth=1,
        alpha=0.40
    )

# ------------------------------------------------------------
# Overall means
# ------------------------------------------------------------

error_mean = (
    correctness_pivot["Error"]
    .mean()
)

correct_mean = (
    correctness_pivot["Correct"]
    .mean()
)

ax.plot(
    x_positions,
    [
        error_mean,
        correct_mean
    ],
    marker="o",
    markersize=10,
    linewidth=3.5,
    label="Mean across specialists"
)

ax.text(
    0,
    error_mean - 4,
    f"{error_mean:.2f}%",
    ha="center",
    fontweight="bold"
)

ax.text(
    1,
    correct_mean + 3,
    f"{correct_mean:.2f}%",
    ha="center",
    fontweight="bold"
)

difference = (
    correct_mean -
    error_mean
)

ax.text(
    0.5,
    72,
    f"Mean difference = +{difference:.2f} percentage points",
    ha="center",
    fontweight="bold"
)

ax.text(
    0.5,
    68,
    "Paired Wilcoxon p = 0.000244",
    ha="center"
)

ax.set_xticks(
    x_positions
)

ax.set_xticklabels(
    [
        "Error predictions\n(FP + FN)",
        "Correct predictions\n(TP + TN)"
    ]
)

ax.set_ylabel(
    "Participant mean HTS (%)"
)

ax.set_xlim(
    -0.3,
    1.3
)

ax.set_ylim(
    0,
    80
)

ax.set_title(
    "Within-Specialist Trust for Correct and Incorrect Predictions"
)

ax.legend(
    frameon=False
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()

save_figure(
    fig,
    "HTS_Correct_vs_Error_Paired"
)

plt.show()
```

```python
# ============================================================
# FIGURE 6
# SPECIALIST EVALUATION DESIGN
# ============================================================

from matplotlib.patches import (
    FancyBboxPatch,
    FancyArrowPatch
)

fig, ax = plt.subplots(
    figsize=(13, 4)
)

ax.set_xlim(
    0,
    13
)

ax.set_ylim(
    0,
    4
)

ax.axis("off")

boxes = [
    (0.2, 1.4, 1.6, 1.1,
     "13 eligible\nspecialists"),

    (2.3, 1.4, 1.6, 1.1,
     "8 selected\ncases"),

    (4.4, 1.4, 1.7, 1.1,
     "4 blinded\nXAI methods"),

    (6.6, 1.4, 1.9, 1.1,
     "32 evaluations\nper specialist"),

    (9.0, 1.4, 1.6, 1.1,
     "416 HTS\nrecords"),

    (11.1, 1.4, 1.6, 1.1,
     "2,496\nitem ratings")
]

for x, y, w, h, label in boxes:

    box = FancyBboxPatch(
        (x, y),
        w,
        h,
        boxstyle="round,pad=0.04",
        linewidth=1.3,
        facecolor="white"
    )

    ax.add_patch(box)

    ax.text(
        x + w/2,
        y + h/2,
        label,
        ha="center",
        va="center",
        fontweight="bold"
    )

for i in range(
    len(boxes) - 1
):

    current = boxes[i]
    nxt = boxes[i+1]

    start = (
        current[0] + current[2] + 0.05,
        current[1] + current[3]/2
    )

    end = (
        nxt[0] - 0.05,
        nxt[1] + nxt[3]/2
    )

    arrow = FancyArrowPatch(
        start,
        end,
        arrowstyle="->",
        mutation_scale=15,
        linewidth=1.4
    )

    ax.add_patch(arrow)

ax.text(
    6.5,
    3.35,
    "Specialist Human-Evaluation Design",
    ha="center",
    fontsize=15,
    fontweight="bold"
)

ax.text(
    6.5,
    0.62,
    "Fully crossed repeated-measures design: each specialist evaluated every case under every XAI condition",
    ha="center",
    fontsize=10.5
)

plt.tight_layout()

save_figure(
    fig,
    "HTS_Evaluation_Design"
)

plt.show()
```
