```python
def extract_state_dict(checkpoint):
    if isinstance(checkpoint, dict):
        for key in ["model_state_dict", "state_dict", "model"]:
            if key in checkpoint:
                return checkpoint[key]
    return checkpoint


def clean_state_dict_keys(state_dict):
    cleaned = {}

    for key, value in state_dict.items():
        new_key = key

        if new_key.startswith("module."):
            new_key = new_key.replace("module.", "", 1)

        cleaned[new_key] = value

    return cleaned


checkpoint = torch.load(resnet_checkpoint_path, map_location=device)
state_dict = extract_state_dict(checkpoint)
state_dict = clean_state_dict_keys(state_dict)

output_features = state_dict["fc.weight"].shape[0]

print("Detected ResNet50 output features:", output_features)

model = models.resnet50(weights=None)
model.fc = nn.Linear(model.fc.in_features, output_features)

model.load_state_dict(state_dict, strict=True)
model = model.to(device)
model.eval()

print("Frozen ResNet50 loaded successfully.")
```

```python
IMAGE_SIZE = 224

model_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("Preprocessing transform defined.")
```

```python
def read_dicom_as_rgb(dicom_path):
    dicom = pydicom.dcmread(str(dicom_path))
    image = dicom.pixel_array.astype(np.float32)

    if getattr(dicom, "PhotometricInterpretation", "") == "MONOCHROME1":
        image = np.max(image) - image

    image = image - np.min(image)

    if np.max(image) > 0:
        image = image / np.max(image)

    image = (image * 255).astype(np.uint8)

    rgb_image = np.stack([image, image, image], axis=-1)

    return rgb_image


def load_rgb_image_from_case(case_row):
    patient_id = str(case_row["patientId"])

    if patient_id not in dicom_index:
        raise FileNotFoundError(f"DICOM image not found for patientId: {patient_id}")

    dicom_path = dicom_index[patient_id]

    rgb_image = read_dicom_as_rgb(dicom_path)

    pil_image = Image.fromarray(rgb_image).convert("RGB")
    pil_image = pil_image.resize((IMAGE_SIZE, IMAGE_SIZE))

    rgb_resized = np.array(pil_image)

    return dicom_path, rgb_resized


def rgb_to_input_tensor(rgb_image):
    pil_image = Image.fromarray(rgb_image.astype(np.uint8)).convert("RGB")
    input_tensor = model_transform(pil_image).unsqueeze(0).to(device)

    return input_tensor
```

```python
def apply_gaussian_noise(rgb_image, noise_std=8):
    noise = np.random.normal(
        loc=0,
        scale=noise_std,
        size=rgb_image.shape
    )

    noisy = rgb_image.astype(np.float32) + noise

    return np.clip(noisy, 0, 255).astype(np.uint8)


def apply_brightness_change(rgb_image, factor=1.10):
    bright = rgb_image.astype(np.float32) * factor

    return np.clip(bright, 0, 255).astype(np.uint8)


def apply_contrast_change(rgb_image, factor=1.10):
    mean = np.mean(
        rgb_image,
        axis=(0, 1),
        keepdims=True
    )

    contrast = (rgb_image.astype(np.float32) - mean) * factor + mean

    return np.clip(contrast, 0, 255).astype(np.uint8)


def apply_translation(rgb_image, shift_x=5, shift_y=5):
    h, w = rgb_image.shape[:2]

    matrix = np.float32([
        [1, 0, shift_x],
        [0, 1, shift_y]
    ])

    translated = cv2.warpAffine(
        rgb_image,
        matrix,
        (w, h),
        borderMode=cv2.BORDER_REFLECT
    )

    return translated.astype(np.uint8)


def apply_rotation(rgb_image, angle=5):
    h, w = rgb_image.shape[:2]
    center = (w // 2, h // 2)

    matrix = cv2.getRotationMatrix2D(
        center,
        angle,
        1.0
    )

    rotated = cv2.warpAffine(
        rgb_image,
        matrix,
        (w, h),
        borderMode=cv2.BORDER_REFLECT
    )

    return rotated.astype(np.uint8)


PERTURBATIONS = {
    "gaussian_noise_std8": lambda img: apply_gaussian_noise(img, noise_std=8),
    "brightness_plus_10": lambda img: apply_brightness_change(img, factor=1.10),
    "contrast_plus_10": lambda img: apply_contrast_change(img, factor=1.10),
    "translation_5px": lambda img: apply_translation(img, shift_x=5, shift_y=5),
    "rotation_5deg": lambda img: apply_rotation(img, angle=5)
}

print("Perturbations defined:")
for name in PERTURBATIONS.keys():
    print("-", name)
```

```python
def reverse_align_heatmap(heatmap, perturbation_name):
    heatmap = np.asarray(heatmap).astype(np.float32)

    h, w = heatmap.shape[:2]

    if perturbation_name == "translation_5px":
        matrix = np.float32([
            [1, 0, -5],
            [0, 1, -5]
        ])

        aligned = cv2.warpAffine(
            heatmap,
            matrix,
            (w, h),
            borderMode=cv2.BORDER_REFLECT
        )

        return aligned.astype(np.float32)

    elif perturbation_name == "rotation_5deg":
        center = (w // 2, h // 2)

        matrix = cv2.getRotationMatrix2D(
            center,
            -5,
            1.0
        )

        aligned = cv2.warpAffine(
            heatmap,
            matrix,
            (w, h),
            borderMode=cv2.BORDER_REFLECT
        )

        return aligned.astype(np.float32)

    else:
        return heatmap
```

```python
def normalise_map(raw_map):
    raw_map = np.asarray(raw_map).astype(np.float32)

    raw_map = np.nan_to_num(
        raw_map,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    raw_map = raw_map - np.min(raw_map)

    if np.max(raw_map) > 0:
        raw_map = raw_map / np.max(raw_map)

    return raw_map.astype(np.float32)
```

```python
from pytorch_grad_cam import GradCAM, GradCAMPlusPlus
from pytorch_grad_cam.utils.model_targets import ClassifierOutputTarget

target_layers = [model.layer4[-1]]

if output_features == 1:
    positive_output_index = 0
else:
    positive_output_index = 1

positive_target = [ClassifierOutputTarget(positive_output_index)]

gradcam_objects = {
    "gradcam": GradCAM(
        model=model,
        target_layers=target_layers
    ),
    "gradcampp": GradCAMPlusPlus(
        model=model,
        target_layers=target_layers
    )
}

print("Grad-CAM and Grad-CAM++ ready.")
```

```python
from lime import lime_image
from skimage.segmentation import slic

LIME_NUM_SAMPLES = 500
LIME_N_SEGMENTS = 50
LIME_COMPACTNESS = 10
LIME_SIGMA = 1

lime_explainer = lime_image.LimeImageExplainer(
    random_state=RANDOM_SEED
)

print("LIME explainer ready.")
```

```python
def lime_predict_function(images):
    model.eval()

    batch_tensors = []

    for image in images:
        image = np.asarray(image)

        if image.max() <= 1.0:
            image = (image * 255).astype(np.uint8)
        else:
            image = image.astype(np.uint8)

        pil_image = Image.fromarray(image).convert("RGB")
        tensor = model_transform(pil_image)
        batch_tensors.append(tensor)

    batch_tensor = torch.stack(batch_tensors).to(device)

    with torch.no_grad():
        logits = model(batch_tensor)
        scaled_logits = logits / TEMPERATURE

        if scaled_logits.shape[1] == 1:
            pneumonia_probability = torch.sigmoid(scaled_logits)
            normal_probability = 1 - pneumonia_probability

            probabilities = torch.cat(
                [normal_probability, pneumonia_probability],
                dim=1
            )

        elif scaled_logits.shape[1] == 2:
            probabilities = torch.softmax(scaled_logits, dim=1)

        else:
            raise ValueError(f"Unexpected output shape: {scaled_logits.shape}")

    return probabilities.detach().cpu().numpy()


def lime_segmentation_function(image):
    return slic(
        image,
        n_segments=LIME_N_SEGMENTS,
        compactness=LIME_COMPACTNESS,
        sigma=LIME_SIGMA,
        start_label=0
    )
```

```python
class CalibratedProbabilityWrapper(nn.Module):
    def __init__(self, base_model, temperature):
        super().__init__()
        self.base_model = base_model
        self.temperature = float(temperature)

    def forward(self, x):
        logits = self.base_model(x)
        scaled_logits = logits / self.temperature

        if scaled_logits.shape[1] == 1:
            pneumonia_probability = torch.sigmoid(scaled_logits)
            normal_probability = 1 - pneumonia_probability

            probabilities = torch.cat(
                [normal_probability, pneumonia_probability],
                dim=1
            )

        elif scaled_logits.shape[1] == 2:
            probabilities = torch.softmax(scaled_logits, dim=1)

        else:
            raise ValueError(f"Unexpected model output shape: {scaled_logits.shape}")

        return probabilities


shap_model = CalibratedProbabilityWrapper(
    base_model=model,
    temperature=TEMPERATURE
).to(device)

shap_model.eval()

print("SHAP calibrated probability wrapper created.")
```

```python
background_rows = []

for outcome_name, group_df in xai_cohort.groupby("outcome"):
    sample_n = min(2, len(group_df))

    sampled_group = group_df.sample(
        n=sample_n,
        random_state=RANDOM_SEED
    )

    background_rows.append(sampled_group)

shap_background_df = pd.concat(background_rows, axis=0).reset_index(drop=True)

background_tensors = []

for _, row in shap_background_df.iterrows():
    _, rgb_image = load_rgb_image_from_case(row)
    input_tensor = rgb_to_input_tensor(rgb_image)
    background_tensors.append(input_tensor.squeeze(0).detach().cpu())

shap_background = torch.stack(background_tensors).to(device)

print("SHAP background tensor shape:", shap_background.shape)

shap_explainer = shap.GradientExplainer(
    shap_model,
    shap_background
)

print("SHAP GradientExplainer initialised.")
```

```python
def extract_positive_class_shap_map(shap_values, target_index=1):
    if isinstance(shap_values, list):
        positive_values = shap_values[target_index]
    else:
        positive_values = shap_values

    positive_values = np.asarray(positive_values)
    positive_values = np.squeeze(positive_values)

    if positive_values.ndim == 4:
        if positive_values.shape[-1] == 2:
            positive_values = positive_values[..., target_index]

        elif positive_values.shape[0] == 2:
            positive_values = positive_values[target_index]

        else:
            raise ValueError(f"Unexpected 4D SHAP shape: {positive_values.shape}")

    if positive_values.ndim == 3:
        if positive_values.shape[0] == 3:
            shap_map = np.mean(np.abs(positive_values), axis=0)

        elif positive_values.shape[-1] == 3:
            shap_map = np.mean(np.abs(positive_values), axis=-1)

        else:
            shap_map = np.mean(np.abs(positive_values), axis=0)

    elif positive_values.ndim == 2:
        shap_map = np.abs(positive_values)

    else:
        raise ValueError(f"Unexpected SHAP shape after processing: {positive_values.shape}")

    return shap_map.astype(np.float32)
```

```python
SHAP_NSAMPLES = 200

def generate_xai_map_for_image(rgb_image, method):
    input_tensor = rgb_to_input_tensor(rgb_image)

    if method in ["gradcam", "gradcampp"]:
        cam_object = gradcam_objects[method]

        cam_map = cam_object(
            input_tensor=input_tensor,
            targets=positive_target
        )[0]

        return normalise_map(cam_map)

    elif method == "lime":
        explanation = lime_explainer.explain_instance(
            image=rgb_image,
            classifier_fn=lime_predict_function,
            labels=[1],
            hide_color=0,
            num_samples=LIME_NUM_SAMPLES,
            segmentation_fn=lime_segmentation_function
        )

        segments = explanation.segments

        if 1 not in explanation.local_exp:
            raise ValueError("LIME did not return label 1 explanation.")

        local_exp = dict(explanation.local_exp[1])

        lime_map = np.zeros(segments.shape, dtype=np.float32)

        for segment_id, weight in local_exp.items():
            if weight > 0:
                lime_map[segments == segment_id] = weight

        return normalise_map(lime_map)

    elif method == "shap":
        shap_values = shap_explainer.shap_values(
            input_tensor,
            nsamples=SHAP_NSAMPLES,
            rseed=RANDOM_SEED
        )

        shap_map = extract_positive_class_shap_map(
            shap_values,
            target_index=1
        )

        return normalise_map(shap_map)

    else:
        raise ValueError(f"Unknown XAI method: {method}")
```

```python
def cosine_similarity_maps(map_a, map_b):
    a = map_a.flatten().astype(np.float32)
    b = map_b.flatten().astype(np.float32)

    denominator = np.linalg.norm(a) * np.linalg.norm(b)

    if denominator == 0:
        return 0.0

    return float(np.dot(a, b) / denominator)


def spearman_similarity_maps(map_a, map_b):
    a = map_a.flatten()
    b = map_b.flatten()

    if np.std(a) == 0 or np.std(b) == 0:
        return 0.0

    correlation, _ = spearmanr(a, b)

    if np.isnan(correlation):
        return 0.0

    return float(correlation)


def top_attribution_overlap(map_a, map_b, top_fraction=0.10):
    a = map_a.flatten()
    b = map_b.flatten()

    k = int(round(top_fraction * len(a)))

    if k <= 0:
        return 0.0

    top_a = set(np.argsort(a)[::-1][:k])
    top_b = set(np.argsort(b)[::-1][:k])

    overlap = len(top_a.intersection(top_b)) / k

    return float(overlap)


def calculate_stability_metrics(original_map, perturbed_map):
    original_map = normalise_map(original_map)
    perturbed_map = normalise_map(perturbed_map)

    ssim_value = ssim(
        original_map,
        perturbed_map,
        data_range=1.0
    )

    cosine_value = cosine_similarity_maps(
        original_map,
        perturbed_map
    )

    spearman_value = spearman_similarity_maps(
        original_map,
        perturbed_map
    )

    top_overlap_value = top_attribution_overlap(
        original_map,
        perturbed_map,
        top_fraction=0.10
    )

    # Convert possible -1 to +1 metrics into 0 to 1 range
    ssim_normalised = (ssim_value + 1) / 2
    spearman_normalised = (spearman_value + 1) / 2

    stability_score = np.mean([
        ssim_normalised,
        max(0.0, min(1.0, cosine_value)),
        spearman_normalised,
        top_overlap_value
    ])

    return {
        "ssim": float(ssim_value),
        "cosine_similarity": float(cosine_value),
        "spearman_correlation": float(spearman_value),
        "top_10_percent_overlap": float(top_overlap_value),
        "stability_score": float(stability_score)
    }
```

```python
def evaluate_stability_for_row(manifest_row, perturbations):
    case_id = manifest_row["xai_case_id"]
    method = manifest_row["method"]

    case_row = xai_cohort[
        xai_cohort["xai_case_id"] == case_id
    ].iloc[0]

    _, rgb_image = load_rgb_image_from_case(case_row)

    original_map = np.load(manifest_row["normalised_map_path"])

    original_map = normalise_map(original_map)

    result_rows = []

    for perturbation_name, perturbation_function in perturbations.items():
        start_time = time.time()

        perturbed_image = perturbation_function(rgb_image)

        perturbed_map = generate_xai_map_for_image(
            rgb_image=perturbed_image,
            method=method
        )

        perturbed_map = reverse_align_heatmap(
            heatmap=perturbed_map,
            perturbation_name=perturbation_name
        )

        perturbed_map = normalise_map(perturbed_map)

        metrics = calculate_stability_metrics(
            original_map=original_map,
            perturbed_map=perturbed_map
        )

        runtime_seconds = time.time() - start_time

        result_rows.append({
            "xai_case_id": case_id,
            "patientId": manifest_row["patientId"],
            "method": method,
            "outcome": manifest_row["outcome"],
            "reference_class": manifest_row["reference_class"],
            "predicted_class": manifest_row["predicted_class"],
            "calibrated_probability": manifest_row["calibrated_probability"],
            "perturbation": perturbation_name,
            "runtime_seconds": runtime_seconds,
            **metrics
        })

    return result_rows
```

```python
pilot_stability_rows = []

for _, manifest_row in tqdm(
    stability_pilot_manifest.iterrows(),
    total=len(stability_pilot_manifest),
    desc="stability pilot"
):
    rows = evaluate_stability_for_row(
        manifest_row=manifest_row,
        perturbations=PERTURBATIONS
    )

    pilot_stability_rows.extend(rows)

    gc.collect()

pilot_stability_results = pd.DataFrame(pilot_stability_rows)

pilot_stability_path = stability_results_folder / "stability_pilot_results.csv"
pilot_stability_results.to_csv(pilot_stability_path, index=False)

print("Stability pilot saved:")
print(pilot_stability_path)

print("Rows:", len(pilot_stability_results))

print("\nMethod counts:")
print(pilot_stability_results["method"].value_counts())

display(pilot_stability_results.head())
```

```python
def extract_state_dict(checkpoint):
    if isinstance(checkpoint, dict):
        for key in ["model_state_dict", "state_dict", "model"]:
            if key in checkpoint:
                return checkpoint[key]
    return checkpoint


def clean_state_dict_keys(state_dict):
    cleaned = {}

    for key, value in state_dict.items():
        new_key = key

        if new_key.startswith("module."):
            new_key = new_key.replace("module.", "", 1)

        cleaned[new_key] = value

    return cleaned


checkpoint = torch.load(resnet_checkpoint_path, map_location=device)
state_dict = extract_state_dict(checkpoint)
state_dict = clean_state_dict_keys(state_dict)

if "fc.weight" not in state_dict:
    raise ValueError("Could not find fc.weight in checkpoint.")

output_features = state_dict["fc.weight"].shape[0]

print("Detected ResNet50 output features:", output_features)

model = models.resnet50(weights=None)
model.fc = nn.Linear(model.fc.in_features, output_features)

model.load_state_dict(state_dict, strict=True)
model = model.to(device)
model.eval()

print("Frozen ResNet50 loaded successfully.")
```

```python
IMAGE_SIZE = 224

model_transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("Preprocessing transform defined.")
```

```python
print("Searching for DICOM files...")

dicom_files = list(BASE_DIR.rglob("*.dcm"))

print("Number of DICOM files found:", len(dicom_files))

dicom_index = {}

for path in dicom_files:
    dicom_index[path.stem] = path

print("DICOM index created.")


def read_dicom_as_rgb(dicom_path):
    dicom = pydicom.dcmread(str(dicom_path))
    image = dicom.pixel_array.astype(np.float32)

    if getattr(dicom, "PhotometricInterpretation", "") == "MONOCHROME1":
        image = np.max(image) - image

    image = image - np.min(image)

    if np.max(image) > 0:
        image = image / np.max(image)

    image = (image * 255).astype(np.uint8)

    rgb_image = np.stack([image, image, image], axis=-1)

    return rgb_image


def load_rgb_image_from_case(case_row):
    patient_id = str(case_row["patientId"])

    if patient_id not in dicom_index:
        raise FileNotFoundError(f"DICOM image not found for patientId: {patient_id}")

    dicom_path = dicom_index[patient_id]

    rgb_image = read_dicom_as_rgb(dicom_path)

    pil_image = Image.fromarray(rgb_image).convert("RGB")
    pil_image = pil_image.resize((IMAGE_SIZE, IMAGE_SIZE))

    rgb_resized = np.array(pil_image)

    return dicom_path, rgb_resized


def rgb_to_input_tensor(rgb_image):
    pil_image = Image.fromarray(rgb_image.astype(np.uint8)).convert("RGB")
    input_tensor = model_transform(pil_image).unsqueeze(0).to(device)

    return input_tensor


print("DICOM loading functions ready.")
```

```python
def apply_gaussian_noise(rgb_image, noise_std=8):
    noise = np.random.normal(
        loc=0,
        scale=noise_std,
        size=rgb_image.shape
    )

    noisy = rgb_image.astype(np.float32) + noise

    return np.clip(noisy, 0, 255).astype(np.uint8)


def apply_brightness_change(rgb_image, factor=1.10):
    bright = rgb_image.astype(np.float32) * factor

    return np.clip(bright, 0, 255).astype(np.uint8)


def apply_contrast_change(rgb_image, factor=1.10):
    mean = np.mean(
        rgb_image,
        axis=(0, 1),
        keepdims=True
    )

    contrast = (rgb_image.astype(np.float32) - mean) * factor + mean

    return np.clip(contrast, 0, 255).astype(np.uint8)


def apply_translation(rgb_image, shift_x=5, shift_y=5):
    h, w = rgb_image.shape[:2]

    matrix = np.float32([
        [1, 0, shift_x],
        [0, 1, shift_y]
    ])

    translated = cv2.warpAffine(
        rgb_image,
        matrix,
        (w, h),
        borderMode=cv2.BORDER_REFLECT
    )

    return translated.astype(np.uint8)


def apply_rotation(rgb_image, angle=5):
    h, w = rgb_image.shape[:2]
    center = (w // 2, h // 2)

    matrix = cv2.getRotationMatrix2D(
        center,
        angle,
        1.0
    )

    rotated = cv2.warpAffine(
        rgb_image,
        matrix,
        (w, h),
        borderMode=cv2.BORDER_REFLECT
    )

    return rotated.astype(np.uint8)


PERTURBATIONS = {
    "gaussian_noise_std8": lambda img: apply_gaussian_noise(img, noise_std=8),
    "brightness_plus_10": lambda img: apply_brightness_change(img, factor=1.10),
    "contrast_plus_10": lambda img: apply_contrast_change(img, factor=1.10),
    "translation_5px": lambda img: apply_translation(img, shift_x=5, shift_y=5),
    "rotation_5deg": lambda img: apply_rotation(img, angle=5)
}

print("Perturbations defined:")
for name in PERTURBATIONS.keys():
    print("-", name)
```

```python
def normalise_map(raw_map):
    raw_map = np.asarray(raw_map).astype(np.float32)

    raw_map = np.nan_to_num(
        raw_map,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    raw_map = raw_map - np.min(raw_map)

    if np.max(raw_map) > 0:
        raw_map = raw_map / np.max(raw_map)

    return raw_map.astype(np.float32)


def reverse_align_heatmap(heatmap, perturbation_name):
    heatmap = np.asarray(heatmap).astype(np.float32)

    h, w = heatmap.shape[:2]

    if perturbation_name == "translation_5px":
        matrix = np.float32([
            [1, 0, -5],
            [0, 1, -5]
        ])

        aligned = cv2.warpAffine(
            heatmap,
            matrix,
            (w, h),
            borderMode=cv2.BORDER_REFLECT
        )

        return aligned.astype(np.float32)

    elif perturbation_name == "rotation_5deg":
        center = (w // 2, h // 2)

        matrix = cv2.getRotationMatrix2D(
            center,
            -5,
            1.0
        )

        aligned = cv2.warpAffine(
            heatmap,
            matrix,
            (w, h),
            borderMode=cv2.BORDER_REFLECT
        )

        return aligned.astype(np.float32)

    else:
        return heatmap


print("Normalisation and reverse alignment functions ready.")
```

```python
from pytorch_grad_cam import GradCAM, GradCAMPlusPlus
from pytorch_grad_cam.utils.model_targets import ClassifierOutputTarget

target_layers = [model.layer4[-1]]

if output_features == 1:
    positive_output_index = 0
else:
    positive_output_index = 1

positive_target = [ClassifierOutputTarget(positive_output_index)]

gradcam_objects = {
    "gradcam": GradCAM(
        model=model,
        target_layers=target_layers
    ),
    "gradcampp": GradCAMPlusPlus(
        model=model,
        target_layers=target_layers
    )
}

print("Grad-CAM and Grad-CAM++ ready.")
```

```python
from lime import lime_image
from skimage.segmentation import slic

LIME_NUM_SAMPLES = 500
LIME_N_SEGMENTS = 50
LIME_COMPACTNESS = 10
LIME_SIGMA = 1

lime_explainer = lime_image.LimeImageExplainer(
    random_state=RANDOM_SEED
)

print("LIME explainer ready.")


def lime_predict_function(images):
    model.eval()

    batch_tensors = []

    for image in images:
        image = np.asarray(image)

        if image.max() <= 1.0:
            image = (image * 255).astype(np.uint8)
        else:
            image = image.astype(np.uint8)

        pil_image = Image.fromarray(image).convert("RGB")
        tensor = model_transform(pil_image)
        batch_tensors.append(tensor)

    batch_tensor = torch.stack(batch_tensors).to(device)

    with torch.no_grad():
        logits = model(batch_tensor)
        scaled_logits = logits / TEMPERATURE

        if scaled_logits.shape[1] == 1:
            pneumonia_probability = torch.sigmoid(scaled_logits)
            normal_probability = 1 - pneumonia_probability

            probabilities = torch.cat(
                [normal_probability, pneumonia_probability],
                dim=1
            )

        elif scaled_logits.shape[1] == 2:
            probabilities = torch.softmax(scaled_logits, dim=1)

        else:
            raise ValueError(f"Unexpected output shape: {scaled_logits.shape}")

    return probabilities.detach().cpu().numpy()


def lime_segmentation_function(image):
    return slic(
        image,
        n_segments=LIME_N_SEGMENTS,
        compactness=LIME_COMPACTNESS,
        sigma=LIME_SIGMA,
        start_label=0
    )


print("LIME prediction and segmentation functions ready.")
```

```python
import shap

print("SHAP version:", shap.__version__)


class CalibratedProbabilityWrapper(nn.Module):
    def __init__(self, base_model, temperature):
        super().__init__()
        self.base_model = base_model
        self.temperature = float(temperature)

    def forward(self, x):
        logits = self.base_model(x)
        scaled_logits = logits / self.temperature

        if scaled_logits.shape[1] == 1:
            pneumonia_probability = torch.sigmoid(scaled_logits)
            normal_probability = 1 - pneumonia_probability

            probabilities = torch.cat(
                [normal_probability, pneumonia_probability],
                dim=1
            )

        elif scaled_logits.shape[1] == 2:
            probabilities = torch.softmax(scaled_logits, dim=1)

        else:
            raise ValueError(f"Unexpected model output shape: {scaled_logits.shape}")

        return probabilities


shap_model = CalibratedProbabilityWrapper(
    base_model=model,
    temperature=TEMPERATURE
).to(device)

shap_model.eval()

print("SHAP calibrated probability wrapper created.")
```

```python
def extract_positive_class_shap_map(shap_values, target_index=1):
    if isinstance(shap_values, list):
        positive_values = shap_values[target_index]
    else:
        positive_values = shap_values

    positive_values = np.asarray(positive_values)
    positive_values = np.squeeze(positive_values)

    if positive_values.ndim == 4:
        if positive_values.shape[-1] == 2:
            positive_values = positive_values[..., target_index]

        elif positive_values.shape[0] == 2:
            positive_values = positive_values[target_index]

        else:
            raise ValueError(f"Unexpected 4D SHAP shape: {positive_values.shape}")

    if positive_values.ndim == 3:
        if positive_values.shape[0] == 3:
            shap_map = np.mean(np.abs(positive_values), axis=0)

        elif positive_values.shape[-1] == 3:
            shap_map = np.mean(np.abs(positive_values), axis=-1)

        else:
            shap_map = np.mean(np.abs(positive_values), axis=0)

    elif positive_values.ndim == 2:
        shap_map = np.abs(positive_values)

    else:
        raise ValueError(f"Unexpected SHAP shape after processing: {positive_values.shape}")

    return shap_map.astype(np.float32)


print("SHAP map extraction function ready.")
```

```python
SHAP_NSAMPLES = 200

def generate_xai_map_for_image(rgb_image, method):
    input_tensor = rgb_to_input_tensor(rgb_image)

    if method in ["gradcam", "gradcampp"]:
        cam_object = gradcam_objects[method]

        cam_map = cam_object(
            input_tensor=input_tensor,
            targets=positive_target
        )[0]

        return normalise_map(cam_map)

    elif method == "lime":
        explanation = lime_explainer.explain_instance(
            image=rgb_image,
            classifier_fn=lime_predict_function,
            labels=[1],
            hide_color=0,
            num_samples=LIME_NUM_SAMPLES,
            segmentation_fn=lime_segmentation_function
        )

        segments = explanation.segments

        if 1 not in explanation.local_exp:
            raise ValueError("LIME did not return label 1 explanation.")

        local_exp = dict(explanation.local_exp[1])

        lime_map = np.zeros(segments.shape, dtype=np.float32)

        for segment_id, weight in local_exp.items():
            if weight > 0:
                lime_map[segments == segment_id] = weight

        return normalise_map(lime_map)

    elif method == "shap":
        shap_values = shap_explainer.shap_values(
            input_tensor,
            nsamples=SHAP_NSAMPLES,
            rseed=RANDOM_SEED
        )

        shap_map = extract_positive_class_shap_map(
            shap_values,
            target_index=1
        )

        return normalise_map(shap_map)

    else:
        raise ValueError(f"Unknown XAI method: {method}")


print("generate_xai_map_for_image function ready.")
```

```python
def cosine_similarity_maps(map_a, map_b):
    a = map_a.flatten().astype(np.float32)
    b = map_b.flatten().astype(np.float32)

    denominator = np.linalg.norm(a) * np.linalg.norm(b)

    if denominator == 0:
        return 0.0

    return float(np.dot(a, b) / denominator)


def spearman_similarity_maps(map_a, map_b):
    a = map_a.flatten()
    b = map_b.flatten()

    if np.std(a) == 0 or np.std(b) == 0:
        return 0.0

    correlation, _ = spearmanr(a, b)

    if np.isnan(correlation):
        return 0.0

    return float(correlation)


def top_attribution_overlap(map_a, map_b, top_fraction=0.10):
    a = map_a.flatten()
    b = map_b.flatten()

    k = int(round(top_fraction * len(a)))

    if k <= 0:
        return 0.0

    top_a = set(np.argsort(a)[::-1][:k])
    top_b = set(np.argsort(b)[::-1][:k])

    overlap = len(top_a.intersection(top_b)) / k

    return float(overlap)


def calculate_stability_metrics(original_map, perturbed_map):
    original_map = normalise_map(original_map)
    perturbed_map = normalise_map(perturbed_map)

    ssim_value = ssim(
        original_map,
        perturbed_map,
        data_range=1.0
    )

    cosine_value = cosine_similarity_maps(
        original_map,
        perturbed_map
    )

    spearman_value = spearman_similarity_maps(
        original_map,
        perturbed_map
    )

    top_overlap_value = top_attribution_overlap(
        original_map,
        perturbed_map,
        top_fraction=0.10
    )

    ssim_normalised = (ssim_value + 1) / 2
    spearman_normalised = (spearman_value + 1) / 2

    stability_score = np.mean([
        ssim_normalised,
        max(0.0, min(1.0, cosine_value)),
        spearman_normalised,
        top_overlap_value
    ])

    return {
        "ssim": float(ssim_value),
        "cosine_similarity": float(cosine_value),
        "spearman_correlation": float(spearman_value),
        "top_10_percent_overlap": float(top_overlap_value),
        "stability_score": float(stability_score)
    }


print("Similarity metric functions ready.")
```

```python
def evaluate_stability_for_row(manifest_row, perturbations):
    case_id = manifest_row["xai_case_id"]
    method = manifest_row["method"]

    case_row = xai_cohort[
        xai_cohort["xai_case_id"] == case_id
    ].iloc[0]

    _, rgb_image = load_rgb_image_from_case(case_row)

    original_map = np.load(manifest_row["normalised_map_path"])
    original_map = normalise_map(original_map)

    result_rows = []

    for perturbation_name, perturbation_function in perturbations.items():
        start_time = time.time()

        perturbed_image = perturbation_function(rgb_image)

        perturbed_map = generate_xai_map_for_image(
            rgb_image=perturbed_image,
            method=method
        )

        perturbed_map = reverse_align_heatmap(
            heatmap=perturbed_map,
            perturbation_name=perturbation_name
        )

        perturbed_map = normalise_map(perturbed_map)

        metrics = calculate_stability_metrics(
            original_map=original_map,
            perturbed_map=perturbed_map
        )

        runtime_seconds = time.time() - start_time

        result_rows.append({
            "xai_case_id": case_id,
            "patientId": manifest_row["patientId"],
            "method": method,
            "outcome": manifest_row["outcome"],
            "reference_class": manifest_row["reference_class"],
            "predicted_class": manifest_row["predicted_class"],
            "calibrated_probability": manifest_row["calibrated_probability"],
            "perturbation": perturbation_name,
            "runtime_seconds": runtime_seconds,
            **metrics
        })

    return result_rows


print("evaluate_stability_for_row function ready.")
```

```python
print("Searching for DICOM files...")

dicom_files = list(BASE_DIR.rglob("*.dcm"))
dicom_index = {path.stem: path for path in dicom_files}

print("DICOM files found:", len(dicom_files))


def read_dicom_as_rgb(dicom_path):
    dicom = pydicom.dcmread(str(dicom_path))
    image = dicom.pixel_array.astype(np.float32)

    if getattr(dicom, "PhotometricInterpretation", "") == "MONOCHROME1":
        image = np.max(image) - image

    image = image - np.min(image)

    if np.max(image) > 0:
        image = image / np.max(image)

    image = (image * 255).astype(np.uint8)

    rgb_image = np.stack([image, image, image], axis=-1)

    return rgb_image


def load_rgb_image_from_case(case_row):
    patient_id = str(case_row["patientId"])

    if patient_id not in dicom_index:
        raise FileNotFoundError(f"DICOM not found for patientId: {patient_id}")

    dicom_path = dicom_index[patient_id]

    rgb_image = read_dicom_as_rgb(dicom_path)

    pil_image = Image.fromarray(rgb_image).convert("RGB")
    pil_image = pil_image.resize((IMAGE_SIZE, IMAGE_SIZE))

    rgb_resized = np.array(pil_image)

    return dicom_path, rgb_resized


def rgb_to_input_tensor(rgb_image):
    pil_image = Image.fromarray(rgb_image.astype(np.uint8)).convert("RGB")
    input_tensor = model_transform(pil_image).unsqueeze(0).to(device)

    return input_tensor


print("DICOM loading functions ready.")
```

```python
def apply_gaussian_noise(rgb_image, noise_std=8):
    noise = np.random.normal(
        loc=0,
        scale=noise_std,
        size=rgb_image.shape
    )

    noisy = rgb_image.astype(np.float32) + noise

    return np.clip(noisy, 0, 255).astype(np.uint8)


def apply_brightness_change(rgb_image, factor=1.10):
    bright = rgb_image.astype(np.float32) * factor

    return np.clip(bright, 0, 255).astype(np.uint8)


def apply_contrast_change(rgb_image, factor=1.10):
    mean = np.mean(
        rgb_image,
        axis=(0, 1),
        keepdims=True
    )

    contrast = (rgb_image.astype(np.float32) - mean) * factor + mean

    return np.clip(contrast, 0, 255).astype(np.uint8)


def apply_translation(rgb_image, shift_x=5, shift_y=5):
    h, w = rgb_image.shape[:2]

    matrix = np.float32([
        [1, 0, shift_x],
        [0, 1, shift_y]
    ])

    translated = cv2.warpAffine(
        rgb_image,
        matrix,
        (w, h),
        borderMode=cv2.BORDER_REFLECT
    )

    return translated.astype(np.uint8)


def apply_rotation(rgb_image, angle=5):
    h, w = rgb_image.shape[:2]
    center = (w // 2, h // 2)

    matrix = cv2.getRotationMatrix2D(
        center,
        angle,
        1.0
    )

    rotated = cv2.warpAffine(
        rgb_image,
        matrix,
        (w, h),
        borderMode=cv2.BORDER_REFLECT
    )

    return rotated.astype(np.uint8)


PERTURBATIONS = {
    "gaussian_noise_std8": lambda img: apply_gaussian_noise(img, noise_std=8),
    "brightness_plus_10": lambda img: apply_brightness_change(img, factor=1.10),
    "contrast_plus_10": lambda img: apply_contrast_change(img, factor=1.10),
    "translation_5px": lambda img: apply_translation(img, shift_x=5, shift_y=5),
    "rotation_5deg": lambda img: apply_rotation(img, angle=5)
}

print("Perturbations ready:")
for name in PERTURBATIONS:
    print("-", name)
```

```python
def normalise_map(raw_map):
    raw_map = np.asarray(raw_map).astype(np.float32)

    raw_map = np.nan_to_num(
        raw_map,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    raw_map = raw_map - np.min(raw_map)

    if np.max(raw_map) > 0:
        raw_map = raw_map / np.max(raw_map)

    return raw_map.astype(np.float32)


def reverse_align_heatmap(heatmap, perturbation_name):
    heatmap = np.asarray(heatmap).astype(np.float32)

    h, w = heatmap.shape[:2]

    if perturbation_name == "translation_5px":
        matrix = np.float32([
            [1, 0, -5],
            [0, 1, -5]
        ])

        aligned = cv2.warpAffine(
            heatmap,
            matrix,
            (w, h),
            borderMode=cv2.BORDER_REFLECT
        )

        return aligned.astype(np.float32)

    elif perturbation_name == "rotation_5deg":
        center = (w // 2, h // 2)

        matrix = cv2.getRotationMatrix2D(
            center,
            -5,
            1.0
        )

        aligned = cv2.warpAffine(
            heatmap,
            matrix,
            (w, h),
            borderMode=cv2.BORDER_REFLECT
        )

        return aligned.astype(np.float32)

    else:
        return heatmap


print("Normalisation and reverse alignment ready.")
```

```python
class CalibratedProbabilityWrapper(nn.Module):
    def __init__(self, base_model, temperature):
        super().__init__()
        self.base_model = base_model
        self.temperature = float(temperature)

    def forward(self, x):
        logits = self.base_model(x)
        scaled_logits = logits / self.temperature

        if scaled_logits.shape[1] == 1:
            pneumonia_probability = torch.sigmoid(scaled_logits)
            normal_probability = 1 - pneumonia_probability

            probabilities = torch.cat(
                [normal_probability, pneumonia_probability],
                dim=1
            )

        elif scaled_logits.shape[1] == 2:
            probabilities = torch.softmax(scaled_logits, dim=1)

        else:
            raise ValueError(f"Unexpected output shape: {scaled_logits.shape}")

        return probabilities


shap_model = CalibratedProbabilityWrapper(
    base_model=model,
    temperature=TEMPERATURE
).to(device)

shap_model.eval()

print("SHAP calibrated probability wrapper ready.")
```

```python
def lime_predict_function(images):
    model.eval()

    batch_tensors = []

    for image in images:
        image = np.asarray(image)

        if image.max() <= 1.0:
            image = (image * 255).astype(np.uint8)
        else:
            image = image.astype(np.uint8)

        pil_image = Image.fromarray(image).convert("RGB")
        tensor = model_transform(pil_image)

        batch_tensors.append(tensor)

    batch_tensor = torch.stack(batch_tensors).to(device)

    with torch.no_grad():
        logits = model(batch_tensor)
        scaled_logits = logits / TEMPERATURE

        if scaled_logits.shape[1] == 1:
            pneumonia_probability = torch.sigmoid(scaled_logits)
            normal_probability = 1 - pneumonia_probability

            probabilities = torch.cat(
                [normal_probability, pneumonia_probability],
                dim=1
            )

        elif scaled_logits.shape[1] == 2:
            probabilities = torch.softmax(scaled_logits, dim=1)

        else:
            raise ValueError(f"Unexpected output shape: {scaled_logits.shape}")

    return probabilities.detach().cpu().numpy()


print("LIME prediction function recreated.")
```

```python
def lime_segmentation_function(image):
    return slic(
        image,
        n_segments=LIME_N_SEGMENTS,
        compactness=LIME_COMPACTNESS,
        sigma=LIME_SIGMA,
        start_label=0
    )


print("LIME segmentation function recreated.")
```

```python
print("Searching for DICOM files...")

dicom_files = list(BASE_DIR.rglob("*.dcm"))
dicom_index = {path.stem: path for path in dicom_files}

print("DICOM files found:", len(dicom_files))


def read_dicom_as_rgb(dicom_path):
    dicom = pydicom.dcmread(str(dicom_path))
    image = dicom.pixel_array.astype(np.float32)

    if getattr(dicom, "PhotometricInterpretation", "") == "MONOCHROME1":
        image = np.max(image) - image

    image = image - np.min(image)

    if np.max(image) > 0:
        image = image / np.max(image)

    image = (image * 255).astype(np.uint8)

    rgb_image = np.stack(
        [image, image, image],
        axis=-1
    )

    return rgb_image


def load_rgb_image_from_case(case_row):
    patient_id = str(case_row["patientId"])

    if patient_id not in dicom_index:
        raise FileNotFoundError(f"DICOM not found for patientId: {patient_id}")

    dicom_path = dicom_index[patient_id]

    rgb_image = read_dicom_as_rgb(dicom_path)

    pil_image = Image.fromarray(rgb_image).convert("RGB")
    pil_image = pil_image.resize((IMAGE_SIZE, IMAGE_SIZE))

    rgb_resized = np.array(pil_image)

    return dicom_path, rgb_resized


def rgb_to_input_tensor(rgb_image):
    pil_image = Image.fromarray(
        rgb_image.astype(np.uint8)
    ).convert("RGB")

    input_tensor = model_transform(pil_image).unsqueeze(0).to(device)

    return input_tensor


print("DICOM loading functions ready.")
```

```python
def apply_gaussian_noise(rgb_image, noise_std=8):
    noise = np.random.normal(
        loc=0,
        scale=noise_std,
        size=rgb_image.shape
    )

    noisy = rgb_image.astype(np.float32) + noise

    return np.clip(noisy, 0, 255).astype(np.uint8)


def apply_brightness_change(rgb_image, factor=1.10):
    bright = rgb_image.astype(np.float32) * factor

    return np.clip(bright, 0, 255).astype(np.uint8)


def apply_contrast_change(rgb_image, factor=1.10):
    mean = np.mean(
        rgb_image,
        axis=(0, 1),
        keepdims=True
    )

    contrast = (
        rgb_image.astype(np.float32) - mean
    ) * factor + mean

    return np.clip(contrast, 0, 255).astype(np.uint8)


def apply_translation(rgb_image, shift_x=5, shift_y=5):
    h, w = rgb_image.shape[:2]

    matrix = np.float32([
        [1, 0, shift_x],
        [0, 1, shift_y]
    ])

    translated = cv2.warpAffine(
        rgb_image,
        matrix,
        (w, h),
        borderMode=cv2.BORDER_REFLECT
    )

    return translated.astype(np.uint8)


def apply_rotation(rgb_image, angle=5):
    h, w = rgb_image.shape[:2]
    center = (w // 2, h // 2)

    matrix = cv2.getRotationMatrix2D(
        center,
        angle,
        1.0
    )

    rotated = cv2.warpAffine(
        rgb_image,
        matrix,
        (w, h),
        borderMode=cv2.BORDER_REFLECT
    )

    return rotated.astype(np.uint8)


PERTURBATIONS = {
    "gaussian_noise_std8": lambda img: apply_gaussian_noise(img, noise_std=8),
    "brightness_plus_10": lambda img: apply_brightness_change(img, factor=1.10),
    "contrast_plus_10": lambda img: apply_contrast_change(img, factor=1.10),
    "translation_5px": lambda img: apply_translation(img, shift_x=5, shift_y=5),
    "rotation_5deg": lambda img: apply_rotation(img, angle=5)
}

print("Perturbations ready:")
for name in PERTURBATIONS:
    print("-", name)
```

```python
def normalise_map(raw_map):
    raw_map = np.asarray(raw_map).astype(np.float32)

    raw_map = np.nan_to_num(
        raw_map,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    raw_map = raw_map - np.min(raw_map)

    if np.max(raw_map) > 0:
        raw_map = raw_map / np.max(raw_map)

    return raw_map.astype(np.float32)


def reverse_align_heatmap(heatmap, perturbation_name):
    heatmap = np.asarray(heatmap).astype(np.float32)

    h, w = heatmap.shape[:2]

    if perturbation_name == "translation_5px":
        matrix = np.float32([
            [1, 0, -5],
            [0, 1, -5]
        ])

        aligned = cv2.warpAffine(
            heatmap,
            matrix,
            (w, h),
            borderMode=cv2.BORDER_REFLECT
        )

        return aligned.astype(np.float32)

    elif perturbation_name == "rotation_5deg":
        center = (w // 2, h // 2)

        matrix = cv2.getRotationMatrix2D(
            center,
            -5,
            1.0
        )

        aligned = cv2.warpAffine(
            heatmap,
            matrix,
            (w, h),
            borderMode=cv2.BORDER_REFLECT
        )

        return aligned.astype(np.float32)

    else:
        return heatmap


print("Normalisation and reverse alignment ready.")
```

```python
class CalibratedProbabilityWrapper(nn.Module):
    def __init__(self, base_model, temperature):
        super().__init__()

        self.base_model = base_model
        self.temperature = float(temperature)

    def forward(self, x):
        logits = self.base_model(x)
        scaled_logits = logits / self.temperature

        if scaled_logits.shape[1] == 1:
            pneumonia_probability = torch.sigmoid(scaled_logits)
            normal_probability = 1 - pneumonia_probability

            probabilities = torch.cat(
                [normal_probability, pneumonia_probability],
                dim=1
            )

        elif scaled_logits.shape[1] == 2:
            probabilities = torch.softmax(
                scaled_logits,
                dim=1
            )

        else:
            raise ValueError(
                f"Unexpected output shape: {scaled_logits.shape}"
            )

        return probabilities


shap_model = CalibratedProbabilityWrapper(
    base_model=model,
    temperature=TEMPERATURE
).to(device)

shap_model.eval()

print("SHAP calibrated probability wrapper ready.")
```

```python
def extract_positive_class_shap_map(shap_values, target_index=1):
    if isinstance(shap_values, list):
        positive_values = shap_values[target_index]
    else:
        positive_values = shap_values

    positive_values = np.asarray(positive_values)
    positive_values = np.squeeze(positive_values)

    if positive_values.ndim == 4:
        if positive_values.shape[-1] == 2:
            positive_values = positive_values[..., target_index]

        elif positive_values.shape[0] == 2:
            positive_values = positive_values[target_index]

        else:
            raise ValueError(
                f"Unexpected 4D SHAP shape: {positive_values.shape}"
            )

    if positive_values.ndim == 3:
        if positive_values.shape[0] == 3:
            shap_map = np.mean(
                np.abs(positive_values),
                axis=0
            )

        elif positive_values.shape[-1] == 3:
            shap_map = np.mean(
                np.abs(positive_values),
                axis=-1
            )

        else:
            shap_map = np.mean(
                np.abs(positive_values),
                axis=0
            )

    elif positive_values.ndim == 2:
        shap_map = np.abs(positive_values)

    else:
        raise ValueError(
            f"Unexpected SHAP shape after processing: {positive_values.shape}"
        )

    return shap_map.astype(np.float32)


def generate_shap_map_for_image(rgb_image):
    input_tensor = rgb_to_input_tensor(rgb_image)

    shap_values = shap_explainer.shap_values(
        input_tensor,
        nsamples=SHAP_NSAMPLES,
        rseed=RANDOM_SEED
    )

    shap_map = extract_positive_class_shap_map(
        shap_values,
        target_index=1
    )

    del input_tensor, shap_values
    gc.collect()

    return normalise_map(shap_map)


print("SHAP map extraction and generation functions ready.")
```

```python
def cosine_similarity_maps(map_a, map_b):
    a = map_a.flatten().astype(np.float32)
    b = map_b.flatten().astype(np.float32)

    denominator = np.linalg.norm(a) * np.linalg.norm(b)

    if denominator == 0:
        return 0.0

    return float(np.dot(a, b) / denominator)


def spearman_similarity_maps(map_a, map_b):
    a = map_a.flatten()
    b = map_b.flatten()

    if np.std(a) == 0 or np.std(b) == 0:
        return 0.0

    correlation, _ = spearmanr(a, b)

    if np.isnan(correlation):
        return 0.0

    return float(correlation)


def top_attribution_overlap(map_a, map_b, top_fraction=0.10):
    a = map_a.flatten()
    b = map_b.flatten()

    k = int(round(top_fraction * len(a)))

    if k <= 0:
        return 0.0

    top_a = set(np.argsort(a)[::-1][:k])
    top_b = set(np.argsort(b)[::-1][:k])

    overlap = len(top_a.intersection(top_b)) / k

    return float(overlap)


def calculate_stability_metrics(original_map, perturbed_map):
    original_map = normalise_map(original_map)
    perturbed_map = normalise_map(perturbed_map)

    ssim_value = ssim(
        original_map,
        perturbed_map,
        data_range=1.0
    )

    cosine_value = cosine_similarity_maps(
        original_map,
        perturbed_map
    )

    spearman_value = spearman_similarity_maps(
        original_map,
        perturbed_map
    )

    top_overlap_value = top_attribution_overlap(
        original_map,
        perturbed_map,
        top_fraction=0.10
    )

    ssim_normalised = (ssim_value + 1) / 2
    spearman_normalised = (spearman_value + 1) / 2

    stability_score = np.mean([
        ssim_normalised,
        max(0.0, min(1.0, cosine_value)),
        spearman_normalised,
        top_overlap_value
    ])

    return {
        "ssim": float(ssim_value),
        "cosine_similarity": float(cosine_value),
        "spearman_correlation": float(spearman_value),
        "top_10_percent_overlap": float(top_overlap_value),
        "stability_score": float(stability_score)
    }


print("Stability metric functions ready.")
```

```python
def evaluate_shap_stability_for_row(manifest_row, perturbations):
    case_id = manifest_row["xai_case_id"]

    case_row = xai_cohort[
        xai_cohort["xai_case_id"] == case_id
    ].iloc[0]

    _, rgb_image = load_rgb_image_from_case(case_row)

    # Regenerate original SHAP map using the same SHAP_NSAMPLES = 100.
    # This avoids comparing older SHAP maps with a different sampling setting.
    original_map = generate_shap_map_for_image(rgb_image)
    original_map = normalise_map(original_map)

    result_rows = []

    for perturbation_name, perturbation_function in perturbations.items():
        start_time = time.time()

        perturbed_image = perturbation_function(rgb_image)

        perturbed_map = generate_shap_map_for_image(
            perturbed_image
        )

        perturbed_map = reverse_align_heatmap(
            heatmap=perturbed_map,
            perturbation_name=perturbation_name
        )

        perturbed_map = normalise_map(perturbed_map)

        metrics = calculate_stability_metrics(
            original_map=original_map,
            perturbed_map=perturbed_map
        )

        runtime_seconds = time.time() - start_time

        result_rows.append({
            "xai_case_id": case_id,
            "patientId": manifest_row["patientId"],
            "method": "shap",
            "outcome": manifest_row["outcome"],
            "reference_class": manifest_row["reference_class"],
            "predicted_class": manifest_row["predicted_class"],
            "calibrated_probability": manifest_row["calibrated_probability"],
            "perturbation": perturbation_name,
            "runtime_seconds": runtime_seconds,
            "shap_nsamples": SHAP_NSAMPLES,
            "shap_background_size": int(shap_background.shape[0]),
            "original_map_regenerated_for_stability": True,
            **metrics
        })

        gc.collect()

    del original_map
    gc.collect()

    return result_rows


print("SHAP stability evaluator ready.")
```

```python
print("Searching for DICOM files...")

dicom_files = list(BASE_DIR.rglob("*.dcm"))
dicom_index = {path.stem: path for path in dicom_files}

print("DICOM files found:", len(dicom_files))


def read_dicom_as_rgb(dicom_path):
    dicom = pydicom.dcmread(str(dicom_path))
    image = dicom.pixel_array.astype(np.float32)

    if getattr(dicom, "PhotometricInterpretation", "") == "MONOCHROME1":
        image = np.max(image) - image

    image = image - np.min(image)

    if np.max(image) > 0:
        image = image / np.max(image)

    image = (image * 255).astype(np.uint8)

    rgb_image = np.stack(
        [image, image, image],
        axis=-1
    )

    return rgb_image


def load_rgb_image_from_case(case_row):
    patient_id = str(case_row["patientId"])

    if patient_id not in dicom_index:
        raise FileNotFoundError(f"DICOM not found for patientId: {patient_id}")

    dicom_path = dicom_index[patient_id]

    rgb_image = read_dicom_as_rgb(dicom_path)

    pil_image = Image.fromarray(rgb_image).convert("RGB")
    pil_image = pil_image.resize((IMAGE_SIZE, IMAGE_SIZE))

    rgb_resized = np.array(pil_image)

    return dicom_path, rgb_resized


def rgb_to_input_tensor(rgb_image):
    pil_image = Image.fromarray(
        rgb_image.astype(np.uint8)
    ).convert("RGB")

    input_tensor = model_transform(pil_image).unsqueeze(0).to(device)

    return input_tensor


print("DICOM loading functions ready.")
```

```python
def apply_gaussian_noise(rgb_image, noise_std=8):
    noise = np.random.normal(
        loc=0,
        scale=noise_std,
        size=rgb_image.shape
    )

    noisy = rgb_image.astype(np.float32) + noise

    return np.clip(noisy, 0, 255).astype(np.uint8)


def apply_brightness_change(rgb_image, factor=1.10):
    bright = rgb_image.astype(np.float32) * factor

    return np.clip(bright, 0, 255).astype(np.uint8)


def apply_contrast_change(rgb_image, factor=1.10):
    mean = np.mean(
        rgb_image,
        axis=(0, 1),
        keepdims=True
    )

    contrast = (
        rgb_image.astype(np.float32) - mean
    ) * factor + mean

    return np.clip(contrast, 0, 255).astype(np.uint8)


def apply_translation(rgb_image, shift_x=5, shift_y=5):
    h, w = rgb_image.shape[:2]

    matrix = np.float32([
        [1, 0, shift_x],
        [0, 1, shift_y]
    ])

    translated = cv2.warpAffine(
        rgb_image,
        matrix,
        (w, h),
        borderMode=cv2.BORDER_REFLECT
    )

    return translated.astype(np.uint8)


def apply_rotation(rgb_image, angle=5):
    h, w = rgb_image.shape[:2]
    center = (w // 2, h // 2)

    matrix = cv2.getRotationMatrix2D(
        center,
        angle,
        1.0
    )

    rotated = cv2.warpAffine(
        rgb_image,
        matrix,
        (w, h),
        borderMode=cv2.BORDER_REFLECT
    )

    return rotated.astype(np.uint8)


PERTURBATIONS = {
    "gaussian_noise_std8": lambda img: apply_gaussian_noise(img, noise_std=8),
    "brightness_plus_10": lambda img: apply_brightness_change(img, factor=1.10),
    "contrast_plus_10": lambda img: apply_contrast_change(img, factor=1.10),
    "translation_5px": lambda img: apply_translation(img, shift_x=5, shift_y=5),
    "rotation_5deg": lambda img: apply_rotation(img, angle=5)
}

print("Perturbations ready:")
for name in PERTURBATIONS:
    print("-", name)
```

```python
def normalise_map(raw_map):
    raw_map = np.asarray(raw_map).astype(np.float32)

    raw_map = np.nan_to_num(
        raw_map,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    raw_map = raw_map - np.min(raw_map)

    if np.max(raw_map) > 0:
        raw_map = raw_map / np.max(raw_map)

    return raw_map.astype(np.float32)


def reverse_align_heatmap(heatmap, perturbation_name):
    heatmap = np.asarray(heatmap).astype(np.float32)

    h, w = heatmap.shape[:2]

    if perturbation_name == "translation_5px":
        matrix = np.float32([
            [1, 0, -5],
            [0, 1, -5]
        ])

        aligned = cv2.warpAffine(
            heatmap,
            matrix,
            (w, h),
            borderMode=cv2.BORDER_REFLECT
        )

        return aligned.astype(np.float32)

    elif perturbation_name == "rotation_5deg":
        center = (w // 2, h // 2)

        matrix = cv2.getRotationMatrix2D(
            center,
            -5,
            1.0
        )

        aligned = cv2.warpAffine(
            heatmap,
            matrix,
            (w, h),
            borderMode=cv2.BORDER_REFLECT
        )

        return aligned.astype(np.float32)

    else:
        return heatmap


print("Normalisation and reverse alignment ready.")
```

```python
class CalibratedProbabilityWrapper(nn.Module):
    def __init__(self, base_model, temperature):
        super().__init__()

        self.base_model = base_model
        self.temperature = float(temperature)

    def forward(self, x):
        logits = self.base_model(x)
        scaled_logits = logits / self.temperature

        if scaled_logits.shape[1] == 1:
            pneumonia_probability = torch.sigmoid(scaled_logits)
            normal_probability = 1 - pneumonia_probability

            probabilities = torch.cat(
                [normal_probability, pneumonia_probability],
                dim=1
            )

        elif scaled_logits.shape[1] == 2:
            probabilities = torch.softmax(
                scaled_logits,
                dim=1
            )

        else:
            raise ValueError(
                f"Unexpected output shape: {scaled_logits.shape}"
            )

        return probabilities


shap_model = CalibratedProbabilityWrapper(
    base_model=model,
    temperature=TEMPERATURE
).to(device)

shap_model.eval()

print("SHAP calibrated probability wrapper ready.")
```

```python
def extract_positive_class_shap_map(shap_values, target_index=1):
    if isinstance(shap_values, list):
        positive_values = shap_values[target_index]
    else:
        positive_values = shap_values

    positive_values = np.asarray(positive_values)
    positive_values = np.squeeze(positive_values)

    if positive_values.ndim == 4:
        if positive_values.shape[-1] == 2:
            positive_values = positive_values[..., target_index]

        elif positive_values.shape[0] == 2:
            positive_values = positive_values[target_index]

        else:
            raise ValueError(
                f"Unexpected 4D SHAP shape: {positive_values.shape}"
            )

    if positive_values.ndim == 3:
        if positive_values.shape[0] == 3:
            shap_map = np.mean(
                np.abs(positive_values),
                axis=0
            )

        elif positive_values.shape[-1] == 3:
            shap_map = np.mean(
                np.abs(positive_values),
                axis=-1
            )

        else:
            shap_map = np.mean(
                np.abs(positive_values),
                axis=0
            )

    elif positive_values.ndim == 2:
        shap_map = np.abs(positive_values)

    else:
        raise ValueError(
            f"Unexpected SHAP shape after processing: {positive_values.shape}"
        )

    return shap_map.astype(np.float32)


def generate_shap_map_for_image(rgb_image):
    input_tensor = rgb_to_input_tensor(rgb_image)

    shap_values = shap_explainer.shap_values(
        input_tensor,
        nsamples=SHAP_NSAMPLES,
        rseed=RANDOM_SEED
    )

    shap_map = extract_positive_class_shap_map(
        shap_values,
        target_index=1
    )

    del input_tensor, shap_values
    gc.collect()

    return normalise_map(shap_map)


print("SHAP map extraction and generation functions ready.")
```

```python
def cosine_similarity_maps(map_a, map_b):
    a = map_a.flatten().astype(np.float32)
    b = map_b.flatten().astype(np.float32)

    denominator = np.linalg.norm(a) * np.linalg.norm(b)

    if denominator == 0:
        return 0.0

    return float(np.dot(a, b) / denominator)


def spearman_similarity_maps(map_a, map_b):
    a = map_a.flatten()
    b = map_b.flatten()

    if np.std(a) == 0 or np.std(b) == 0:
        return 0.0

    correlation, _ = spearmanr(a, b)

    if np.isnan(correlation):
        return 0.0

    return float(correlation)


def top_attribution_overlap(map_a, map_b, top_fraction=0.10):
    a = map_a.flatten()
    b = map_b.flatten()

    k = int(round(top_fraction * len(a)))

    if k <= 0:
        return 0.0

    top_a = set(np.argsort(a)[::-1][:k])
    top_b = set(np.argsort(b)[::-1][:k])

    overlap = len(top_a.intersection(top_b)) / k

    return float(overlap)


def calculate_stability_metrics(original_map, perturbed_map):
    original_map = normalise_map(original_map)
    perturbed_map = normalise_map(perturbed_map)

    ssim_value = ssim(
        original_map,
        perturbed_map,
        data_range=1.0
    )

    cosine_value = cosine_similarity_maps(
        original_map,
        perturbed_map
    )

    spearman_value = spearman_similarity_maps(
        original_map,
        perturbed_map
    )

    top_overlap_value = top_attribution_overlap(
        original_map,
        perturbed_map,
        top_fraction=0.10
    )

    ssim_normalised = (ssim_value + 1) / 2
    spearman_normalised = (spearman_value + 1) / 2

    stability_score = np.mean([
        ssim_normalised,
        max(0.0, min(1.0, cosine_value)),
        spearman_normalised,
        top_overlap_value
    ])

    return {
        "ssim": float(ssim_value),
        "cosine_similarity": float(cosine_value),
        "spearman_correlation": float(spearman_value),
        "top_10_percent_overlap": float(top_overlap_value),
        "stability_score": float(stability_score)
    }


print("Stability metric functions ready.")
```

```python
def evaluate_shap_stability_for_row(manifest_row, perturbations):
    case_id = manifest_row["xai_case_id"]

    case_row = xai_cohort[
        xai_cohort["xai_case_id"] == case_id
    ].iloc[0]

    _, rgb_image = load_rgb_image_from_case(case_row)

    print(f"\nStarting SHAP stability for {case_id}")

    original_start = time.time()

    original_map = generate_shap_map_for_image(rgb_image)
    original_map = normalise_map(original_map)

    print(
        f"{case_id} | original SHAP map completed | "
        f"{round(time.time() - original_start, 2)} seconds"
    )

    result_rows = []

    for perturbation_name, perturbation_function in perturbations.items():
        start_time = time.time()

        print(f"{case_id} | starting perturbation: {perturbation_name}")

        perturbed_image = perturbation_function(rgb_image)

        perturbed_map = generate_shap_map_for_image(
            perturbed_image
        )

        perturbed_map = reverse_align_heatmap(
            heatmap=perturbed_map,
            perturbation_name=perturbation_name
        )

        perturbed_map = normalise_map(perturbed_map)

        metrics = calculate_stability_metrics(
            original_map=original_map,
            perturbed_map=perturbed_map
        )

        runtime_seconds = time.time() - start_time

        print(
            f"{case_id} | {perturbation_name} completed | "
            f"{round(runtime_seconds, 2)} seconds"
        )

        result_rows.append({
            "xai_case_id": case_id,
            "patientId": manifest_row["patientId"],
            "method": "shap",
            "outcome": manifest_row["outcome"],
            "reference_class": manifest_row["reference_class"],
            "predicted_class": manifest_row["predicted_class"],
            "calibrated_probability": manifest_row["calibrated_probability"],
            "perturbation": perturbation_name,
            "runtime_seconds": runtime_seconds,
            "shap_nsamples": SHAP_NSAMPLES,
            "shap_background_size": int(shap_background.shape[0]),
            "original_map_regenerated_for_stability": True,
            **metrics
        })

        gc.collect()

    del original_map
    gc.collect()

    print(f"{case_id} | all SHAP perturbations completed")

    return result_rows


print("SHAP stability evaluator ready.")
```
