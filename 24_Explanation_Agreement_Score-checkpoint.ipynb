```python
def normalise_01(arr):
    arr = np.asarray(arr, dtype=np.float32)
    
    if arr.ndim == 3:
        arr = arr.mean(axis=2)
    
    min_val = np.nanmin(arr)
    max_val = np.nanmax(arr)
    
    if max_val - min_val < 1e-8:
        return np.zeros_like(arr, dtype=np.float32)
    
    return (arr - min_val) / (max_val - min_val)


def load_heatmap(path, target_size=(224, 224)):
    """
    Loads .npy or image heatmap and returns 2D 0-1 array.
    """
    path = Path(path)
    
    if path.suffix.lower() == ".npy":
        arr = np.load(path)
        arr = np.squeeze(arr)
        
        if arr.ndim == 3:
            arr = arr.mean(axis=2)
        
        arr = normalise_01(arr)
        
        img = Image.fromarray((arr * 255).astype(np.uint8))
        img = img.resize(target_size, Image.BILINEAR)
        arr = np.asarray(img).astype(np.float32) / 255.0
        
        return normalise_01(arr)
    
    else:
        img = Image.open(path).convert("L")
        img = img.resize(target_size, Image.BILINEAR)
        arr = np.asarray(img).astype(np.float32) / 255.0
        
        return normalise_01(arr)
```

```python
from scipy.stats import spearmanr

def cosine_similarity_map(a, b):
    a = a.flatten().astype(np.float32)
    b = b.flatten().astype(np.float32)
    
    denom = np.linalg.norm(a) * np.linalg.norm(b)
    
    if denom < 1e-8:
        return np.nan
    
    value = np.dot(a, b) / denom
    return float(np.clip(value, -1, 1))


def spearman_similarity_map(a, b):
    a = a.flatten()
    b = b.flatten()
    
    if np.std(a) < 1e-8 or np.std(b) < 1e-8:
        return np.nan
    
    rho, p = spearmanr(a, b)
    return float(rho)


def topk_overlap(a, b, top_percent=10):
    a = a.flatten()
    b = b.flatten()
    
    n = len(a)
    k = int(np.ceil(n * top_percent / 100))
    
    top_a = np.argsort(a)[-k:]
    top_b = np.argsort(b)[-k:]
    
    overlap = len(set(top_a).intersection(set(top_b))) / k
    return float(overlap)


def ssim_similarity_map(a, b):
    if not SKIMAGE_AVAILABLE:
        return np.nan
    
    return float(ssim(a, b, data_range=1.0))


def safe_01(value):
    if pd.isna(value):
        return np.nan
    return float(np.clip(value, 0, 1))


def spearman_to_01(rho):
    if pd.isna(rho):
        return np.nan
    return (rho + 1) / 2
```

```python
pairwise_rows = []
case_map_cache = {}

for _, case_row in survey_cases.iterrows():
    case_id = case_row["xai_case_id"]
    outcome = case_row["outcome"] if "outcome" in case_row.index else None
    
    maps = {}
    
    for method in methods:
        file_path = find_heatmap_file_exact(case_id, method)
        
        if file_path is None:
            print(f"Missing map: {case_id} - {method}")
            continue
        
        maps[method] = load_heatmap(file_path)
    
    case_map_cache[case_id] = maps
    
    for i in range(len(methods)):
        for j in range(i + 1, len(methods)):
            m1 = methods[i]
            m2 = methods[j]
            
            if m1 not in maps or m2 not in maps:
                continue
            
            a = maps[m1]
            b = maps[m2]
            
            ssim_val = ssim_similarity_map(a, b)
            cosine_val = cosine_similarity_map(a, b)
            spearman_val = spearman_similarity_map(a, b)
            top10_val = topk_overlap(a, b, top_percent=10)
            
            ssim_01 = safe_01(ssim_val)
            cosine_01 = (cosine_val + 1) / 2 if not pd.isna(cosine_val) else np.nan
            spearman_01 = spearman_to_01(spearman_val)
            top10_01 = safe_01(top10_val)
            
            composite = np.nanmean([
                ssim_01,
                cosine_01,
                spearman_01,
                top10_01
            ])
            
            pairwise_rows.append({
                "xai_case_id": case_id,
                "outcome": outcome,
                "method_1": m1,
                "method_2": m2,
                "pair": f"{m1} vs {m2}",
                "SSIM": ssim_val,
                "Cosine": cosine_val,
                "Spearman": spearman_val,
                "Top10_Overlap": top10_val,
                "SSIM_01": ssim_01,
                "Cosine_01": cosine_01,
                "Spearman_01": spearman_01,
                "Top10_Overlap_01": top10_01,
                "Pairwise_EAS": composite
            })

pairwise_eas_df = pd.DataFrame(pairwise_rows)

print("Pairwise EAS rows:", pairwise_eas_df.shape)
display(pairwise_eas_df.head())
```

```python
case_eas_df = pairwise_eas_df.groupby(["xai_case_id", "outcome"]).agg(
    EAS_SSIM=("SSIM_01", "mean"),
    EAS_Cosine=("Cosine_01", "mean"),
    EAS_Spearman=("Spearman_01", "mean"),
    EAS_Top10=("Top10_Overlap_01", "mean"),
    EAS_Composite=("Pairwise_EAS", "mean"),
    pairwise_comparisons=("Pairwise_EAS", "count")
).reset_index()

case_eas_df["EAS_Percentage"] = case_eas_df["EAS_Composite"] * 100

display(case_eas_df)
display(case_eas_df.describe())
```

```python
eas_by_outcome = case_eas_df.groupby("outcome").agg(
    mean_EAS=("EAS_Percentage", "mean"),
    median_EAS=("EAS_Percentage", "median"),
    sd_EAS=("EAS_Percentage", "std"),
    min_EAS=("EAS_Percentage", "min"),
    max_EAS=("EAS_Percentage", "max"),
    n_cases=("EAS_Percentage", "count")
).reset_index()

display(eas_by_outcome)
```

```python
merged_df = analysis_df.merge(
    case_eas_df[["xai_case_id", "EAS_Composite", "EAS_Percentage"]],
    left_on="xai_case_id_standard",
    right_on="xai_case_id",
    how="left"
)

print("Merged shape:", merged_df.shape)
display(merged_df.head())
```

```python
def spearman_table(df, x_col, y_cols):
    rows = []
    
    for y_col in y_cols:
        temp = df[[x_col, y_col]].dropna()
        
        if len(temp) < 3:
            rho, p = np.nan, np.nan
        else:
            rho, p = spearmanr(temp[x_col], temp[y_col])
        
        rows.append({
            "x": x_col,
            "y": y_col,
            "n": len(temp),
            "spearman_rho": rho,
            "p_value": p
        })
    
    return pd.DataFrame(rows)


possible_y_cols = [
    "HTS",
    "CTS_baseline",
    "CTS_TAI_optimised",
    "TAI_TAI_optimised",
    "TrustGap_TAI_optimised",
    "Abs_TrustGap_TAI_optimised"
]

available_y_cols = [c for c in possible_y_cols if c in merged_df.columns]

eas_correlation_df = spearman_table(
    merged_df,
    x_col="EAS_Percentage",
    y_cols=available_y_cols
)

display(eas_correlation_df)
```

```python
merged_df["CTS_EAS_extended"] = merged_df[
    ["confidence", "robustness", "faithfulness", "stability", "EAS_Percentage"]
].mean(axis=1)

merged_df["TrustGap_EAS_extended"] = merged_df["HTS"] - merged_df["CTS_EAS_extended"]
merged_df["Abs_TrustGap_EAS_extended"] = merged_df["TrustGap_EAS_extended"].abs()
merged_df["TAI_EAS_extended"] = 100 - merged_df["Abs_TrustGap_EAS_extended"]

display(merged_df.head())
```

```python
def make_alignment_version(df, cts_col, version_name):
    temp = df.copy()
    temp["CTS_version"] = version_name
    temp["CTS"] = temp[cts_col]
    temp["TrustGap"] = temp["HTS"] - temp["CTS"]
    temp["Abs_TrustGap"] = temp["TrustGap"].abs()
    temp["TAI"] = 100 - temp["Abs_TrustGap"]
    return temp


versions = []

versions.append(make_alignment_version(merged_df, "CTS_baseline", "Baseline equal-weight CTS"))

if "CTS_regression_derived" in merged_df.columns:
    versions.append(make_alignment_version(merged_df, "CTS_regression_derived", "Regression-derived CTS"))

if "CTS_TAI_optimised" in merged_df.columns:
    versions.append(make_alignment_version(merged_df, "CTS_TAI_optimised", "TAI-optimised CTS"))

versions.append(make_alignment_version(merged_df, "CTS_EAS_extended", "EAS-extended CTS"))

eas_cts_comparison_rows = pd.concat(versions, ignore_index=True)

eas_cts_comparison_summary = eas_cts_comparison_rows.groupby("CTS_version").agg(
    mean_CTS=("CTS", "mean"),
    mean_HTS=("HTS", "mean"),
    mean_TrustGap=("TrustGap", "mean"),
    mean_abs_TrustGap=("Abs_TrustGap", "mean"),
    mean_TAI=("TAI", "mean"),
    median_TAI=("TAI", "median"),
    min_TAI=("TAI", "min"),
    max_TAI=("TAI", "max"),
    n=("TAI", "count")
).reset_index()

display(eas_cts_comparison_summary)
```

```python
case_eas_df["Explanation_Disagreement_Warning"] = case_eas_df["EAS_Percentage"] < 50

display(case_eas_df[[
    "xai_case_id",
    "outcome",
    "EAS_Percentage",
    "Explanation_Disagreement_Warning"
]])

warning_summary = case_eas_df.groupby("Explanation_Disagreement_Warning").size().reset_index(name="count")
warning_summary["percentage"] = warning_summary["count"] / warning_summary["count"].sum() * 100

display(warning_summary)
```
