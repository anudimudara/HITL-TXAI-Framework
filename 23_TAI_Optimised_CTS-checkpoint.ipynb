```python
import pandas as pd
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt

from scipy.optimize import minimize
from sklearn.utils import resample

# -----------------------------
# Project paths
# -----------------------------
PROJECT_ROOT = Path(r"C:\Users\anudi\Downloads\MSc_Research")

# Use the clean file generated from Regression-derived CTS notebook
REGRESSION_FILE = (
    PROJECT_ROOT 
    / "Advanced_Framework_Analysis" 
    / "Regression_Derived_CTS" 
    / "Regression_Derived_CTS_Results.xlsx"
)

# Output folder for TAI-optimised CTS
OUT_DIR = PROJECT_ROOT / "Advanced_Framework_Analysis" / "TAI_Optimised_CTS"
OUT_DIR.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
print("Regression results file exists:", REGRESSION_FILE.exists())
print("Output folder:", OUT_DIR)
```

```python
# -----------------------------
# Load clean 32-row CTS-HTS data
# -----------------------------
analysis_df = pd.read_excel(REGRESSION_FILE, sheet_name="Clean_Analysis_Data")

print("Shape:", analysis_df.shape)
display(analysis_df.head())

print("\nColumns:")
for c in analysis_df.columns:
    print(c)
```

```python
required_cols = [
    "confidence",
    "robustness",
    "faithfulness",
    "stability",
    "HTS",
    "CTS_baseline"
]

missing_cols = [c for c in required_cols if c not in analysis_df.columns]

if missing_cols:
    raise ValueError(f"Missing columns: {missing_cols}")

predictor_cols = ["confidence", "robustness", "faithfulness", "stability"]

# Make sure values are numeric
for col in predictor_cols + ["HTS", "CTS_baseline"]:
    analysis_df[col] = pd.to_numeric(analysis_df[col], errors="coerce")

analysis_df = analysis_df.dropna(subset=predictor_cols + ["HTS", "CTS_baseline"]).reset_index(drop=True)

print("Final dataset shape:", analysis_df.shape)
display(analysis_df[predictor_cols + ["HTS", "CTS_baseline"]].describe())
```

```python
def calculate_alignment_metrics(df, cts_col, label):
    """
    Calculates CTS, TrustGap, Abs TrustGap and TAI for a CTS version.
    """
    temp = df.copy()
    temp["CTS_version"] = label
    temp["CTS"] = temp[cts_col]
    temp["TrustGap"] = temp["HTS"] - temp["CTS"]
    temp["Abs_TrustGap"] = temp["TrustGap"].abs()
    temp["TAI"] = 100 - temp["Abs_TrustGap"]
    return temp


def summarise_alignment(rows_df):
    """
    Summarises alignment metrics.
    """
    return rows_df.groupby("CTS_version").agg(
        mean_CTS=("CTS", "mean"),
        median_CTS=("CTS", "median"),
        mean_HTS=("HTS", "mean"),
        mean_TrustGap=("TrustGap", "mean"),
        median_TrustGap=("TrustGap", "median"),
        mean_abs_TrustGap=("Abs_TrustGap", "mean"),
        median_abs_TrustGap=("Abs_TrustGap", "median"),
        mean_TAI=("TAI", "mean"),
        median_TAI=("TAI", "median"),
        min_TAI=("TAI", "min"),
        max_TAI=("TAI", "max"),
        n=("TAI", "count")
    ).reset_index()
```

```python
baseline_rows = calculate_alignment_metrics(
    analysis_df,
    cts_col="CTS_baseline",
    label="Baseline equal-weight CTS"
)

baseline_summary = summarise_alignment(baseline_rows)

display(baseline_summary)
```

```python
X = analysis_df[predictor_cols].values
y_hts = analysis_df["HTS"].values

def objective_mean_absolute_trustgap(weights, X, y):
    """
    Objective function:
    minimise mean absolute TrustGap = mean(|HTS - CTS_weighted|)
    """
    cts_weighted = np.dot(X, weights)
    mean_abs_trustgap = np.mean(np.abs(y - cts_weighted))
    return mean_abs_trustgap


# Initial equal weights
initial_weights = np.array([0.25, 0.25, 0.25, 0.25])

# Bounds: each weight must be between 0 and 1
bounds = [(0, 1) for _ in predictor_cols]

# Constraint: all weights must sum to 1
constraints = {
    "type": "eq",
    "fun": lambda w: np.sum(w) - 1
}

print("Initial objective value:")
print(objective_mean_absolute_trustgap(initial_weights, X, y_hts))
```

```python
optimisation_result = minimize(
    objective_mean_absolute_trustgap,
    initial_weights,
    args=(X, y_hts),
    method="SLSQP",
    bounds=bounds,
    constraints=constraints,
    options={"maxiter": 1000}
)

print("Optimisation success:", optimisation_result.success)
print("Message:", optimisation_result.message)

optimised_weights = optimisation_result.x

weights_opt_df = pd.DataFrame({
    "component": predictor_cols,
    "optimised_weight": optimised_weights
})

display(weights_opt_df)

print("\nOptimised weights:")
for comp, weight in zip(predictor_cols, optimised_weights):
    print(f"{comp}: {weight:.4f}")

print("\nSum of weights:", optimised_weights.sum())
print("Optimised mean absolute TrustGap:", optimisation_result.fun)
```

```python
analysis_df["CTS_TAI_optimised"] = np.dot(
    analysis_df[predictor_cols].values,
    optimised_weights
)

analysis_df["TrustGap_TAI_optimised"] = analysis_df["HTS"] - analysis_df["CTS_TAI_optimised"]
analysis_df["Abs_TrustGap_TAI_optimised"] = analysis_df["TrustGap_TAI_optimised"].abs()
analysis_df["TAI_TAI_optimised"] = 100 - analysis_df["Abs_TrustGap_TAI_optimised"]

display(analysis_df.head())
```

```python
all_versions = []

# Baseline
all_versions.append(
    calculate_alignment_metrics(
        analysis_df,
        "CTS_baseline",
        "Baseline equal-weight CTS"
    )
)

# Regression-derived CTS, if available
if "CTS_regression_derived" in analysis_df.columns:
    all_versions.append(
        calculate_alignment_metrics(
            analysis_df,
            "CTS_regression_derived",
            "Regression-derived CTS"
        )
    )

# TAI-optimised CTS
all_versions.append(
    calculate_alignment_metrics(
        analysis_df,
        "CTS_TAI_optimised",
        "TAI-optimised CTS"
    )
)

comparison_rows = pd.concat(all_versions, ignore_index=True)
comparison_summary = summarise_alignment(comparison_rows)

display(comparison_summary)
```

```python
method_comparison = comparison_rows.groupby(["CTS_version", "method"]).agg(
    mean_CTS=("CTS", "mean"),
    mean_HTS=("HTS", "mean"),
    mean_TrustGap=("TrustGap", "mean"),
    mean_abs_TrustGap=("Abs_TrustGap", "mean"),
    mean_TAI=("TAI", "mean"),
    median_TAI=("TAI", "median"),
    n=("TAI", "count")
).reset_index()

display(method_comparison)
```

```python
if "outcome" in comparison_rows.columns:
    outcome_comparison = comparison_rows.groupby(["CTS_version", "outcome"]).agg(
        mean_CTS=("CTS", "mean"),
        mean_HTS=("HTS", "mean"),
        mean_TrustGap=("TrustGap", "mean"),
        mean_abs_TrustGap=("Abs_TrustGap", "mean"),
        mean_TAI=("TAI", "mean"),
        median_TAI=("TAI", "median"),
        n=("TAI", "count")
    ).reset_index()

    display(outcome_comparison)
else:
    outcome_comparison = pd.DataFrame()
    print("No outcome column found.")
```

```python
def assign_tai_band(tai):
    if tai >= 85:
        return "Well-aligned"
    elif tai >= 70:
        return "Moderate misalignment"
    else:
        return "Significant misalignment"


comparison_rows["TAI_band"] = comparison_rows["TAI"].apply(assign_tai_band)

tai_band_summary = comparison_rows.groupby(["CTS_version", "TAI_band"]).size().reset_index(name="count")

tai_band_summary["percentage"] = tai_band_summary.groupby("CTS_version")["count"].transform(
    lambda x: x / x.sum() * 100
)

display(tai_band_summary)
```

```python
def assign_cts_overestimation_risk(row):
    if "outcome" not in row.index:
        return False
    
    return (
        row["CTS"] >= 70 and
        row["HTS"] < 40 and
        str(row["outcome"]).upper() in ["FP", "FN"]
    )


comparison_rows["CTS_overestimation_risk"] = comparison_rows.apply(
    assign_cts_overestimation_risk,
    axis=1
)

risk_summary = comparison_rows.groupby("CTS_version").agg(
    risk_count=("CTS_overestimation_risk", "sum"),
    total_outputs=("CTS_overestimation_risk", "count")
).reset_index()

risk_summary["risk_percentage"] = (
    risk_summary["risk_count"] / risk_summary["total_outputs"] * 100
)

display(risk_summary)
```

```python
N_BOOT = 1000
rng = np.random.default_rng(42)

bootstrap_weights = []
bootstrap_metrics = []

for i in range(N_BOOT):
    sample_idx = rng.choice(analysis_df.index, size=len(analysis_df), replace=True)
    boot_df = analysis_df.loc[sample_idx].reset_index(drop=True)
    
    X_boot = boot_df[predictor_cols].values
    y_boot = boot_df["HTS"].values
    
    try:
        result = minimize(
            objective_mean_absolute_trustgap,
            initial_weights,
            args=(X_boot, y_boot),
            method="SLSQP",
            bounds=bounds,
            constraints=constraints,
            options={"maxiter": 1000}
        )
        
        if result.success:
            w = result.x
            bootstrap_weights.append(w)
            
            cts_boot = np.dot(X_boot, w)
            trustgap_boot = y_boot - cts_boot
            tai_boot = 100 - np.abs(trustgap_boot)
            
            bootstrap_metrics.append({
                "mean_abs_TrustGap": np.mean(np.abs(trustgap_boot)),
                "mean_TAI": np.mean(tai_boot),
                "median_TAI": np.median(tai_boot)
            })
            
    except Exception:
        continue

bootstrap_weights_df = pd.DataFrame(
    bootstrap_weights,
    columns=predictor_cols
)

bootstrap_metrics_df = pd.DataFrame(bootstrap_metrics)

bootstrap_weight_summary = bootstrap_weights_df.agg(["mean", "std"]).T
bootstrap_weight_summary["ci_lower_2_5"] = bootstrap_weights_df.quantile(0.025)
bootstrap_weight_summary["ci_upper_97_5"] = bootstrap_weights_df.quantile(0.975)
bootstrap_weight_summary = bootstrap_weight_summary.reset_index().rename(columns={"index": "component"})

bootstrap_metric_summary = bootstrap_metrics_df.agg(["mean", "std"]).T
bootstrap_metric_summary["ci_lower_2_5"] = bootstrap_metrics_df.quantile(0.025)
bootstrap_metric_summary["ci_upper_97_5"] = bootstrap_metrics_df.quantile(0.975)
bootstrap_metric_summary = bootstrap_metric_summary.reset_index().rename(columns={"index": "metric"})

display(bootstrap_weight_summary)
display(bootstrap_metric_summary)
```
