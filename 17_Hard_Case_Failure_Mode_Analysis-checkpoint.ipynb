```python
from pathlib import Path
import pandas as pd
import numpy as np
import json
from datetime import datetime

# ============================================================
# HARD-CASE, FAILURE-MODE AND DEFERRAL ANALYSIS
# ============================================================

BASE_DIR = Path(r"C:\Users\anudi\Downloads\MSc_Research")
MODEL_THRESHOLD = 0.35

OUTPUT_DIR = BASE_DIR / "xai_analysis" / "hard_case_failure_mode_deferral_analysis" / "results"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Output folder:")
print(OUTPUT_DIR)
```

```python
def find_file(base_dir, filename):
    matches = list(base_dir.rglob(filename))
    if len(matches) == 0:
        raise FileNotFoundError(f"Could not find {filename} under {base_dir}")
    return matches[0]


def standardise_method_column(df):
    """
    Ensures the dataframe has a column called 'method'.
    """
    possible_method_cols = ["method", "xai_method", "explanation_method"]
    
    for col in possible_method_cols:
        if col in df.columns:
            if col != "method":
                df = df.rename(columns={col: "method"})
            return df
    
    raise ValueError("No method column found. Expected one of: method, xai_method, explanation_method")


def method_clean(x):
    x = str(x).lower().replace("-", "").replace("_", "").replace(" ", "")
    
    if x in ["gradcam"]:
        return "gradcam"
    if x in ["gradcampp", "gradcamplusplus"]:
        return "gradcampp"
    if x == "lime":
        return "lime"
    if x == "shap":
        return "shap"
    
    return x


def method_label(x):
    labels = {
        "gradcam": "Grad-CAM",
        "gradcampp": "Grad-CAM++",
        "lime": "LIME",
        "shap": "SHAP"
    }
    return labels.get(str(x).lower(), str(x))


def minmax(series):
    """
    Normalises a numeric series to 0–1.
    If all values are the same, returns 0.5 for all rows.
    """
    series = pd.to_numeric(series, errors="coerce")
    min_val = series.min()
    max_val = series.max()
    
    if pd.isna(min_val) or pd.isna(max_val) or max_val == min_val:
        return pd.Series(np.ones(len(series)) * 0.5, index=series.index)
    
    return (series - min_val) / (max_val - min_val)


def confidence_from_threshold(probability, threshold=0.35):
    """
    Converts calibrated probability into confidence based on distance from threshold.
    This is not model accuracy. It shows how far the prediction is from the decision boundary.
    """
    probability = float(probability)
    
    if probability >= threshold:
        return (probability - threshold) / (1 - threshold)
    else:
        return (threshold - probability) / threshold


def confidence_band(confidence_score):
    if pd.isna(confidence_score):
        return "Unknown"
    elif confidence_score < 0.25:
        return "Low confidence / near-threshold"
    elif confidence_score < 0.60:
        return "Medium confidence"
    else:
        return "High confidence"


def safe_quantile(series, q, default=None):
    series = pd.to_numeric(series, errors="coerce").dropna()
    if len(series) == 0:
        return default
    return float(series.quantile(q))


def make_flag_text(row):
    flags = []
    
    flag_columns = [
        "flag_high_confidence_fp",
        "flag_high_confidence_fn",
        "flag_low_confidence_prediction",
        "flag_low_prediction_robustness",
        "flag_low_faithfulness",
        "flag_low_stability",
        "flag_weak_localisation",
        "flag_correct_prediction_weak_explanation",
        "flag_incorrect_prediction_stable_explanation"
    ]
    
    labels = {
        "flag_high_confidence_fp": "High-confidence false positive",
        "flag_high_confidence_fn": "High-confidence false negative",
        "flag_low_confidence_prediction": "Low-confidence / near-threshold prediction",
        "flag_low_prediction_robustness": "Low prediction robustness",
        "flag_low_faithfulness": "Low explanation faithfulness",
        "flag_low_stability": "Low explanation stability",
        "flag_weak_localisation": "Weak localisation evidence",
        "flag_correct_prediction_weak_explanation": "Correct prediction with weak explanation",
        "flag_incorrect_prediction_stable_explanation": "Incorrect prediction with stable explanation"
    }
    
    for col in flag_columns:
        if col in row and row[col] == 1:
            flags.append(labels[col])
    
    if len(flags) == 0:
        return "No major hard-case flag"
    
    return "; ".join(flags)
```

```python
# Find files automatically
full_manifest_path = find_file(BASE_DIR, "full_xai_manifest.csv")
faithfulness_path = find_file(BASE_DIR, "xai_faithfulness_results.csv")
stability_path = find_file(BASE_DIR, "case_method_explanation_stability.csv")
robustness_path = find_file(BASE_DIR, "prediction_robustness_results.csv")
localisation_path = find_file(BASE_DIR, "xai_localisation_results.csv")
cohort_path = find_file(BASE_DIR, "xai_evaluation_cohort.csv")

print("Files found:")
print("Full XAI manifest:", full_manifest_path)
print("Faithfulness:", faithfulness_path)
print("Stability:", stability_path)
print("Prediction robustness:", robustness_path)
print("Localisation:", localisation_path)
print("XAI cohort:", cohort_path)

# Load files
manifest = pd.read_csv(full_manifest_path)
faithfulness = pd.read_csv(faithfulness_path)
stability = pd.read_csv(stability_path)
robustness = pd.read_csv(robustness_path)
localisation = pd.read_csv(localisation_path)
cohort = pd.read_csv(cohort_path)

# Standardise method column
manifest = standardise_method_column(manifest)
faithfulness = standardise_method_column(faithfulness)
stability = standardise_method_column(stability)
localisation = standardise_method_column(localisation)

# Clean method names
manifest["method"] = manifest["method"].apply(method_clean)
faithfulness["method"] = faithfulness["method"].apply(method_clean)
stability["method"] = stability["method"].apply(method_clean)
localisation["method"] = localisation["method"].apply(method_clean)

print("\nRows loaded:")
print("manifest:", len(manifest))
print("faithfulness:", len(faithfulness))
print("stability:", len(stability))
print("robustness:", len(robustness))
print("localisation:", len(localisation))
print("cohort:", len(cohort))

print("\nManifest columns:")
print(list(manifest.columns))
```

```python
faithfulness_summary = faithfulness.copy()

if "faithfulness_score" not in faithfulness_summary.columns:
    score_parts = []
    
    if "aopc" in faithfulness_summary.columns:
        faithfulness_summary["aopc_norm"] = faithfulness_summary.groupby("method")["aopc"].transform(minmax)
        score_parts.append("aopc_norm")
    
    if "insertion_auc" in faithfulness_summary.columns:
        faithfulness_summary["insertion_auc_norm"] = faithfulness_summary.groupby("method")["insertion_auc"].transform(minmax)
        score_parts.append("insertion_auc_norm")
    
    if "deletion_auc" in faithfulness_summary.columns:
        faithfulness_summary["deletion_auc_inverse_norm"] = (
            1 - faithfulness_summary.groupby("method")["deletion_auc"].transform(minmax)
        )
        score_parts.append("deletion_auc_inverse_norm")
    
    if len(score_parts) == 0:
        raise ValueError("Could not create faithfulness_score. Check faithfulness columns.")
    
    faithfulness_summary["faithfulness_score"] = faithfulness_summary[score_parts].mean(axis=1)

faithfulness_summary = faithfulness_summary.groupby(
    ["xai_case_id", "method"], as_index=False
).agg(
    faithfulness_score=("faithfulness_score", "mean")
)

print("Faithfulness summary rows:", len(faithfulness_summary))
display(faithfulness_summary.head())
```

```python
stability_summary = stability.copy()

if "explanation_stability_score" not in stability_summary.columns:
    if "stability_score" in stability_summary.columns:
        stability_summary = stability_summary.rename(
            columns={"stability_score": "explanation_stability_score"}
        )
    else:
        raise ValueError("No explanation_stability_score or stability_score column found.")

agg_dict = {
    "explanation_stability_score": ("explanation_stability_score", "mean")
}

if "mean_ssim" in stability_summary.columns:
    agg_dict["mean_ssim"] = ("mean_ssim", "mean")

if "mean_cosine_similarity" in stability_summary.columns:
    agg_dict["mean_cosine_similarity"] = ("mean_cosine_similarity", "mean")

if "mean_spearman_correlation" in stability_summary.columns:
    agg_dict["mean_spearman_correlation"] = ("mean_spearman_correlation", "mean")

if "mean_top_10_percent_overlap" in stability_summary.columns:
    agg_dict["mean_top_10_percent_overlap"] = ("mean_top_10_percent_overlap", "mean")

stability_summary = stability_summary.groupby(
    ["xai_case_id", "method"], as_index=False
).agg(**agg_dict)

print("Stability summary rows:", len(stability_summary))
display(stability_summary.head())
```

```python
localisation_summary = localisation.copy()

if "localisation_score" not in localisation_summary.columns:
    localisation_score_parts = []
    
    possible_cols = [
        "attribution_energy_inside_region",
        "energy_inside_region",
        "energy_inside_bbox",
        "pointing_game",
        "pointing_game_accuracy",
        "pointing_game_hit",
        "iou",
        "dice"
    ]
    
    for col in possible_cols:
        if col in localisation_summary.columns:
            new_col = col + "_norm"
            localisation_summary[new_col] = localisation_summary.groupby("method")[col].transform(minmax)
            localisation_score_parts.append(new_col)
    
    if len(localisation_score_parts) == 0:
        raise ValueError("Could not create localisation_score. Check localisation columns.")
    
    localisation_summary["localisation_score"] = localisation_summary[localisation_score_parts].mean(axis=1)

localisation_summary = localisation_summary.groupby(
    ["xai_case_id", "method"], as_index=False
).agg(
    localisation_score=("localisation_score", "mean")
)

print("Localisation summary rows:", len(localisation_summary))
display(localisation_summary.head())
```

```python
robustness_summary = robustness.copy()

print("Robustness columns:")
print(list(robustness_summary.columns))

# Identify absolute probability change column
possible_abs_cols = [
    "absolute_probability_change",
    "abs_probability_change",
    "probability_change_abs",
    "abs_change",
    "mean_abs_probability_change"
]

abs_change_col = None
for col in possible_abs_cols:
    if col in robustness_summary.columns:
        abs_change_col = col
        break

# Identify prediction change column
possible_flip_cols = [
    "prediction_changed",
    "label_changed",
    "class_changed",
    "threshold_crossed"
]

flip_col = None
for col in possible_flip_cols:
    if col in robustness_summary.columns:
        flip_col = col
        break

if flip_col is None and "prediction_preserved" in robustness_summary.columns:
    robustness_summary["prediction_changed_generated"] = 1 - robustness_summary["prediction_preserved"].astype(int)
    flip_col = "prediction_changed_generated"

if flip_col is None and "label_preserved" in robustness_summary.columns:
    robustness_summary["prediction_changed_generated"] = 1 - robustness_summary["label_preserved"].astype(int)
    flip_col = "prediction_changed_generated"

# Create case-level robustness summary
if "prediction_robustness_score" in robustness_summary.columns:
    robustness_case_summary = robustness_summary.groupby(
        "xai_case_id", as_index=False
    ).agg(
        prediction_robustness_score=("prediction_robustness_score", "mean")
    )
    
    if abs_change_col is not None:
        temp_abs = robustness_summary.groupby("xai_case_id", as_index=False).agg(
            mean_abs_probability_change=(abs_change_col, "mean"),
            max_abs_probability_change=(abs_change_col, "max")
        )
        robustness_case_summary = robustness_case_summary.merge(temp_abs, on="xai_case_id", how="left")
    else:
        robustness_case_summary["mean_abs_probability_change"] = np.nan
        robustness_case_summary["max_abs_probability_change"] = np.nan

else:
    if abs_change_col is None:
        raise ValueError(
            "Could not create prediction robustness score. "
            "No prediction_robustness_score or probability-change column found."
        )
    
    robustness_case_summary = robustness_summary.groupby(
        "xai_case_id", as_index=False
    ).agg(
        mean_abs_probability_change=(abs_change_col, "mean"),
        max_abs_probability_change=(abs_change_col, "max")
    )
    
    robustness_case_summary["prediction_robustness_score"] = (
        1 - robustness_case_summary["mean_abs_probability_change"]
    ).clip(0, 1)

# Add prediction flip count if available
if flip_col is not None:
    flip_summary = robustness_summary.groupby("xai_case_id", as_index=False).agg(
        prediction_flip_count=(flip_col, "sum"),
        prediction_preservation_rate=(flip_col, lambda x: 1 - np.mean(x.astype(int)))
    )
    robustness_case_summary = robustness_case_summary.merge(flip_summary, on="xai_case_id", how="left")
else:
    robustness_case_summary["prediction_flip_count"] = np.nan
    robustness_case_summary["prediction_preservation_rate"] = np.nan

print("Robustness case summary rows:", len(robustness_case_summary))
display(robustness_case_summary.head())
```

```python
analysis = manifest.copy()
analysis["method"] = analysis["method"].apply(method_clean)

# Merge faithfulness
analysis = analysis.merge(
    faithfulness_summary,
    on=["xai_case_id", "method"],
    how="left"
)

# Merge stability
analysis = analysis.merge(
    stability_summary,
    on=["xai_case_id", "method"],
    how="left"
)

# Merge localisation
analysis = analysis.merge(
    localisation_summary,
    on=["xai_case_id", "method"],
    how="left"
)

# Merge prediction robustness
analysis = analysis.merge(
    robustness_case_summary,
    on="xai_case_id",
    how="left"
)

# Check calibrated probability column
if "calibrated_probability" not in analysis.columns:
    raise ValueError("calibrated_probability column not found in full_xai_manifest.csv")

# Create confidence score and confidence band
analysis["confidence_score"] = analysis["calibrated_probability"].apply(
    lambda x: confidence_from_threshold(x, MODEL_THRESHOLD)
)

analysis["confidence_band"] = analysis["confidence_score"].apply(confidence_band)

# Fill important missing values
analysis["faithfulness_score"] = analysis["faithfulness_score"].fillna(0)
analysis["explanation_stability_score"] = analysis["explanation_stability_score"].fillna(0)

if analysis["prediction_robustness_score"].isna().all():
    analysis["prediction_robustness_score"] = 0.5
else:
    analysis["prediction_robustness_score"] = analysis["prediction_robustness_score"].fillna(
        analysis["prediction_robustness_score"].mean()
    )

# Localisation remains NaN for cases without bounding boxes
analysis["method_label"] = analysis["method"].apply(method_label)

# Save evidence table
evidence_path = OUTPUT_DIR / "case_method_evidence_table.csv"
analysis.to_csv(evidence_path, index=False)

print("Evidence table saved:")
print(evidence_path)
print("Rows:", len(analysis))
print("Expected rows: 480")

display(analysis.head())
```

```python
faithfulness_q25 = safe_quantile(analysis["faithfulness_score"], 0.25, default=0.25)
stability_q25 = safe_quantile(analysis["explanation_stability_score"], 0.25, default=0.25)
stability_q75 = safe_quantile(analysis["explanation_stability_score"], 0.75, default=0.75)
robustness_q25 = safe_quantile(analysis["prediction_robustness_score"], 0.25, default=0.75)

if "localisation_score" in analysis.columns:
    localisation_q25 = safe_quantile(analysis["localisation_score"], 0.25, default=np.nan)
else:
    localisation_q25 = np.nan

thresholds = {
    "model_threshold": MODEL_THRESHOLD,
    "low_confidence_threshold": 0.25,
    "high_confidence_threshold": 0.60,
    "faithfulness_low_q25": faithfulness_q25,
    "stability_low_q25": stability_q25,
    "stability_high_q75": stability_q75,
    "robustness_low_q25": robustness_q25,
    "localisation_low_q25": localisation_q25
}

print("Thresholds used:")
print(json.dumps(thresholds, indent=4))
```

```python
df = analysis.copy()
df["outcome_upper"] = df["outcome"].astype(str).str.upper()

# Prediction-related flags
df["flag_high_confidence_fp"] = (
    (df["outcome_upper"] == "FP") &
    (df["confidence_score"] >= thresholds["high_confidence_threshold"])
).astype(int)

df["flag_high_confidence_fn"] = (
    (df["outcome_upper"] == "FN") &
    (df["confidence_score"] >= thresholds["high_confidence_threshold"])
).astype(int)

df["flag_low_confidence_prediction"] = (
    df["confidence_score"] < thresholds["low_confidence_threshold"]
).astype(int)

df["flag_low_prediction_robustness"] = (
    df["prediction_robustness_score"] <= thresholds["robustness_low_q25"]
).astype(int)

# Explanation-related flags
df["flag_low_faithfulness"] = (
    df["faithfulness_score"] <= thresholds["faithfulness_low_q25"]
).astype(int)

df["flag_low_stability"] = (
    df["explanation_stability_score"] <= thresholds["stability_low_q25"]
).astype(int)

# Localisation flag only where localisation exists
if pd.notna(thresholds["localisation_low_q25"]):
    df["flag_weak_localisation"] = (
        df["localisation_score"].notna() &
        (df["localisation_score"] <= thresholds["localisation_low_q25"])
    ).astype(int)
else:
    df["flag_weak_localisation"] = 0

# Combined flags
df["flag_correct_prediction_weak_explanation"] = (
    df["outcome_upper"].isin(["TP", "TN"]) &
    (
        (df["flag_low_faithfulness"] == 1) |
        (df["flag_low_stability"] == 1) |
        (df["flag_weak_localisation"] == 1)
    )
).astype(int)

df["flag_incorrect_prediction_stable_explanation"] = (
    df["outcome_upper"].isin(["FP", "FN"]) &
    (df["explanation_stability_score"] >= thresholds["stability_high_q75"])
).astype(int)

flag_columns = [
    "flag_high_confidence_fp",
    "flag_high_confidence_fn",
    "flag_low_confidence_prediction",
    "flag_low_prediction_robustness",
    "flag_low_faithfulness",
    "flag_low_stability",
    "flag_weak_localisation",
    "flag_correct_prediction_weak_explanation",
    "flag_incorrect_prediction_stable_explanation"
]

df["failure_flag_count"] = df[flag_columns].sum(axis=1)
df["failure_mode_summary"] = df.apply(make_flag_text, axis=1)

# Research-level technical review flag only
df["technical_review_flag"] = np.where(
    df["failure_flag_count"] >= 2,
    "Review candidate",
    "No major review flag"
)

df["technical_review_flag"] = np.where(
    (df["flag_high_confidence_fn"] == 1) |
    (df["flag_high_confidence_fp"] == 1),
    "High-priority review candidate",
    df["technical_review_flag"]
)

# Save failure modes
failure_modes_path = OUTPUT_DIR / "case_method_failure_modes.csv"
df.to_csv(failure_modes_path, index=False)

print("Failure modes saved:")
print(failure_modes_path)
print("Rows:", len(df))
print("Expected rows: 480")

display(
    df[
        [
            "xai_case_id",
            "patientId",
            "outcome",
            "method_label",
            "calibrated_probability",
            "confidence_score",
            "confidence_band",
            "prediction_robustness_score",
            "faithfulness_score",
            "explanation_stability_score",
            "localisation_score",
            "failure_flag_count",
            "technical_review_flag",
            "failure_mode_summary"
        ]
    ].head(12)
)
```

```python
case_level = df.groupby("xai_case_id", as_index=False).agg(
    patientId=("patientId", "first"),
    outcome=("outcome", "first"),
    reference_class=("reference_class", "first"),
    predicted_class=("predicted_class", "first"),
    calibrated_probability=("calibrated_probability", "first"),
    confidence_score=("confidence_score", "first"),
    confidence_band=("confidence_band", "first"),
    prediction_robustness_score=("prediction_robustness_score", "first"),
    mean_abs_probability_change=("mean_abs_probability_change", "first") if "mean_abs_probability_change" in df.columns else ("prediction_robustness_score", "first"),
    max_abs_probability_change=("max_abs_probability_change", "first") if "max_abs_probability_change" in df.columns else ("prediction_robustness_score", "first"),
    prediction_flip_count=("prediction_flip_count", "first") if "prediction_flip_count" in df.columns else ("prediction_robustness_score", "first"),
    mean_faithfulness_score=("faithfulness_score", "mean"),
    mean_explanation_stability_score=("explanation_stability_score", "mean"),
    mean_localisation_score=("localisation_score", "mean"),
    total_failure_flags=("failure_flag_count", "sum"),
    max_failure_flags_single_method=("failure_flag_count", "max"),
    methods_with_review_flag=("technical_review_flag", lambda x: int((x != "No major review flag").sum())),
    methods_high_priority=("technical_review_flag", lambda x: int((x == "High-priority review candidate").sum())),
    low_faithfulness_methods=("flag_low_faithfulness", "sum"),
    low_stability_methods=("flag_low_stability", "sum"),
    weak_localisation_methods=("flag_weak_localisation", "sum")
)

case_level["outcome_upper"] = case_level["outcome"].astype(str).str.upper()

def assign_case_priority(row):
    if row["outcome_upper"] == "FN" and row["confidence_score"] >= thresholds["high_confidence_threshold"]:
        return "High"
    
    if row["outcome_upper"] == "FP" and row["confidence_score"] >= thresholds["high_confidence_threshold"]:
        return "High"
    
    if row["methods_high_priority"] >= 1:
        return "High"
    
    if row["outcome_upper"] in ["FP", "FN"] and row["prediction_robustness_score"] <= thresholds["robustness_low_q25"]:
        return "High"
    
    if row["total_failure_flags"] >= 8:
        return "High"
    
    if row["confidence_score"] < thresholds["low_confidence_threshold"]:
        return "Medium"
    
    if row["outcome_upper"] in ["FP", "FN"]:
        return "Medium"
    
    if row["methods_with_review_flag"] >= 2:
        return "Medium"
    
    if row["prediction_robustness_score"] <= thresholds["robustness_low_q25"]:
        return "Medium"
    
    if row["total_failure_flags"] >= 4:
        return "Medium"
    
    return "Low"


case_level["hard_case_priority"] = case_level.apply(assign_case_priority, axis=1)

def case_level_reason(row):
    reasons = []
    
    if row["outcome_upper"] == "FN":
        reasons.append("False negative case")
    
    if row["outcome_upper"] == "FP":
        reasons.append("False positive case")
    
    if row["confidence_score"] >= thresholds["high_confidence_threshold"] and row["outcome_upper"] in ["FP", "FN"]:
        reasons.append("High-confidence error")
    
    if row["confidence_score"] < thresholds["low_confidence_threshold"]:
        reasons.append("Low-confidence / near-threshold prediction")
    
    if row["prediction_robustness_score"] <= thresholds["robustness_low_q25"]:
        reasons.append("Low prediction robustness")
    
    if row["low_faithfulness_methods"] >= 2:
        reasons.append("Low faithfulness in multiple XAI methods")
    
    if row["low_stability_methods"] >= 2:
        reasons.append("Low stability in multiple XAI methods")
    
    if row["weak_localisation_methods"] >= 2:
        reasons.append("Weak localisation in multiple XAI methods")
    
    if len(reasons) == 0:
        return "No major case-level hard-case issue"
    
    return "; ".join(reasons)

case_level["hard_case_reason"] = case_level.apply(case_level_reason, axis=1)

# Save case-level summary
case_level_path = OUTPUT_DIR / "case_level_hard_case_summary.csv"
case_level.to_csv(case_level_path, index=False)

print("Case-level hard-case summary saved:")
print(case_level_path)
print("Rows:", len(case_level))
print("Expected rows: 120")

display(
    case_level[
        [
            "xai_case_id",
            "patientId",
            "outcome",
            "calibrated_probability",
            "confidence_score",
            "confidence_band",
            "prediction_robustness_score",
            "mean_faithfulness_score",
            "mean_explanation_stability_score",
            "mean_localisation_score",
            "total_failure_flags",
            "methods_with_review_flag",
            "hard_case_priority",
            "hard_case_reason"
        ]
    ].head(12)
)
```

```python
priority_order = {
    "High": 1,
    "Medium": 2,
    "Low": 3
}

top_hard_cases = case_level.copy()
top_hard_cases["priority_order"] = top_hard_cases["hard_case_priority"].map(priority_order)

top_hard_cases = top_hard_cases.sort_values(
    by=[
        "priority_order",
        "total_failure_flags",
        "methods_with_review_flag",
        "confidence_score"
    ],
    ascending=[True, False, False, False]
)

top_hard_cases_path = OUTPUT_DIR / "top_hard_cases_for_review.csv"
top_hard_cases.to_csv(top_hard_cases_path, index=False)

print("Top hard cases saved:")
print(top_hard_cases_path)

display(
    top_hard_cases[
        [
            "xai_case_id",
            "patientId",
            "outcome",
            "calibrated_probability",
            "confidence_score",
            "confidence_band",
            "prediction_robustness_score",
            "total_failure_flags",
            "methods_with_review_flag",
            "hard_case_priority",
            "hard_case_reason"
        ]
    ].head(20)
)
```
