```python
validation_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("Validation transformation created.")
```

```python
val_dataset = datasets.ImageFolder(
    root=VAL_DIR,
    transform=validation_transform
)

print("Validation images:", len(val_dataset))
print("Classes:", val_dataset.classes)
print("Class mapping:", val_dataset.class_to_idx)
```

```python
BATCH_SIZE = 16

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

print("Validation batches:", len(val_loader))
```

```python
model = models.densenet121(
    weights=None
)

number_of_features = model.classifier.in_features

model.classifier = nn.Linear(
    number_of_features,
    2
)

model = model.to(device)

print(model.classifier)
```

```python
try:
    state_dict = torch.load(
        CHECKPOINT_PATH,
        map_location=device,
        weights_only=True
    )
except TypeError:
    state_dict = torch.load(
        CHECKPOINT_PATH,
        map_location=device
    )

model.load_state_dict(state_dict)

model.eval()

print("DenseNet121 checkpoint loaded successfully.")
print("Checkpoint:", CHECKPOINT_PATH)
```

```python
all_true_labels = []
all_pneumonia_probabilities = []

start_time = time.time()

model.eval()

with torch.no_grad():

    for images, labels in tqdm(
        val_loader,
        desc="Evaluating DenseNet121"
    ):

        images = images.to(device)
        labels = labels.to(device)

        outputs = model(images)

        probabilities = torch.softmax(
            outputs,
            dim=1
        )

        pneumonia_probabilities = probabilities[:, 1]

        all_true_labels.extend(
            labels.cpu().numpy()
        )

        all_pneumonia_probabilities.extend(
            pneumonia_probabilities.cpu().numpy()
        )

end_time = time.time()

y_true = np.asarray(
    all_true_labels,
    dtype=int
)

y_prob = np.asarray(
    all_pneumonia_probabilities,
    dtype=float
)

elapsed_time = end_time - start_time

print()
print("Predictions generated:", len(y_true))
print(f"Total inference time: {elapsed_time:.2f} seconds")
print(
    f"Average inference time per image: "
    f"{elapsed_time / len(y_true):.4f} seconds"
)
```

```python
def calculate_ece(
    y_true,
    y_prob,
    n_bins=10
):
    bin_edges = np.linspace(
        0.0,
        1.0,
        n_bins + 1
    )

    bin_numbers = np.digitize(
        y_prob,
        bin_edges[1:-1]
    )

    total_ece = 0.0
    calibration_rows = []

    for bin_index in range(n_bins):

        bin_mask = bin_numbers == bin_index
        bin_count = int(bin_mask.sum())

        if bin_count == 0:
            continue

        mean_probability = float(
            y_prob[bin_mask].mean()
        )

        observed_frequency = float(
            y_true[bin_mask].mean()
        )

        bin_weight = bin_count / len(y_true)

        absolute_error = abs(
            observed_frequency - mean_probability
        )

        total_ece += (
            bin_weight * absolute_error
        )

        calibration_rows.append({
            "bin": bin_index + 1,
            "lower_boundary": bin_edges[bin_index],
            "upper_boundary": bin_edges[bin_index + 1],
            "number_of_samples": bin_count,
            "mean_predicted_probability": mean_probability,
            "observed_pneumonia_frequency": observed_frequency,
            "absolute_calibration_error": absolute_error
        })

    calibration_table = pd.DataFrame(
        calibration_rows
    )

    return total_ece, calibration_table
```

```python
def calculate_binary_metrics(
    y_true,
    y_prob,
    threshold=0.50
):
    y_pred = (
        y_prob >= threshold
    ).astype(int)

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    tn, fp, fn, tp = cm.ravel()

    sensitivity = (
        tp / (tp + fn)
        if (tp + fn) > 0
        else 0.0
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else 0.0
    )

    negative_predictive_value = (
        tn / (tn + fn)
        if (tn + fn) > 0
        else 0.0
    )

    false_positive_rate = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else 0.0
    )

    false_negative_rate = (
        fn / (fn + tp)
        if (fn + tp) > 0
        else 0.0
    )

    ece, _ = calculate_ece(
        y_true,
        y_prob,
        n_bins=10
    )

    metrics = {
        "threshold": float(threshold),

        "accuracy": accuracy_score(
            y_true,
            y_pred
        ),

        "balanced_accuracy": balanced_accuracy_score(
            y_true,
            y_pred
        ),

        "precision_pneumonia": precision_score(
            y_true,
            y_pred,
            zero_division=0
        ),

        "sensitivity_pneumonia": sensitivity,

        "specificity_normal": specificity,

        "f1_pneumonia": f1_score(
            y_true,
            y_pred,
            zero_division=0
        ),

        "negative_predictive_value": negative_predictive_value,

        "false_positive_rate": false_positive_rate,

        "false_negative_rate": false_negative_rate,

        "matthews_correlation_coefficient": matthews_corrcoef(
            y_true,
            y_pred
        ),

        "roc_auc": roc_auc_score(
            y_true,
            y_prob
        ),

        "average_precision_auprc": average_precision_score(
            y_true,
            y_prob
        ),

        "brier_score": brier_score_loss(
            y_true,
            y_prob
        ),

        "log_loss": log_loss(
            y_true,
            np.column_stack([
                1 - y_prob,
                y_prob
            ]),
            labels=[0, 1]
        ),

        "expected_calibration_error": ece,

        "true_negatives": int(tn),
        "false_positives": int(fp),
        "false_negatives": int(fn),
        "true_positives": int(tp)
    }

    return metrics, y_pred
```

```python
threshold_rows = []

threshold_values = np.arange(
    0.05,
    0.951,
    0.01
)

for threshold in threshold_values:

    threshold_metrics, _ = calculate_binary_metrics(
        y_true,
        y_prob,
        threshold=float(threshold)
    )

    threshold_metrics["youden_j"] = (
        threshold_metrics["sensitivity_pneumonia"]
        + threshold_metrics["specificity_normal"]
        - 1
    )

    threshold_rows.append(
        threshold_metrics
    )

threshold_df = pd.DataFrame(
    threshold_rows
)

display(
    threshold_df[
        [
            "threshold",
            "accuracy",
            "precision_pneumonia",
            "sensitivity_pneumonia",
            "specificity_normal",
            "f1_pneumonia",
            "balanced_accuracy",
            "youden_j"
        ]
    ].head(10)
)
```

```python
best_threshold_index = threshold_df[
    "youden_j"
].idxmax()

best_threshold_row = threshold_df.loc[
    best_threshold_index
]

selected_threshold = float(
    best_threshold_row["threshold"]
)

print(
    f"Selected threshold: "
    f"{selected_threshold:.2f}"
)

print()
print("Metrics at selected threshold:")

print(
    best_threshold_row[
        [
            "accuracy",
            "precision_pneumonia",
            "sensitivity_pneumonia",
            "specificity_normal",
            "f1_pneumonia",
            "balanced_accuracy",
            "youden_j"
        ]
    ]
)
```

```python
high_sensitivity_candidates = threshold_df[
    threshold_df["sensitivity_pneumonia"] >= 0.90
].copy()

if len(high_sensitivity_candidates) > 0:

    high_sensitivity_row = (
        high_sensitivity_candidates
        .sort_values(
            by=[
                "specificity_normal",
                "f1_pneumonia"
            ],
            ascending=False
        )
        .iloc[0]
    )

    print("High-sensitivity operating point:")
    print(
        high_sensitivity_row[
            [
                "threshold",
                "sensitivity_pneumonia",
                "specificity_normal",
                "precision_pneumonia",
                "f1_pneumonia"
            ]
        ]
    )

else:
    print(
        "No threshold achieved pneumonia "
        "sensitivity of at least 0.90."
    )
```
