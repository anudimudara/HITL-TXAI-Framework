```python
def load_checkpoint(path):
    try:
        return torch.load(path, map_location=device, weights_only=False)
    except TypeError:
        return torch.load(path, map_location=device)


def extract_state_dict(checkpoint):
    if isinstance(checkpoint, dict):
        for key in ["model_state_dict", "state_dict", "model"]:
            if key in checkpoint and isinstance(checkpoint[key], dict):
                return checkpoint[key]

        if checkpoint and all(torch.is_tensor(v) for v in checkpoint.values()):
            return checkpoint

    raise ValueError("Could not extract model state dictionary.")


checkpoint = load_checkpoint(model_checkpoint_path)
state_dict = extract_state_dict(checkpoint)

clean_state_dict = {
    key.replace("module.", ""): value
    for key, value in state_dict.items()
}

# Detect final output size from checkpoint
fc_weight_keys = [key for key in clean_state_dict.keys() if key.endswith("fc.weight")]

if len(fc_weight_keys) == 0:
    raise ValueError("Could not find fc.weight in checkpoint.")

fc_weight_key = fc_weight_keys[0]
output_features = clean_state_dict[fc_weight_key].shape[0]

print("Detected output features:", output_features)

model = resnet50(weights=None)
model.fc = nn.Linear(model.fc.in_features, output_features)

model.load_state_dict(clean_state_dict, strict=True)
model = model.to(device)
model.eval()

print("Frozen ResNet50 loaded successfully.")
print("Final layer:", model.fc)
print("Grad-CAM target layer:", model.layer4[-1])
```

```python
model_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


def read_dicom_as_uint8(dicom_path):
    ds = pydicom.dcmread(str(dicom_path))
    image = ds.pixel_array.astype(np.float32)

    slope = float(getattr(ds, "RescaleSlope", 1))
    intercept = float(getattr(ds, "RescaleIntercept", 0))
    image = image * slope + intercept

    photometric = getattr(ds, "PhotometricInterpretation", "")

    if photometric == "MONOCHROME1":
        image = np.max(image) - image

    image = image - np.min(image)

    if np.max(image) > 0:
        image = image / np.max(image)

    image = (image * 255).astype(np.uint8)

    return image


def load_image_for_xai(row):
    possible_path_columns = [
        "image_path",
        "processed_image_path",
        "png_path",
        "jpg_path",
        "dicom_path",
        "dcm_path"
    ]

    selected_path = None

    for col in possible_path_columns:
        if col in row.index and pd.notna(row[col]):
            candidate_path = Path(str(row[col]))
            if candidate_path.exists():
                selected_path = candidate_path
                break

    if selected_path is None:
        raise FileNotFoundError(
            f"No valid image path found for patientId={row['patientId']}. "
            f"Available columns: {row.index.tolist()}"
        )

    if selected_path.suffix.lower() == ".dcm":
        image = read_dicom_as_uint8(selected_path)
        pil_image = Image.fromarray(image).convert("RGB")
    else:
        pil_image = Image.open(selected_path).convert("RGB")

    pil_image_224 = pil_image.resize((224, 224))
    display_rgb_224 = np.array(pil_image_224).astype(np.uint8)
    input_tensor = model_transform(pil_image_224).unsqueeze(0).to(device)

    return pil_image_224, display_rgb_224, input_tensor


def predict_calibrated_probability(input_tensor):
    with torch.no_grad():
        logits = model(input_tensor)

    if logits.shape[1] == 1:
        scaled_logits = logits / TEMPERATURE
        pneumonia_probability = torch.sigmoid(scaled_logits).item()
    elif logits.shape[1] == 2:
        scaled_logits = logits / TEMPERATURE
        probabilities = torch.softmax(scaled_logits, dim=1)
        pneumonia_probability = probabilities[:, 1].item()
    else:
        raise ValueError(f"Unexpected model output shape: {logits.shape}")

    predicted_label = int(pneumonia_probability >= LOCKED_THRESHOLD)

    return pneumonia_probability, predicted_label
```

```python
verification_rows = []

for _, row in xai_pilot_cohort.iterrows():
    _, _, input_tensor = load_image_for_xai(row)

    reproduced_probability, reproduced_label = predict_calibrated_probability(input_tensor)

    saved_probability = float(row["calibrated_probability"])

    if "predicted_label" in row.index:
        saved_label = int(row["predicted_label"])
    elif "predicted_class" in row.index:
        saved_label = 1 if str(row["predicted_class"]).lower() in ["pneumonia", "lung opacity", "1"] else 0
    else:
        saved_label = int(saved_probability >= LOCKED_THRESHOLD)

    verification_rows.append({
        "pilot_case_id": row["pilot_case_id"],
        "xai_case_id": row["xai_case_id"],
        "patientId": row["patientId"],
        "outcome": row["outcome"],
        "saved_probability": saved_probability,
        "reproduced_probability": reproduced_probability,
        "absolute_probability_difference": abs(saved_probability - reproduced_probability),
        "saved_label": saved_label,
        "reproduced_label": reproduced_label,
        "label_match": saved_label == reproduced_label
    })

verification_df = pd.DataFrame(verification_rows)

display(verification_df)

verification_path = pilot_folder / "xai_pilot_prediction_verification.csv"
verification_df.to_csv(verification_path, index=False)

print("Saved verification file:")
print(verification_path)

print("\nMaximum probability difference:")
print(verification_df["absolute_probability_difference"].max())

print("\nAll labels match:")
print(verification_df["label_match"].all())
```

```python
from pytorch_grad_cam import GradCAM, GradCAMPlusPlus
from pytorch_grad_cam.utils.model_targets import ClassifierOutputTarget

import matplotlib.pyplot as plt
import cv2
from skimage.transform import resize
```

```python
def normalise_map(attribution_map):
    """
    Converts any attribution map into a clean 0-1 range.
    """
    attribution_map = np.asarray(attribution_map, dtype=np.float32)

    attribution_map = np.nan_to_num(
        attribution_map,
        nan=0.0,
        posinf=0.0,
        neginf=0.0
    )

    attribution_map = attribution_map - attribution_map.min()

    if attribution_map.max() > 0:
        attribution_map = attribution_map / attribution_map.max()

    return attribution_map.astype(np.float32)


def overlay_heatmap(rgb_image, heatmap, alpha=0.35):
    """
    Creates a heatmap overlay for visual checking.
    This is only for visual inspection; technical evaluation will use raw/normalised maps.
    """
    rgb_image = np.asarray(rgb_image).astype(np.uint8)
    heatmap = normalise_map(heatmap)

    heatmap_uint8 = np.uint8(255 * heatmap)

    heatmap_colour = cv2.applyColorMap(
        heatmap_uint8,
        cv2.COLORMAP_JET
    )

    heatmap_colour = cv2.cvtColor(
        heatmap_colour,
        cv2.COLOR_BGR2RGB
    )

    overlay = cv2.addWeighted(
        rgb_image,
        1 - alpha,
        heatmap_colour,
        alpha,
        0
    )

    return overlay


def save_map_and_overlay(case_id, method_name, raw_map, rgb_image):
    """
    Saves:
    1. Raw attribution map as .npy
    2. Normalised attribution map as .npy
    3. Overlay image as .png
    """
    raw_map = np.asarray(raw_map, dtype=np.float32)

    if raw_map.shape != (224, 224):
        raw_map = resize(
            raw_map,
            (224, 224),
            preserve_range=True,
            anti_aliasing=True
        ).astype(np.float32)

    normalised_map = normalise_map(raw_map)

    raw_path = maps_folder / f"{case_id}_{method_name}_raw.npy"
    normalised_path = maps_folder / f"{case_id}_{method_name}_normalised.npy"
    overlay_path = overlays_folder / f"{case_id}_{method_name}_overlay.png"

    np.save(raw_path, raw_map)
    np.save(normalised_path, normalised_map)

    overlay = overlay_heatmap(
        rgb_image=rgb_image,
        heatmap=normalised_map
    )

    plt.figure(figsize=(5, 5))
    plt.imshow(overlay)
    plt.axis("off")
    plt.title(f"{case_id} - {method_name}")
    plt.tight_layout()
    plt.savefig(
        overlay_path,
        dpi=200,
        bbox_inches="tight"
    )
    plt.close()

    return raw_path, normalised_path, overlay_path, normalised_map
```

```python
target_layers = [model.layer4[-1]]

output_features = model.fc.out_features

if output_features == 1:
    # One-logit binary model: class index 0 represents the positive pneumonia logit.
    positive_target_index = 0
elif output_features == 2:
    # Two-logit model: class index 1 represents Lung Opacity / Pneumonia.
    positive_target_index = 1
else:
    raise ValueError(f"Unexpected number of output features: {output_features}")

positive_target = [ClassifierOutputTarget(positive_target_index)]

print("Grad-CAM target layer:")
print(target_layers[0])

print("\nModel output features:", output_features)
print("Positive target index for Grad-CAM:", positive_target_index)
```

```python
gradcam = GradCAM(
    model=model,
    target_layers=target_layers
)

gradcampp = GradCAMPlusPlus(
    model=model,
    target_layers=target_layers
)

print("Grad-CAM initialised successfully.")
print("Grad-CAM++ initialised successfully.")
```

```python
from lime import lime_image
from skimage.segmentation import slic

print("LIME imported successfully.")
```

```python
def lime_segmentation_function(image):
    """
    SLIC segmentation for LIME.
    These settings are fixed for the feasibility pilot.
    """
    return slic(
        image,
        n_segments=50,
        compactness=10,
        sigma=1,
        start_label=1
    )


def lime_predict_function(images):
    """
    LIME sends a batch of RGB images as NumPy arrays.
    This function returns class probabilities in the format:
    [Normal probability, Pneumonia/Lung Opacity probability]
    """
    tensors = []

    for image in images:
        image = np.asarray(image).astype(np.uint8)
        pil_image = Image.fromarray(image).convert("RGB")
        tensor = model_transform(pil_image)
        tensors.append(tensor)

    batch_tensor = torch.stack(tensors).to(device)

    with torch.no_grad():
        logits = model(batch_tensor)
        scaled_logits = logits / TEMPERATURE

        if logits.shape[1] == 1:
            pneumonia_probability = torch.sigmoid(scaled_logits)
            normal_probability = 1 - pneumonia_probability
            probabilities = torch.cat(
                [normal_probability, pneumonia_probability],
                dim=1
            )
        elif logits.shape[1] == 2:
            probabilities = torch.softmax(scaled_logits, dim=1)
        else:
            raise ValueError(f"Unexpected model output shape: {logits.shape}")

    return probabilities.detach().cpu().numpy()


lime_explainer = lime_image.LimeImageExplainer(
    random_state=SEED
)

print("LIME explainer initialised.")
```

```python
from skimage.segmentation import mark_boundaries
import matplotlib.pyplot as plt
import numpy as np

def show_lime_top_superpixels(rgb_image, explanation, target_index=1, top_k=10):
    segments = explanation.segments
    local_exp = explanation.local_exp[target_index]

    # Sort by positive contribution
    positive_segments = [
        (segment_id, weight)
        for segment_id, weight in local_exp
        if weight > 0
    ]

    positive_segments = sorted(
        positive_segments,
        key=lambda x: x[1],
        reverse=True
    )[:top_k]

    mask = np.zeros(segments.shape, dtype=bool)

    for segment_id, weight in positive_segments:
        mask[segments == segment_id] = True

    masked_image = rgb_image.copy()
    masked_image[~mask] = masked_image[~mask] * 0.35

    boundary_image = mark_boundaries(
        masked_image / 255.0,
        segments,
        color=(1, 1, 0)
    )

    plt.figure(figsize=(6, 6))
    plt.imshow(boundary_image)
    plt.axis("off")
    plt.title(f"LIME top {top_k} positive superpixels")
    plt.show()

    return mask, positive_segments
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

def show_selected_lime_superpixels_clean(
    rgb_image,
    explanation,
    target_index=1,
    top_k=10,
    dim_unselected=True
):
    segments = explanation.segments
    local_exp = explanation.local_exp[target_index]

    positive_segments = [
        (segment_id, weight)
        for segment_id, weight in local_exp
        if weight > 0
    ]

    positive_segments = sorted(
        positive_segments,
        key=lambda x: x[1],
        reverse=True
    )[:top_k]

    selected_mask = np.zeros(segments.shape, dtype=np.uint8)

    for segment_id, weight in positive_segments:
        selected_mask[segments == segment_id] = 1

    display_image = rgb_image.copy()

    if dim_unselected:
        display_image = (display_image * 0.35).astype(np.uint8)
        display_image[selected_mask == 1] = rgb_image[selected_mask == 1]

    # Draw boundary only around selected superpixels
    contours, _ = cv2.findContours(
        selected_mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    outlined_image = display_image.copy()

    cv2.drawContours(
        outlined_image,
        contours,
        -1,
        color=(255, 255, 0),
        thickness=2
    )

    plt.figure(figsize=(6, 6))
    plt.imshow(outlined_image)
    plt.axis("off")
    plt.title(f"LIME top {top_k} positive superpixels")
    plt.tight_layout()
    plt.show()

    print("Top positive LIME segments:")
    for segment_id, weight in positive_segments:
        print(f"Segment {segment_id}: weight={weight:.6f}")

    return selected_mask, positive_segments
```

```python
class CalibratedProbabilityWrapper(nn.Module):
    def __init__(self, base_model, temperature):
        super().__init__()
        self.base_model = base_model
        self.temperature = float(temperature)

    def forward(self, x):
        logits = self.base_model(x)
        scaled_logits = logits / self.temperature

        if scaled_logits.shape[1] == 1:
            pneumonia_probability = torch.sigmoid(scaled_logits)
            normal_probability = 1 - pneumonia_probability
            probabilities = torch.cat(
                [normal_probability, pneumonia_probability],
                dim=1
            )
        elif scaled_logits.shape[1] == 2:
            probabilities = torch.softmax(scaled_logits, dim=1)
        else:
            raise ValueError(f"Unexpected model output shape: {scaled_logits.shape}")

        return probabilities


shap_model = CalibratedProbabilityWrapper(
    base_model=model,
    temperature=TEMPERATURE
).to(device)

shap_model.eval()

print("SHAP probability wrapper created.")
```

```python
def extract_positive_class_shap_map(shap_values, target_index=1):
    """
    Converts SHAP output into a 2D attribution map of shape (224, 224).
    Handles common SHAP return formats.
    """

    if isinstance(shap_values, list):
        positive_values = shap_values[target_index]
    else:
        positive_values = shap_values

    positive_values = np.asarray(positive_values)

    # Remove batch dimension when present
    positive_values = np.squeeze(positive_values)

    # Possible formats:
    # (3, 224, 224)
    # (224, 224, 3)
    # (3, 224, 224, 2)
    # (224, 224, 3, 2)
    # (2, 3, 224, 224)

    if positive_values.ndim == 4:
        # Class dimension at the end
        if positive_values.shape[-1] == 2:
            positive_values = positive_values[..., target_index]

        # Class dimension at the beginning
        elif positive_values.shape[0] == 2:
            positive_values = positive_values[target_index]

        else:
            raise ValueError(f"Unexpected 4D SHAP shape: {positive_values.shape}")

    if positive_values.ndim == 3:
        # CHW format
        if positive_values.shape[0] == 3:
            shap_map = np.mean(np.abs(positive_values), axis=0)

        # HWC format
        elif positive_values.shape[-1] == 3:
            shap_map = np.mean(np.abs(positive_values), axis=-1)

        else:
            shap_map = np.mean(np.abs(positive_values), axis=0)

    elif positive_values.ndim == 2:
        shap_map = positive_values

    else:
        raise ValueError(f"Unexpected SHAP shape after processing: {positive_values.shape}")

    return shap_map.astype(np.float32)
```
