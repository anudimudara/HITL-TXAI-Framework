```python
# ============================================================
# SECTION 12 — EQAG CRITERION 2: APPLICABILITY
# ============================================================

valid_outcomes = {"TP", "TN", "FP", "FN"}

eqag_df["Outcome_Applicable"] = (
    eqag_df["Outcome"]
    .astype(str)
    .str.upper()
    .isin(valid_outcomes)
)

eqag_df["CTS_Applicable"] = (
    eqag_df["CTS_star"].notna()
)

eqag_df["HTS_Applicable"] = (
    eqag_df["HTS"].notna()
)

eqag_df["EAS_Applicable"] = (
    eqag_df["EAS"].notna()
)

eqag_df["Applicable"] = (
    eqag_df["Outcome_Applicable"] &
    eqag_df["CTS_Applicable"] &
    eqag_df["HTS_Applicable"] &
    eqag_df["EAS_Applicable"]
)

print("Applicability results:")
print(eqag_df["Applicable"].value_counts())
```

```python
# ============================================================
# SECTION 13 — LOCALISATION APPLICABILITY
# ============================================================

def localisation_rule(outcome):

    outcome = str(outcome).upper()

    # Reference-positive cases
    if outcome in ["TP", "FN"]:
        return "Conditional - requires valid bounding box"

    # Reference-negative cases
    elif outcome in ["TN", "FP"]:
        return "Not applicable"

    else:
        return "Unknown"


eqag_df["Localisation_Applicability"] = (
    eqag_df["Outcome"]
    .apply(localisation_rule)
)

print(
    eqag_df["Localisation_Applicability"]
    .value_counts()
)
```

```python
# ============================================================
# SECTION 14 — EQAG CRITERION 3: NORMALISATION
# ============================================================

def unit_interval_valid(series):
    return (
        pd.to_numeric(series, errors="coerce")
        .between(0, 1, inclusive="both")
    )


eqag_df["CTS_Normalised"] = (
    unit_interval_valid(eqag_df["CTS_star"])
)

eqag_df["HTS_Normalised"] = (
    unit_interval_valid(eqag_df["HTS"])
)

eqag_df["EAS_Normalised"] = (
    unit_interval_valid(eqag_df["EAS"])
)

eqag_df["UATAI_Normalised"] = (
    unit_interval_valid(eqag_df["UA_TAI"])
)

eqag_df["Normalised"] = (
    eqag_df["CTS_Normalised"] &
    eqag_df["HTS_Normalised"] &
    eqag_df["EAS_Normalised"] &
    eqag_df["UATAI_Normalised"]
)

print("Normalisation results:")
print(eqag_df["Normalised"].value_counts())
```

```python
# ============================================================
# SECTION 15 — NORMALISATION RANGE SUMMARY
# ============================================================

for col in [
    "CTS_star",
    "HTS",
    "EAS",
    "UA_TAI"
]:

    print(
        f"{col}: "
        f"min={eqag_df[col].min():.6f}, "
        f"max={eqag_df[col].max():.6f}"
    )
```

```python
# ============================================================
# SECTION 16 — UNIT/PERCENTAGE CONSISTENCY
# ============================================================

TOLERANCE_PERCENTAGE = 1e-4

pairs = {
    "CTS": ("CTS_star", "CTS_star_pct"),
    "HTS": ("HTS", "HTS_pct"),
    "EAS": ("EAS", "EAS_pct"),
    "UA_TAI": ("UA_TAI", "UA_TAI_pct")
}

for name, (unit_col, pct_col) in pairs.items():

    expected_pct = (
        eqag_df[unit_col] * 100
    )

    difference = np.abs(
        expected_pct -
        eqag_df[pct_col]
    )

    max_difference = difference.max()

    print(
        f"{name}: maximum unit/percentage difference = "
        f"{max_difference:.8f}"
    )
```

```python
# ============================================================
# SECTION 17 — VERIFY FINAL UA-TAI EQUATION
# ============================================================

KAPPA = 0.5

eqag_df["UA_TAI_Recomputed"] = (
    (
        1 -
        np.abs(
            eqag_df["HTS"] -
            eqag_df["CTS_star"]
        )
    )
    *
    np.exp(
        -KAPPA *
        (1 - eqag_df["EAS"])
    )
)

eqag_df["UA_TAI_Formula_Difference"] = (
    np.abs(
        eqag_df["UA_TAI"] -
        eqag_df["UA_TAI_Recomputed"]
    )
)

FORMULA_TOLERANCE = 1e-6

eqag_df["UA_TAI_Formula_Consistent"] = (
    eqag_df["UA_TAI_Formula_Difference"]
    <= FORMULA_TOLERANCE
)

print("UA-TAI formula consistency:")
print(
    eqag_df[
        "UA_TAI_Formula_Consistent"
    ].value_counts()
)

print(
    "\nMaximum difference:",
    eqag_df[
        "UA_TAI_Formula_Difference"
    ].max()
)
```

```python
# ============================================================
# SECTION 18 — CHECK MAIN UA-TAI AGAINST KAPPA 0.5 RESULT
# ============================================================

eqag_df["UA_TAI_Kappa_05_Saved"] = (
    df["UA_TAI_kappa_0_5_unit"]
)

eqag_df["Kappa_05_Difference"] = (
    np.abs(
        eqag_df["UA_TAI"] -
        eqag_df["UA_TAI_Kappa_05_Saved"]
    )
)

eqag_df["Kappa_05_Consistent"] = (
    eqag_df["Kappa_05_Difference"]
    <= 1e-6
)

print(
    eqag_df["Kappa_05_Consistent"]
    .value_counts()
)

print(
    "Maximum difference:",
    eqag_df["Kappa_05_Difference"].max()
)
```

```python
# ============================================================
# SECTION 19 — VERIFY RAW ALIGNMENT
# ============================================================

eqag_df["Raw_Alignment_Recomputed"] = (
    1 -
    np.abs(
        eqag_df["HTS"] -
        eqag_df["CTS_star"]
    )
)

eqag_df["Raw_Alignment_Difference"] = (
    np.abs(
        eqag_df["Raw_Alignment"] -
        eqag_df["Raw_Alignment_Recomputed"]
    )
)

eqag_df["Raw_Alignment_Consistent"] = (
    eqag_df["Raw_Alignment_Difference"]
    <= 1e-6
)

print(
    eqag_df[
        "Raw_Alignment_Consistent"
    ].value_counts()
)
```

```python
# ============================================================
# SECTION 20 — VERIFY EAS DISCOUNT
# ============================================================

eqag_df["EAS_Discount_Saved"] = (
    df["EAS_discount"]
)

eqag_df["EAS_Discount_Recomputed"] = (
    np.exp(
        -KAPPA *
        (1 - eqag_df["EAS"])
    )
)

eqag_df["EAS_Discount_Difference"] = (
    np.abs(
        eqag_df["EAS_Discount_Saved"] -
        eqag_df["EAS_Discount_Recomputed"]
    )
)

eqag_df["EAS_Discount_Consistent"] = (
    eqag_df["EAS_Discount_Difference"]
    <= 1e-6
)

print(
    eqag_df[
        "EAS_Discount_Consistent"
    ].value_counts()
)
```

```python
# ============================================================
# STEP 3 — REPRODUCIBILITY STATUS
# ============================================================

eqag_df["Reproducible"] = True

eqag_df["Reproducibility_Note"] = (
    "Evidence derived from retained model configuration, "
    "saved XAI outputs, specialist evaluation data, "
    "analysis notebooks and final TrustAlign workbook."
)

print(eqag_df["Reproducible"].value_counts())
```

```python
# ============================================================
# STEP 4 — STATISTICAL ADEQUACY
# ============================================================

eqag_df["Statistical_Adequacy"] = (
    "Qualified with limitations"
)

eqag_df["Statistical_Limitation"] = (
    "TrustAlign analysis is based on 32 case-method records "
    "from 8 specialist-evaluated cases and 13 eligible specialists."
)

print(eqag_df["Statistical_Adequacy"].value_counts())
```

```python
# ============================================================
# STEP 5 — CORE EQAG ELIGIBILITY
# ============================================================

eqag_df["Core_EQAG_Eligible"] = (
    eqag_df["Complete"] &
    eqag_df["Applicable"] &
    eqag_df["Normalised"] &
    eqag_df["Reproducible"] &
    eqag_df["Provenance_Complete"] &
    eqag_df["UA_TAI_Formula_Consistent"] &
    eqag_df["Governance_Output_Complete"]
)

print(eqag_df["Core_EQAG_Eligible"].value_counts())
```

```python
# ============================================================
# STEP 6 — FINAL EQAG STATUS
# ============================================================

def assign_eqag_status(row):

    if not row["Core_EQAG_Eligible"]:
        return "Not qualified"

    if row["Statistical_Adequacy"] == "Qualified with limitations":
        return "Qualified with limitations"

    return "Qualified"


eqag_df["EQAG_Status"] = eqag_df.apply(
    assign_eqag_status,
    axis=1
)

print("Final EQAG status:")
print(eqag_df["EQAG_Status"].value_counts())
```

```python
# ============================================================
# STEP 7 — UA-TAI ELIGIBILITY
# ============================================================

eqag_df["UA_TAI_Eligible"] = (
    eqag_df["EQAG_Status"]
    .isin([
        "Qualified",
        "Qualified with limitations"
    ])
)

print(eqag_df["UA_TAI_Eligible"].value_counts())
```

```python
# ============================================================
# STEP 8 — EQAG REASON CODES
# ============================================================

def generate_reason_codes(row):

    codes = []

    if not row["Complete"]:
        codes.append("N01_MISSING_CORE_EVIDENCE")

    if not row["Applicable"]:
        codes.append("N02_NOT_APPLICABLE")

    if not row["Normalised"]:
        codes.append("N03_INVALID_SCALE")

    if not row["Provenance_Complete"]:
        codes.append("N04_PROVENANCE_INCOMPLETE")

    if not row["Reproducible"]:
        codes.append("N05_REPRODUCIBILITY_FAILURE")

    if not row["UA_TAI_Formula_Consistent"]:
        codes.append("N06_FORMULA_INCONSISTENCY")

    if row["Statistical_Adequacy"] == "Qualified with limitations":
        codes.append("Q01_SMALL_ALIGNMENT_SUBSET")

    if row["Localisation_Applicability"] == \
       "Conditional - requires valid bounding box":
        codes.append("Q02_LOCALISATION_CONDITIONAL")

    if len(codes) == 0:
        codes.append("Q00_FULLY_QUALIFIED")

    return "; ".join(codes)


eqag_df["EQAG_Reason_Codes"] = (
    eqag_df.apply(
        generate_reason_codes,
        axis=1
    )
)

display(
    eqag_df[
        [
            "Case_ID",
            "Method",
            "Outcome",
            "EQAG_Status",
            "EQAG_Reason_Codes",
            "UA_TAI_Eligible"
        ]
    ].head(10)
)
```
