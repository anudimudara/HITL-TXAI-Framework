```python
def calculate_ece(
    y_true,
    y_prob,
    n_bins=10
):
    y_true = np.asarray(y_true, dtype=int)
    y_prob = np.asarray(y_prob, dtype=float)

    bin_edges = np.linspace(
        0.0,
        1.0,
        n_bins + 1
    )

    bin_numbers = np.digitize(
        y_prob,
        bin_edges[1:-1],
        right=True
    )

    total_ece = 0.0
    rows = []

    for bin_index in range(n_bins):

        mask = bin_numbers == bin_index
        count = int(mask.sum())

        if count == 0:
            continue

        mean_probability = float(
            y_prob[mask].mean()
        )

        observed_frequency = float(
            y_true[mask].mean()
        )

        weight = count / len(y_true)

        absolute_error = abs(
            observed_frequency - mean_probability
        )

        total_ece += weight * absolute_error

        rows.append({
            "bin": bin_index + 1,
            "lower_boundary": bin_edges[bin_index],
            "upper_boundary": bin_edges[bin_index + 1],
            "number_of_samples": count,
            "mean_predicted_probability": mean_probability,
            "observed_pneumonia_frequency": observed_frequency,
            "absolute_error": absolute_error,
            "weighted_error": weight * absolute_error
        })

    return float(total_ece), pd.DataFrame(rows)


print("ECE function created.")
```

```python
def calculate_binary_metrics(
    y_true,
    y_prob,
    threshold
):
    y_true = np.asarray(y_true, dtype=int)
    y_prob = np.asarray(y_prob, dtype=float)

    y_pred = (
        y_prob >= threshold
    ).astype(int)

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    tn, fp, fn, tp = cm.ravel()

    sensitivity = (
        tp / (tp + fn)
        if (tp + fn) > 0
        else 0.0
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else 0.0
    )

    npv = (
        tn / (tn + fn)
        if (tn + fn) > 0
        else 0.0
    )

    fpr = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else 0.0
    )

    fnr = (
        fn / (fn + tp)
        if (fn + tp) > 0
        else 0.0
    )

    ece, _ = calculate_ece(
        y_true,
        y_prob,
        n_bins=10
    )

    clipped_probabilities = np.clip(
        y_prob,
        1e-7,
        1 - 1e-7
    )

    metrics = {
        "threshold": float(threshold),

        "accuracy": float(
            accuracy_score(
                y_true,
                y_pred
            )
        ),

        "balanced_accuracy": float(
            balanced_accuracy_score(
                y_true,
                y_pred
            )
        ),

        "precision_pneumonia": float(
            precision_score(
                y_true,
                y_pred,
                zero_division=0
            )
        ),

        "sensitivity_pneumonia": float(
            sensitivity
        ),

        "specificity_normal": float(
            specificity
        ),

        "f1_pneumonia": float(
            f1_score(
                y_true,
                y_pred,
                zero_division=0
            )
        ),

        "negative_predictive_value": float(
            npv
        ),

        "false_positive_rate": float(
            fpr
        ),

        "false_negative_rate": float(
            fnr
        ),

        "mcc": float(
            matthews_corrcoef(
                y_true,
                y_pred
            )
        ),

        "roc_auc": float(
            roc_auc_score(
                y_true,
                y_prob
            )
        ),

        "average_precision": float(
            average_precision_score(
                y_true,
                y_prob
            )
        ),

        "brier_score": float(
            brier_score_loss(
                y_true,
                y_prob
            )
        ),

        "log_loss": float(
            log_loss(
                y_true,
                np.column_stack([
                    1 - clipped_probabilities,
                    clipped_probabilities
                ]),
                labels=[0, 1]
            )
        ),

        "ece_10_bins": float(ece),

        "true_negatives": int(tn),
        "false_positives": int(fp),
        "false_negatives": int(fn),
        "true_positives": int(tp)
    }

    return metrics, y_pred


print("Metric function created.")
```

```python
def collect_test_logits(
    model,
    data_loader,
    device,
    description
):
    model.eval()

    logits_list = []
    labels_list = []

    start_time = time.time()

    with torch.no_grad():

        for images, labels in tqdm(
            data_loader,
            desc=description
        ):

            images = images.to(device)

            logits = model(images)

            logits_list.append(
                logits.cpu()
            )

            labels_list.append(
                labels.cpu()
            )

    elapsed_time = time.time() - start_time

    all_logits = torch.cat(
        logits_list,
        dim=0
    )

    all_labels = torch.cat(
        labels_list,
        dim=0
    )

    assert all_logits.shape == (
        len(test_dataset),
        2
    )

    assert all_labels.shape[0] == len(
        test_dataset
    )

    return (
        all_logits,
        all_labels,
        elapsed_time
    )


print("Inference function created.")
```

```python
def logits_to_probabilities(
    logits,
    temperature=1.0
):
    scaled_logits = (
        logits.detach().float().cpu()
        / float(temperature)
    )

    probabilities = torch.softmax(
        scaled_logits,
        dim=1
    )

    return probabilities[:, 1].numpy()


print("Probability conversion function created.")
```

```python
def calculate_probability_metrics(
    y_true,
    probability
):
    probability = np.asarray(
        probability,
        dtype=float
    )

    clipped = np.clip(
        probability,
        1e-7,
        1 - 1e-7
    )

    ece, bins = calculate_ece(
        y_true,
        probability,
        n_bins=10
    )

    return {
        "brier_score": brier_score_loss(
            y_true,
            probability
        ),

        "log_loss": log_loss(
            y_true,
            np.column_stack([
                1 - clipped,
                clipped
            ]),
            labels=[0, 1]
        ),

        "ece_10_bins": ece,

        "roc_auc": roc_auc_score(
            y_true,
            probability
        ),

        "average_precision": (
            average_precision_score(
                y_true,
                probability
            )
        )
    }, bins
```

```python
def bootstrap_confidence_intervals(
    y_true,
    y_prob,
    threshold,
    n_iterations=1000,
    seed=42
):
    y_true = np.asarray(
        y_true,
        dtype=int
    )

    y_prob = np.asarray(
        y_prob,
        dtype=float
    )

    rng = np.random.default_rng(seed)

    normal_indices = np.where(
        y_true == 0
    )[0]

    pneumonia_indices = np.where(
        y_true == 1
    )[0]

    metric_names = [
        "accuracy",
        "balanced_accuracy",
        "precision_pneumonia",
        "sensitivity_pneumonia",
        "specificity_normal",
        "f1_pneumonia",
        "mcc",
        "roc_auc",
        "average_precision"
    ]

    distributions = {
        name: []
        for name in metric_names
    }

    for _ in tqdm(
        range(n_iterations),
        desc="Bootstrap confidence intervals"
    ):
        sampled_normal = rng.choice(
            normal_indices,
            size=len(normal_indices),
            replace=True
        )

        sampled_pneumonia = rng.choice(
            pneumonia_indices,
            size=len(pneumonia_indices),
            replace=True
        )

        sampled_indices = np.concatenate([
            sampled_normal,
            sampled_pneumonia
        ])

        rng.shuffle(sampled_indices)

        sample_metrics, _ = (
            calculate_binary_metrics(
                y_true[sampled_indices],
                y_prob[sampled_indices],
                threshold
            )
        )

        for name in metric_names:
            distributions[name].append(
                sample_metrics[name]
            )

    original_metrics, _ = (
        calculate_binary_metrics(
            y_true,
            y_prob,
            threshold
        )
    )

    rows = []

    for name in metric_names:

        values = np.asarray(
            distributions[name],
            dtype=float
        )

        rows.append({
            "metric": name,
            "point_estimate": (
                original_metrics[name]
            ),
            "ci_lower_95": float(
                np.percentile(values, 2.5)
            ),
            "ci_upper_95": float(
                np.percentile(values, 97.5)
            ),
            "bootstrap_mean": float(
                values.mean()
            ),
            "bootstrap_iterations": (
                n_iterations
            )
        })

    return pd.DataFrame(rows)
```

```python
from scipy.stats import binomtest

dense_correct = (
    densenet_pred_primary == y_test
)

resnet_correct = (
    resnet_pred_primary == y_test
)

both_correct = int(
    np.sum(dense_correct & resnet_correct)
)

both_wrong = int(
    np.sum(~dense_correct & ~resnet_correct)
)

dense_only_correct = int(
    np.sum(dense_correct & ~resnet_correct)
)

resnet_only_correct = int(
    np.sum(~dense_correct & resnet_correct)
)

discordant_total = (
    dense_only_correct
    + resnet_only_correct
)

if discordant_total > 0:
    mcnemar_p_value = binomtest(
        k=dense_only_correct,
        n=discordant_total,
        p=0.5,
        alternative="two-sided"
    ).pvalue
else:
    mcnemar_p_value = 1.0

mcnemar_results = pd.DataFrame([{
    "both_correct": both_correct,
    "both_wrong": both_wrong,
    "densenet_only_correct": dense_only_correct,
    "resnet_only_correct": resnet_only_correct,
    "discordant_total": discordant_total,
    "exact_mcnemar_p_value": mcnemar_p_value
}])

display(mcnemar_results)

mcnemar_results.to_csv(
    os.path.join(
        FINAL_RESULTS_DIR,
        "mcnemar_test_results.csv"
    ),
    index=False
)
```

```python
def paired_bootstrap_auc_difference(
    y_true,
    densenet_prob,
    resnet_prob,
    n_iterations=1000,
    seed=42
):
    y_true = np.asarray(
        y_true,
        dtype=int
    )

    densenet_prob = np.asarray(
        densenet_prob,
        dtype=float
    )

    resnet_prob = np.asarray(
        resnet_prob,
        dtype=float
    )

    rng = np.random.default_rng(seed)

    normal_indices = np.where(
        y_true == 0
    )[0]

    pneumonia_indices = np.where(
        y_true == 1
    )[0]

    differences = []

    for _ in tqdm(
        range(n_iterations),
        desc="Paired AUC bootstrap"
    ):
        sampled_normal = rng.choice(
            normal_indices,
            len(normal_indices),
            replace=True
        )

        sampled_pneumonia = rng.choice(
            pneumonia_indices,
            len(pneumonia_indices),
            replace=True
        )

        indices = np.concatenate([
            sampled_normal,
            sampled_pneumonia
        ])

        dense_auc = roc_auc_score(
            y_true[indices],
            densenet_prob[indices]
        )

        resnet_auc = roc_auc_score(
            y_true[indices],
            resnet_prob[indices]
        )

        differences.append(
            resnet_auc - dense_auc
        )

    differences = np.asarray(
        differences
    )

    dense_point_auc = roc_auc_score(
        y_true,
        densenet_prob
    )

    resnet_point_auc = roc_auc_score(
        y_true,
        resnet_prob
    )

    point_difference = (
        resnet_point_auc
        - dense_point_auc
    )

    lower = float(
        np.percentile(
            differences,
            2.5
        )
    )

    upper = float(
        np.percentile(
            differences,
            97.5
        )
    )

    probability_nonpositive = (
        np.sum(differences <= 0) + 1
    ) / (
        n_iterations + 1
    )

    probability_nonnegative = (
        np.sum(differences >= 0) + 1
    ) / (
        n_iterations + 1
    )

    two_sided_p = min(
        1.0,
        2 * min(
            probability_nonpositive,
            probability_nonnegative
        )
    )

    return {
        "densenet_auc": dense_point_auc,
        "resnet_auc": resnet_point_auc,
        "resnet_minus_densenet_auc": (
            point_difference
        ),
        "ci_lower_95": lower,
        "ci_upper_95": upper,
        "paired_bootstrap_p_value": (
            two_sided_p
        ),
        "bootstrap_iterations": (
            n_iterations
        )
    }
```

```python
def confidence_band_analysis(
    y_true,
    calibrated_probability,
    model_name
):
    y_true = np.asarray(
        y_true,
        dtype=int
    )

    calibrated_probability = np.asarray(
        calibrated_probability,
        dtype=float
    )

    prediction = (
        calibrated_probability >= 0.50
    ).astype(int)

    confidence = np.maximum(
        calibrated_probability,
        1 - calibrated_probability
    )

    bands = [
        (
            "Uncertain",
            0.50,
            0.60
        ),
        (
            "Moderate",
            0.60,
            0.80
        ),
        (
            "High",
            0.80,
            1.000001
        )
    ]

    rows = []

    for band_name, lower, upper in bands:

        mask = (
            (confidence >= lower)
            & (confidence < upper)
        )

        count = int(mask.sum())

        if count == 0:
            continue

        y_band = y_true[mask]
        pred_band = prediction[mask]

        cm = confusion_matrix(
            y_band,
            pred_band,
            labels=[0, 1]
        )

        tn, fp, fn, tp = cm.ravel()

        sensitivity = (
            tp / (tp + fn)
            if (tp + fn) > 0
            else np.nan
        )

        specificity = (
            tn / (tn + fp)
            if (tn + fp) > 0
            else np.nan
        )

        accuracy = accuracy_score(
            y_band,
            pred_band
        )

        rows.append({
            "model": model_name,
            "confidence_band": band_name,
            "lower_boundary": lower,
            "upper_boundary": upper,
            "cases": count,
            "coverage": count / len(y_true),
            "accuracy": accuracy,
            "error_rate": 1 - accuracy,
            "sensitivity": sensitivity,
            "specificity": specificity,
            "false_positives": int(fp),
            "false_negatives": int(fn),
            "mean_confidence": float(
                confidence[mask].mean()
            )
        })

    return pd.DataFrame(rows)
```
