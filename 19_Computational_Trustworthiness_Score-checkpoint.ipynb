```python
def find_file(base_dir, filename, required=True):
    """
    Finds the newest matching file under BASE_DIR.
    If required=True and the file is missing, it raises an error.
    """
    matches = list(base_dir.rglob(filename))
    
    if len(matches) == 0:
        if required:
            raise FileNotFoundError(f"Could not find {filename} under {base_dir}")
        return None
    
    matches = sorted(matches, key=lambda p: p.stat().st_mtime, reverse=True)
    return matches[0]


def safe_numeric(series):
    """
    Converts values to numeric format.
    Non-numeric values become NaN.
    """
    return pd.to_numeric(series, errors="coerce")


def method_clean(x):
    """
    Standardises XAI method names.
    """
    x = str(x).lower().replace("-", "").replace("_", "").replace(" ", "")
    
    if x in ["gradcam"]:
        return "gradcam"
    if x in ["gradcampp", "gradcamplusplus"]:
        return "gradcampp"
    if x == "lime":
        return "lime"
    if x == "shap":
        return "shap"
    
    return x


def method_label(x):
    """
    Converts cleaned method names into dissertation-friendly labels.
    """
    labels = {
        "gradcam": "Grad-CAM",
        "gradcampp": "Grad-CAM++",
        "lime": "LIME",
        "shap": "SHAP"
    }
    return labels.get(str(x).lower(), str(x))


def clean_outcome(x):
    """
    Cleans outcome labels such as TP, TN, FP, FN.
    """
    return str(x).upper().strip()


def clip_score(series):
    """
    Ensures score values are between 0 and 1.
    """
    return safe_numeric(series).clip(lower=0, upper=1)


def cts_level(score):
    """
    Converts numerical CTS into interpretive trustworthiness category.
    """
    if pd.isna(score):
        return "Not available"
    elif score < 0.40:
        return "Low computational trustworthiness"
    elif score < 0.70:
        return "Moderate computational trustworthiness"
    else:
        return "High computational trustworthiness"


def weighted_score(row, components, weights):
    """
    Calculates weighted CTS.
    
    components = list of column names
    weights = dictionary of weights for each component
    
    Missing values are skipped and remaining weights are renormalised.
    """
    values = []
    used_weights = []
    
    for comp in components:
        value = row.get(comp, np.nan)
        weight = weights.get(comp, 0)
        
        if pd.notna(value):
            values.append(float(value))
            used_weights.append(float(weight))
    
    if len(values) == 0 or sum(used_weights) == 0:
        return np.nan
    
    used_weights = np.array(used_weights)
    used_weights = used_weights / used_weights.sum()
    
    return float(np.dot(values, used_weights))


def summary_stats(group, score_col):
    """
    Creates summary statistics for CTS.
    """
    return pd.Series({
        "count": group[score_col].count(),
        "mean": group[score_col].mean(),
        "std": group[score_col].std(),
        "median": group[score_col].median(),
        "min": group[score_col].min(),
        "max": group[score_col].max()
    })
```

```python
cts_df = case_method.copy()

# Clean method and outcome
cts_df["method"] = cts_df["method"].apply(method_clean)
cts_df["method_label"] = cts_df["method"].apply(method_label)
cts_df["outcome_upper"] = cts_df["outcome"].apply(clean_outcome)

# Ensure component scores are numeric and between 0 and 1
cts_df["confidence_score"] = clip_score(cts_df["confidence_score"])
cts_df["prediction_robustness_score"] = clip_score(cts_df["prediction_robustness_score"])
cts_df["faithfulness_score"] = clip_score(cts_df["faithfulness_score"])
cts_df["explanation_stability_score"] = clip_score(cts_df["explanation_stability_score"])

# Localisation may be missing for cases without bounding-box reference evidence
if "localisation_score" in cts_df.columns:
    cts_df["localisation_score"] = clip_score(cts_df["localisation_score"])
else:
    cts_df["localisation_score"] = np.nan

# Correct/error labels for analysis only
cts_df["is_correct"] = cts_df["outcome_upper"].isin(["TP", "TN"]).astype(int)
cts_df["is_error"] = cts_df["outcome_upper"].isin(["FP", "FN"]).astype(int)

print("Prepared CTS case-method table.")
print("Rows:", len(cts_df))
print("Expected rows: 480")
print("Unique cases:", cts_df["xai_case_id"].nunique())
print("Expected unique cases: 120")
print("Methods:")
print(cts_df["method_label"].value_counts())

display(cts_df.head())
```

```python
# Main CTS components available for all case-method rows
main_components = [
    "confidence_score",
    "prediction_robustness_score",
    "faithfulness_score",
    "explanation_stability_score"
]

# Extended CTS includes localisation where available
extended_components = [
    "confidence_score",
    "prediction_robustness_score",
    "faithfulness_score",
    "explanation_stability_score",
    "localisation_score"
]

# Equal weights for main CTS
main_equal_weights = {
    "confidence_score": 0.25,
    "prediction_robustness_score": 0.25,
    "faithfulness_score": 0.25,
    "explanation_stability_score": 0.25
}

# Equal weights for extended CTS
extended_equal_weights = {
    "confidence_score": 0.20,
    "prediction_robustness_score": 0.20,
    "faithfulness_score": 0.20,
    "explanation_stability_score": 0.20,
    "localisation_score": 0.20
}

print("Main CTS components:")
print(main_components)

print("\nExtended CTS components:")
print(extended_components)
```

```python
sensitivity_scenarios = {
    "Equal weights": {
        "confidence_score": 0.25,
        "prediction_robustness_score": 0.25,
        "faithfulness_score": 0.25,
        "explanation_stability_score": 0.25
    },
    "Confidence-focused": {
        "confidence_score": 0.40,
        "prediction_robustness_score": 0.20,
        "faithfulness_score": 0.20,
        "explanation_stability_score": 0.20
    },
    "Robustness-focused": {
        "confidence_score": 0.20,
        "prediction_robustness_score": 0.40,
        "faithfulness_score": 0.20,
        "explanation_stability_score": 0.20
    },
    "XAI-quality-focused": {
        "confidence_score": 0.15,
        "prediction_robustness_score": 0.15,
        "faithfulness_score": 0.35,
        "explanation_stability_score": 0.35
    },
    "Faithfulness-focused": {
        "confidence_score": 0.20,
        "prediction_robustness_score": 0.20,
        "faithfulness_score": 0.40,
        "explanation_stability_score": 0.20
    },
    "Stability-focused": {
        "confidence_score": 0.20,
        "prediction_robustness_score": 0.20,
        "faithfulness_score": 0.20,
        "explanation_stability_score": 0.40
    },
    "Safety-focused": {
        "confidence_score": 0.15,
        "prediction_robustness_score": 0.35,
        "faithfulness_score": 0.35,
        "explanation_stability_score": 0.15
    }
}

sensitivity_case_method_rows = []
sensitivity_method_rows = []

for scenario_name, weights in sensitivity_scenarios.items():
    temp = cts_df.copy()
    
    temp["sensitivity_scenario"] = scenario_name
    temp["sensitivity_cts"] = temp.apply(
        lambda row: weighted_score(row, main_components, weights),
        axis=1
    )
    
    # Save case-method sensitivity rows
    sensitivity_case_method_rows.append(
        temp[
            [
                "xai_case_id",
                "method",
                "method_label",
                "outcome_upper",
                "sensitivity_scenario",
                "sensitivity_cts"
            ]
        ]
    )
    
    # Method-level sensitivity summary
    method_summary = temp.groupby(["sensitivity_scenario", "method", "method_label"]).agg(
        mean_cts=("sensitivity_cts", "mean"),
        std_cts=("sensitivity_cts", "std"),
        median_cts=("sensitivity_cts", "median"),
        min_cts=("sensitivity_cts", "min"),
        max_cts=("sensitivity_cts", "max"),
        records=("xai_case_id", "count")
    ).reset_index()
    
    method_summary["rank"] = method_summary.groupby("sensitivity_scenario")["mean_cts"].rank(
        method="min",
        ascending=False
    ).astype(int)
    
    sensitivity_method_rows.append(method_summary)

cts_sensitivity_case_method = pd.concat(sensitivity_case_method_rows, ignore_index=True)
cts_sensitivity_method_summary = pd.concat(sensitivity_method_rows, ignore_index=True)

# Overall sensitivity analysis by method
cts_sensitivity_analysis = cts_sensitivity_method_summary.groupby(["method", "method_label"]).agg(
    mean_of_scenario_means=("mean_cts", "mean"),
    std_across_scenarios=("mean_cts", "std"),
    min_scenario_mean=("mean_cts", "min"),
    max_scenario_mean=("mean_cts", "max"),
    best_rank=("rank", "min"),
    worst_rank=("rank", "max"),
    average_rank=("rank", "mean")
).reset_index()

cts_sensitivity_analysis = cts_sensitivity_analysis.sort_values("average_rank")

# Save files
sensitivity_case_method_path = OUTPUT_DIR / "cts_sensitivity_case_method_scores.csv"
sensitivity_method_summary_path = OUTPUT_DIR / "cts_sensitivity_method_summary.csv"
sensitivity_analysis_path = OUTPUT_DIR / "cts_sensitivity_analysis.csv"

cts_sensitivity_case_method.to_csv(sensitivity_case_method_path, index=False)
cts_sensitivity_method_summary.to_csv(sensitivity_method_summary_path, index=False)
cts_sensitivity_analysis.to_csv(sensitivity_analysis_path, index=False)

print("CTS sensitivity analysis saved:")
print(sensitivity_case_method_path)
print(sensitivity_method_summary_path)
print(sensitivity_analysis_path)

print("\nSensitivity method summary:")
display(cts_sensitivity_method_summary)

print("\nSensitivity analysis:")
display(cts_sensitivity_analysis)
```

```python
def add_percentage_column(df, score_col, percentage_col=None):
    """
    Adds a percentage column by multiplying a 0–1 score by 100.
    """
    if score_col not in df.columns:
        return df
    
    if percentage_col is None:
        percentage_col = score_col + "_percentage"
    
    df[percentage_col] = (pd.to_numeric(df[score_col], errors="coerce") * 100).round(2)
    return df
```
