```python
import torch
from torch.utils.data import DataLoader
from torchvision import datasets, transforms

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

validation_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

val_dataset = datasets.ImageFolder(
    root=VAL_DIR,
    transform=validation_transform
)

val_loader = DataLoader(
    val_dataset,
    batch_size=16,
    shuffle=False,
    num_workers=0
)

print("Device:", device)
print("Validation images:", len(val_dataset))
print("Validation batches:", len(val_loader))
print("Class mapping:", val_dataset.class_to_idx)
```

```python
import torch.nn as nn
from torchvision.models import resnet50

model = resnet50(weights=None)

number_of_features = model.fc.in_features

model.fc = nn.Linear(
    number_of_features,
    2
)

model = model.to(device)

try:
    state_dict = torch.load(
        CHECKPOINT_PATH,
        map_location=device,
        weights_only=True
    )
except TypeError:
    state_dict = torch.load(
        CHECKPOINT_PATH,
        map_location=device
    )

model.load_state_dict(state_dict)
model.eval()

print("Best ResNet50 checkpoint loaded.")
print(model.fc)
```

```python
import time
import numpy as np
from tqdm import tqdm

all_true_labels = []
all_pneumonia_probabilities = []

start_time = time.time()

with torch.no_grad():

    for images, labels in tqdm(
        val_loader,
        desc="Evaluating ResNet50"
    ):

        images = images.to(device)

        outputs = model(images)

        probabilities = torch.softmax(
            outputs,
            dim=1
        )

        pneumonia_probabilities = probabilities[:, 1]

        all_true_labels.extend(
            labels.numpy()
        )

        all_pneumonia_probabilities.extend(
            pneumonia_probabilities.cpu().numpy()
        )

elapsed_time = time.time() - start_time

y_true = np.asarray(
    all_true_labels,
    dtype=int
)

y_prob = np.asarray(
    all_pneumonia_probabilities,
    dtype=float
)

print("Predictions generated:", len(y_true))
print(f"Total inference time: {elapsed_time:.2f} seconds")
print(
    f"Average time per image: "
    f"{elapsed_time / len(y_true):.4f} seconds"
)
```

```python
def calculate_ece(
    y_true,
    y_prob,
    n_bins=10
):
    """
    Calculate binary Expected Calibration Error for
    predicted pneumonia probabilities.
    """

    y_true = np.asarray(y_true, dtype=int)
    y_prob = np.asarray(y_prob, dtype=float)

    bin_edges = np.linspace(
        0.0,
        1.0,
        n_bins + 1
    )

    bin_numbers = np.digitize(
        y_prob,
        bin_edges[1:-1],
        right=True
    )

    total_ece = 0.0
    calibration_rows = []

    for bin_index in range(n_bins):

        bin_mask = bin_numbers == bin_index
        bin_count = int(bin_mask.sum())

        if bin_count == 0:
            continue

        mean_probability = float(
            y_prob[bin_mask].mean()
        )

        observed_frequency = float(
            y_true[bin_mask].mean()
        )

        bin_weight = (
            bin_count / len(y_true)
        )

        absolute_error = abs(
            observed_frequency
            - mean_probability
        )

        total_ece += (
            bin_weight * absolute_error
        )

        calibration_rows.append({
            "bin": bin_index + 1,
            "lower_boundary": bin_edges[bin_index],
            "upper_boundary": bin_edges[bin_index + 1],
            "number_of_samples": bin_count,
            "mean_predicted_probability": mean_probability,
            "observed_pneumonia_frequency": observed_frequency,
            "absolute_calibration_error": absolute_error,
            "weighted_calibration_error": (
                bin_weight * absolute_error
            )
        })

    calibration_table = pd.DataFrame(
        calibration_rows
    )

    return float(total_ece), calibration_table


print("ECE function created.")
```

```python
def calculate_binary_metrics(
    y_true,
    y_prob,
    threshold=0.50
):
    """
    Calculate classification, discrimination and
    calibration metrics for binary pneumonia detection.
    """

    y_true = np.asarray(y_true, dtype=int)
    y_prob = np.asarray(y_prob, dtype=float)

    y_pred = (
        y_prob >= threshold
    ).astype(int)

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    tn, fp, fn, tp = cm.ravel()

    sensitivity = (
        tp / (tp + fn)
        if (tp + fn) > 0
        else 0.0
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else 0.0
    )

    negative_predictive_value = (
        tn / (tn + fn)
        if (tn + fn) > 0
        else 0.0
    )

    false_positive_rate = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else 0.0
    )

    false_negative_rate = (
        fn / (fn + tp)
        if (fn + tp) > 0
        else 0.0
    )

    ece, _ = calculate_ece(
        y_true,
        y_prob,
        n_bins=10
    )

    precision_values, recall_values, _ = (
        precision_recall_curve(
            y_true,
            y_prob
        )
    )

    trapezoidal_pr_auc = auc(
        recall_values,
        precision_values
    )

    clipped_probabilities = np.clip(
        y_prob,
        1e-7,
        1 - 1e-7
    )

    metrics = {
        "threshold": float(threshold),

        "accuracy": float(
            accuracy_score(
                y_true,
                y_pred
            )
        ),

        "balanced_accuracy": float(
            balanced_accuracy_score(
                y_true,
                y_pred
            )
        ),

        "precision_pneumonia": float(
            precision_score(
                y_true,
                y_pred,
                zero_division=0
            )
        ),

        "sensitivity_pneumonia": float(
            sensitivity
        ),

        "specificity_normal": float(
            specificity
        ),

        "f1_pneumonia": float(
            f1_score(
                y_true,
                y_pred,
                zero_division=0
            )
        ),

        "negative_predictive_value": float(
            negative_predictive_value
        ),

        "false_positive_rate": float(
            false_positive_rate
        ),

        "false_negative_rate": float(
            false_negative_rate
        ),

        "matthews_correlation_coefficient": float(
            matthews_corrcoef(
                y_true,
                y_pred
            )
        ),

        "roc_auc": float(
            roc_auc_score(
                y_true,
                y_prob
            )
        ),

        "average_precision_auprc": float(
            average_precision_score(
                y_true,
                y_prob
            )
        ),

        "trapezoidal_pr_auc": float(
            trapezoidal_pr_auc
        ),

        "brier_score": float(
            brier_score_loss(
                y_true,
                y_prob
            )
        ),

        "log_loss": float(
            log_loss(
                y_true,
                np.column_stack([
                    1 - clipped_probabilities,
                    clipped_probabilities
                ]),
                labels=[0, 1]
            )
        ),

        "expected_calibration_error": float(
            ece
        ),

        "true_negatives": int(tn),
        "false_positives": int(fp),
        "false_negatives": int(fn),
        "true_positives": int(tp)
    }

    return metrics, y_pred


print("Binary evaluation function created.")
```

```python
threshold_rows = []

threshold_values = np.arange(
    0.05,
    0.951,
    0.01
)

for threshold in threshold_values:

    threshold_metrics, _ = (
        calculate_binary_metrics(
            y_true,
            y_prob,
            threshold=float(threshold)
        )
    )

    threshold_metrics["youden_j"] = (
        threshold_metrics[
            "sensitivity_pneumonia"
        ]
        + threshold_metrics[
            "specificity_normal"
        ]
        - 1
    )

    threshold_rows.append(
        threshold_metrics
    )

threshold_df = pd.DataFrame(
    threshold_rows
)

print(
    "Number of thresholds evaluated:",
    len(threshold_df)
)

display(
    threshold_df[
        [
            "threshold",
            "accuracy",
            "precision_pneumonia",
            "sensitivity_pneumonia",
            "specificity_normal",
            "f1_pneumonia",
            "balanced_accuracy",
            "youden_j"
        ]
    ].head(10)
)
```
