```python
import pandas as pd

threshold_path = (
    project_folder
    / "resnet50_validation_results"
    / "resnet50_metrics_selected_threshold.csv"
)

calibration_path = (
    project_folder
    / "temperature_scaling_results"
    / "resnet50_validation_calibration.csv"
)

threshold_results = pd.read_csv(threshold_path)
calibration_results = pd.read_csv(calibration_path)

print("Threshold file:")
display(threshold_results)

print("\nCalibration file:")
display(calibration_results)

print("\nThreshold columns:")
print(threshold_results.columns.tolist())

print("\nCalibration columns:")
print(calibration_results.columns.tolist())
```

```python
import numpy as np

epsilon = 1e-8

probability_after = calibration_results[
    "pneumonia_probability_after"
].clip(epsilon, 1 - epsilon)

logit_difference = (
    calibration_results["pneumonia_logit"]
    - calibration_results["normal_logit"]
)

estimated_temperatures = (
    logit_difference
    / np.log(
        probability_after
        / (1 - probability_after)
    )
)

# Remove unstable values close to probability 0.5
stable_temperatures = estimated_temperatures[
    np.isfinite(estimated_temperatures)
    & (np.abs(logit_difference) > 1e-6)
]

TEMPERATURE = float(
    stable_temperatures.median()
)

LOCKED_THRESHOLD = float(
    threshold_results.loc[0, "threshold"]
)

print("Locked threshold:", LOCKED_THRESHOLD)
print("Estimated temperature:", TEMPERATURE)
print(
    "Temperature variation:",
    stable_temperatures.std()
)
```

```python
scaled_probability = 1 / (
    1 + np.exp(
        -logit_difference / TEMPERATURE
    )
)

maximum_error = np.max(
    np.abs(
        scaled_probability
        - calibration_results[
            "pneumonia_probability_after"
        ]
    )
)

print(
    "Maximum reconstruction error:",
    maximum_error
)
```

```python
from pathlib import Path
import torch

project_folder = Path(
    r"C:\Users\anudi\Downloads\MSc_Research"
)

best_path = project_folder / "best_resnet50_rsna.pth"
final_path = project_folder / "final_resnet50_rsna.pth"


def load_checkpoint(path):
    try:
        return torch.load(
            path,
            map_location="cpu",
            weights_only=True
        )
    except TypeError:
        return torch.load(
            path,
            map_location="cpu"
        )


def extract_state_dict(checkpoint):
    if not isinstance(checkpoint, dict):
        raise TypeError("Checkpoint is not dictionary-based.")

    for key in [
        "model_state_dict",
        "state_dict",
        "model"
    ]:
        if (
            key in checkpoint
            and isinstance(checkpoint[key], dict)
        ):
            return checkpoint[key]

    if checkpoint and all(
        torch.is_tensor(value)
        for value in checkpoint.values()
    ):
        return checkpoint

    raise ValueError(
        "No model state dictionary was found."
    )


best_checkpoint = load_checkpoint(best_path)
final_checkpoint = load_checkpoint(final_path)

best_state = extract_state_dict(best_checkpoint)
final_state = extract_state_dict(final_checkpoint)

print("Best tensors:", len(best_state))
print("Final tensors:", len(final_state))
```

```python
best_keys = set(best_state.keys())
final_keys = set(final_state.keys())

print("Keys only in best:", best_keys - final_keys)
print("Keys only in final:", final_keys - best_keys)

common_keys = sorted(
    best_keys.intersection(final_keys)
)

different_parameters = []
maximum_parameter_difference = 0.0

for key in common_keys:
    best_tensor = best_state[key]
    final_tensor = final_state[key]

    if best_tensor.shape != final_tensor.shape:
        different_parameters.append(
            (key, "different shape")
        )
        continue

    difference = (
        best_tensor.float()
        - final_tensor.float()
    ).abs().max().item()

    maximum_parameter_difference = max(
        maximum_parameter_difference,
        difference
    )

    if difference > 0:
        different_parameters.append(
            (key, difference)
        )

print(
    "Number of different parameters:",
    len(different_parameters)
)

print(
    "Maximum parameter difference:",
    maximum_parameter_difference
)

print(
    "Checkpoints are identical:",
    len(different_parameters) == 0
)
```

```python
def calculate_sha256(path, chunk_size=1024 * 1024):
    digest = hashlib.sha256()

    with open(path, "rb") as file:
        for chunk in iter(
            lambda: file.read(chunk_size),
            b""
        ):
            digest.update(chunk)

    return digest.hexdigest()


checkpoint_sha256 = calculate_sha256(
    MODEL_CHECKPOINT
)

print("Checkpoint SHA-256:")
print(checkpoint_sha256)
```

```python
def extract_logits(saved_object):
    if torch.is_tensor(saved_object):
        return saved_object

    if isinstance(saved_object, dict):
        preferred_keys = [
            "logits",
            "test_logits",
            "outputs",
            "model_outputs"
        ]

        for key in preferred_keys:
            if key in saved_object:
                value = saved_object[key]

                if torch.is_tensor(value):
                    return value

                return torch.tensor(value)

    if isinstance(saved_object, (list, tuple)):
        tensor_candidates = [
            item for item in saved_object
            if torch.is_tensor(item)
            and item.ndim == 2
            and item.shape[1] == 2
        ]

        if len(tensor_candidates) == 1:
            return tensor_candidates[0]

        if len(tensor_candidates) > 1:
            return torch.cat(
                tensor_candidates,
                dim=0
            )

    raise ValueError(
        "A two-class logits tensor could not be identified."
    )


test_logits = extract_logits(saved_object)

test_logits = (
    test_logits
    .detach()
    .cpu()
    .float()
)

print("Recovered logits shape:", test_logits.shape)
```

```python
import numpy as np
import pandas as pd

test_predictions["calibrated_probability"] = pd.to_numeric(
    test_predictions["calibrated_probability"],
    errors="raise"
)

probability = test_predictions[
    "calibrated_probability"
]

test_predictions["decision_certainty"] = np.where(
    probability >= LOCKED_THRESHOLD,
    (
        probability - LOCKED_THRESHOLD
    )
    / (
        1 - LOCKED_THRESHOLD
    ),
    (
        LOCKED_THRESHOLD - probability
    )
    / LOCKED_THRESHOLD
)

test_predictions["decision_certainty"] = (
    test_predictions["decision_certainty"]
    .clip(0, 1)
)
```

```python
test_predictions["certainty_band"] = pd.Series(
    index=test_predictions.index,
    dtype="object"
)

for outcome in ["TP", "TN", "FP", "FN"]:

    outcome_index = test_predictions.index[
        test_predictions["outcome"] == outcome
    ]

    certainty_values = test_predictions.loc[
        outcome_index,
        "decision_certainty"
    ]

    ranked_values = certainty_values.rank(
        method="first"
    )

    test_predictions.loc[
        outcome_index,
        "certainty_band"
    ] = pd.qcut(
        ranked_values,
        q=3,
        labels=[
            "Low",
            "Medium",
            "High"
        ]
    ).astype(str)
```

```python
TARGET_PER_BAND = 10
TARGET_PER_OUTCOME = 30
SELECTION_SEED = 42
```

```python
def select_outcome_cases(
    dataframe,
    outcome,
    seed
):
    selected_parts = []

    for band_index, band in enumerate(
        ["Low", "Medium", "High"]
    ):
        available_cases = dataframe[
            (dataframe["outcome"] == outcome)
            & (
                dataframe["certainty_band"]
                == band
            )
        ]

        if len(available_cases) < TARGET_PER_BAND:
            raise ValueError(
                f"{outcome}-{band} contains only "
                f"{len(available_cases)} cases."
            )

        selected_band = available_cases.sample(
            n=TARGET_PER_BAND,
            random_state=seed + band_index
        )

        selected_parts.append(
            selected_band
        )

    return pd.concat(
        selected_parts,
        axis=0
    )
```

```python
selected_groups = []

for outcome_index, outcome in enumerate(
    ["TP", "TN", "FP", "FN"]
):
    selected_group = select_outcome_cases(
        dataframe=test_predictions,
        outcome=outcome,
        seed=(
            SELECTION_SEED
            + outcome_index * 10
        )
    )

    selected_groups.append(
        selected_group
    )

xai_cohort = (
    pd.concat(
        selected_groups,
        axis=0
    )
    .reset_index(drop=True)
)
```
